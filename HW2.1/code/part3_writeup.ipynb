# Part III: Expected Value and Bankroll Allocation

Suppose the player has a fixed bankroll

$$
B.
$$

The player chooses

$$
b_R
$$

dollars to bet on roulette and

$$
b_B
$$

dollars to bet on blackjack.

The constraints are

$$
b_R\ge0,
$$

$$
b_B\ge0,
$$

and

$$
b_R+b_B\le B.
$$

The inequality is important because the player is allowed to keep some or all of the bankroll instead of gambling it.

Let

$$
e_R=E[X_R]
$$

be the expected payoff per dollar bet on roulette and

$$
e_B=E[X_B]
$$

be the expected payoff per dollar bet on blackjack.

Then the expected total profit is

$$
E[\Pi]
=
b_Re_R+b_Be_B.
$$

Therefore, the optimization problem is

$$
\max_{b_R,b_B\ge0,\ b_R+b_B\le B}
\left(
b_Re_R+b_Be_B
\right).
$$

Because this objective is linear, the solution depends on the signs and relative values of $e_R$ and $e_B$.

# Case 1: $e_R<0$ and $e_B<0$

Suppose both games have negative expected value.

Then

$$
b_Re_R\le0
$$

and

$$
b_Be_B\le0.
$$

Every positive amount placed into either game decreases expected profit.

The player is allowed to choose

$$
b_R=0
$$

and

$$
b_B=0.
$$

This gives

$$
E[\Pi]=0.
$$

Any positive bet gives negative expected profit, so the optimum is

$$
b_R^*=0,
\qquad
b_B^*=0.
$$

Therefore,

$$
\boxed{\text{if both games have negative expected value, do not bet.}}
$$

# Case 2: $e_R<e_B<0$

In this case blackjack has a better expected value than roulette because

$$
e_B>e_R.
$$

However, both values are still negative.

If the player were required to gamble the entire bankroll, blackjack would be the better choice.

But the constraint is

$$
b_R+b_B\le B,
$$

so not betting is allowed.

Keeping one dollar has expected return

$$
0,
$$

while betting it on either game gives a negative expected return.

Therefore,

$$
0>e_B>e_R.
$$

The optimal solution is still

$$
b_R^*=0,
\qquad
b_B^*=0.
$$

This demonstrates that a game being less bad than another game does not mean that it is worth playing.

# Case 3: $e_R<0<e_B$

Suppose roulette has negative expected value but blackjack has positive expected value.

Since

$$
e_R<0,
$$

no money should be placed into roulette:

$$
b_R^*=0.
$$

Since

$$
e_B>0,
$$

each additional dollar placed into blackjack increases expected profit.

Therefore the full bankroll should be allocated to blackjack:

$$
b_B^*=B.
$$

Thus,

$$
\boxed{
b_R^*=0,
\qquad
b_B^*=B
}
$$

and the maximum expected profit is

$$
E[\Pi]^*=Be_B.
$$

This conclusion follows from maximizing expected value only. It does not account for risk aversion, variance, or the possibility of losing the bankroll.

# General Decision Rule

The three possible returns for each dollar are

$$
0,
\qquad
e_R,
\qquad
e_B.
$$

The value $0$ corresponds to keeping the money.

Therefore, the player should compare

$$
\max\{0,e_R,e_B\}.
$$

If both games have negative expectation, keep the bankroll.

If roulette has the largest positive expectation, allocate the bankroll to roulette.

If blackjack has the largest positive expectation, allocate the bankroll to blackjack.

# Conclusion

The expected total profit is

$$
E[\Pi]
=
b_Re_R+b_Be_B.
$$

For the models used in this homework,

$$
E[\Pi]
=
-0.0270270b_R
-
0.0800918b_B.
$$

Both coefficients are negative.

Therefore, every positive allocation to roulette or blackjack reduces expected profit.

The optimal solution is

$$
\boxed{
b_R^*=0,
\qquad
b_B^*=0
}
$$

and

$$
\boxed{
E[\Pi]^*=0.
}
$$

The standalone Monte Carlo experiment independently reached the same conclusion.

The main lesson is that choosing the better of two negative expected-value games is not necessarily the optimal decision. Because the player is allowed to keep the bankroll, not gambling dominates both games when both expected returns are negative.
