# Part I: European Roulette

European roulette has sample space

$$
\Omega_R=\{0,1,2,\ldots,36\}.
$$

Each of the 37 outcomes is equally likely, so

$$
P(\{\omega\})=\frac{1}{37}.
$$

Because the sample space is finite, the natural sigma-algebra is

$$
\mathcal F_R=2^{\Omega_R}.
$$

Thus every subset of $\Omega_R$ is an event.

The event that the ball lands on red is

$$
A_{\mathrm{red}}
=
\{1,3,5,7,9,12,14,16,18,19,21,23,25,27,30,32,34,36\}.
$$

There are 18 red numbers. The remaining 19 outcomes consist of 18 black numbers and zero.

Therefore,

$$
P(A_{\mathrm{red}})
=
\frac{18}{37}
\approx
0.486486
$$

and

$$
P(A_{\mathrm{red}}^c)
=
\frac{19}{37}
\approx
0.513514.
$$

# Roulette Payoff Random Variable

For a one-unit bet on red, define

$$
X_R(\omega)
=
\begin{cases}
1, & \omega\in A_{\mathrm{red}},\\
-1, & \omega\notin A_{\mathrm{red}}.
\end{cases}
$$

Thus,

$$
P(X_R=1)=\frac{18}{37}
$$

and

$$
P(X_R=-1)=\frac{19}{37}.
$$

# Expected Value

The expected payoff is

$$
E[X_R]
=
1\left(\frac{18}{37}\right)
+
(-1)\left(\frac{19}{37}\right).
$$

Therefore,

$$
E[X_R]
=
-\frac{1}{37}
\approx
-0.027027.
$$

Thus every one dollar bet on red loses approximately

$$
0.027
$$

dollars on average.

Equivalently, the expected loss is approximately 2.70 cents per dollar bet.

# Variance

Since

$$
X_R^2=1
$$

for both possible payoffs,

$$
E[X_R^2]=1.
$$

Therefore,

$$
\mathrm{Var}(X_R)
=
E[X_R^2]
-
(E[X_R])^2.
$$

Hence,

$$
\mathrm{Var}(X_R)
=
1-
\left(-\frac{1}{37}\right)^2
=
\frac{1368}{1369}.
$$

Numerically,

$$
\mathrm{Var}(X_R)
\approx
0.999270.
$$

# Monte Carlo Verification of the Probability Model

I simulated

$$
1000000
$$

independent European roulette spins.

The theoretical expected payoff per unit bet was

$$
E[X_R]
=
-0.0270270.
$$

The simulated mean payoff was

$$
-0.028258.
$$

The difference between the simulated and theoretical means was

$$
-0.001231.
$$

The Monte Carlo standard error was

$$
0.0009996.
$$

Therefore, the difference measured in standard-error units was

$$
z
=
\frac{-0.001231}{0.0009996}
\approx
-1.23.
$$

A difference of about 1.23 standard errors is consistent with ordinary Monte Carlo sampling variation.

The theoretical variance was

$$
0.999270,
$$

while the simulated variance was

$$
0.999201.
$$

These values are also extremely close.

Therefore, the simulation supports both the theoretical expected value and variance.

# How Much Should Be Bet on Roulette?

Suppose the player has bankroll $B$ and chooses to bet

$$
b_R
$$

dollars on roulette, where

$$
0\le b_R\le B.
$$

The total roulette profit is

$$
\Pi_R=b_RX_R.
$$

Therefore,

$$
E[\Pi_R]
=
b_RE[X_R].
$$

Since

$$
E[X_R]
=
-\frac{1}{37},
$$

we obtain

$$
E[\Pi_R]
=
-\frac{b_R}{37}.
$$

The expected profit decreases linearly as the roulette bet increases.

Therefore, the optimization problem

$$
\max_{0\le b_R\le B}
E[\Pi_R]
$$

has solution

$$
b_R^*=0.
$$

Thus, if the objective is to maximize expected profit, the optimal decision is to bet nothing on roulette.

# Numerical Verification of the Betting Decision

To verify this result numerically, I used a bankroll of

$$
B=100
$$

and tested five possible roulette allocations.

| Roulette Bet | Theoretical Expected Profit | Simulated Mean Profit |
| ---: | ---: | ---: |
| $0$ | $0.0000$ | $0.0000$ |
| $25$ | $-0.6757$ | $-0.7065$ |
| $50$ | $-1.3514$ | $-1.4129$ |
| $75$ | $-2.0270$ | $-2.1194$ |
| $100$ | $-2.7027$ | $-2.8258$ |

![Expected profit versus roulette bet size](roulette_bet_strategy_verification.png)

Both the theoretical calculation and the simulation show the same pattern.

As the amount bet on roulette increases, average profit decreases.

For example, theoretically,

$$
E[\Pi_R\mid b_R=25]
=
-\frac{25}{37}
\approx
-0.676,
$$

while betting the entire $100$ bankroll gives

$$
E[\Pi_R\mid b_R=100]
=
-\frac{100}{37}
\approx
-2.703.
$$

In contrast, betting nothing gives

$$
E[\Pi_R\mid b_R=0]=0.
$$

Therefore,

$$
0
>
-\frac{25}{37}
>
-\frac{50}{37}
>
-\frac{75}{37}
>
-\frac{100}{37}.
$$

The simulation produced the same ordering.

# LLM Prediction and Verification

I asked the LLM to determine how much should be bet on roulette if the objective is to maximize expected profit.

The LLM predicted

$$
b_R^*=0
$$

because the expected payoff per dollar is negative:

$$
E[X_R]
=
-\frac{1}{37}
<
0.
$$

This prediction was verified analytically because

$$
E[\Pi_R]
=
-\frac{b_R}{37}
$$

is a decreasing function of $b_R$.

It was also verified numerically using one million simulated roulette outcomes. Every positive bet size produced a negative average profit, while betting zero produced exactly zero profit.

Therefore, the LLM's recommendation was correct.

# Conclusion

European roulette is a negative expected-value game for a bet on red.

The theoretical results are

$$
E[X_R]
=
-\frac{1}{37}
\approx
-0.027027
$$

and

$$
\mathrm{Var}(X_R)
=
\frac{1368}{1369}
\approx
0.999270.
$$

The Monte Carlo simulation closely reproduced both quantities.

More importantly, because

$$
E[\Pi_R]
=
-\frac{b_R}{37},
$$

any positive roulette bet reduces expected wealth.

Therefore, the expected-value maximizing strategy is

$$
\boxed{b_R^*=0}.
$$

The simulation confirmed this decision: increasing the amount bet produced increasingly negative average profits.

An individual roulette bet can still win, but when the objective is maximizing expected profit, the mathematically optimal decision is not to play.
