# Part II: Blackjack

Blackjack is more complicated than roulette because it is a sequential game.

Cards are drawn without replacement, so the probability of later cards depends on which cards have already been removed from the deck. The player also makes decisions during the hand.

To make the problem completely defined, I used the simplified blackjack rules from the homework together with a fixed player strategy.

# Probability Space

The sample space is

$$
\Omega_B
=
\{\text{all permutations of a standard 52-card deck}\}.
$$

There are

$$
52!
$$

possible ordered decks.

Assuming a uniformly shuffled deck, every ordered deck has probability

$$
P(\{\omega\})
=
\frac{1}{52!}.
$$

Since the sample space is finite, the natural sigma-algebra is

$$
\mathcal F_B
=
2^{\Omega_B}.
$$

Once the deck order and the fixed player strategy are known, the entire hand is determined.

# Simplified Blackjack Rules

The following rules were used:

- one standard 52-card deck;
- no splitting;
- no doubling;
- no surrender;
- player hits while the hand total is less than 17;
- player stands on 17 or greater;
- dealer hits while the hand total is less than 17;
- dealer stands on 17 or greater;
- an ace counts as 11 unless this would cause the hand to bust, in which case it counts as 1;
- a win gives payoff $+1$;
- a loss gives payoff $-1$;
- a push gives payoff $0$.

The homework does not specify a special payout for a natural blackjack. Therefore, in this simplified model, an initial blackjack is treated as an ordinary total of 21 and does not receive a special $3:2$ payout.

The model also treats the dealer as standing on any total of 17, including soft 17.

These assumptions are important because changing the blackjack rules can change the expected value.

# Fixed Player Strategy

The player's fixed strategy $S$ is

$$
S=
\begin{cases}
\text{hit}, & \text{if total}<17,\\
\text{stand}, & \text{if total}\ge17.
\end{cases}
$$

This is intentionally a simple strategy.

It is not claimed to be optimal blackjack strategy.

# Payoff Random Variable

Define

$$
X_B:\Omega_B\rightarrow\mathbb R
$$

by

$$
X_B(\omega)
=
\begin{cases}
1, & \text{if the player wins},\\
0, & \text{if the result is a push},\\
-1, & \text{if the player loses}.
\end{cases}
$$

The ordered deck and strategy determine the value of $X_B$.

# Why Blackjack Is More Complicated Than Roulette

Roulette can be represented as

$$
\text{spin}
\rightarrow
\text{outcome}
\rightarrow
\text{payoff}.
$$

Blackjack instead has a sequence such as

$$
\text{shuffle}
\rightarrow
\text{initial cards}
\rightarrow
\text{player decisions}
\rightarrow
\text{additional cards}
\rightarrow
\text{dealer decisions}
\rightarrow
\text{payoff}.
$$

The draws are also not independent.

For example, if an ace is removed from the deck, the probability of drawing an ace on the next draw decreases.

Therefore,

$$
P(\text{next card}\mid\text{previous cards})
$$

changes throughout the hand.

# Recursive Probability Calculation

Directly enumerating all

$$
52!
$$

ordered decks would be impractical.

Instead, I computed the probabilities recursively.

If $n$ cards remain in the deck and $n_j$ of them are of card type $j$, then

$$
P(\text{next card is type }j)
=
\frac{n_j}{n}.
$$

The recursion considers every possible next card and continues until the hand reaches a win, loss, or push.

For the fixed strategy used here, the theoretical probabilities were

$$
P(\text{win})
=
0.4088236578,
$$

$$
P(\text{loss})
=
0.4889154379,
$$

and

$$
P(\text{push})
=
0.1022609043.
$$

As a verification,

$$
P(\text{win})
+
P(\text{loss})
+
P(\text{push})
=
1.
$$

Numerically,

$$
0.4088236578
+
0.4889154379
+
0.1022609043
=
1.
$$

# Expected Value

The expected payoff is

$$
E[X_B]
=
1P(\text{win})
+
0P(\text{push})
-
1P(\text{loss}).
$$

Therefore,

$$
E[X_B]
=
P(\text{win})
-
P(\text{loss}).
$$

Substituting the calculated probabilities gives

$$
E[X_B]
=
0.4088236578
-
0.4889154379.
$$

Hence,

$$
E[X_B]
=
-0.0800917801.
$$

Therefore, under this simplified strategy, each dollar bet has an expected loss of approximately

$$
0.08009
$$

dollars.

This corresponds to an expected loss of approximately

$$
8.01\%
$$

of the amount bet.

# Variance

Since

$$
X_B^2
=
\begin{cases}
1, & \text{for a win or loss},\\
0, & \text{for a push},
\end{cases}
$$

we have

$$
E[X_B^2]
=
P(\text{win})
+
P(\text{loss}).
$$

Therefore,

$$
E[X_B^2]
=
0.4088236578
+
0.4889154379
=
0.8977390957.
$$

The variance is

$$
\mathrm{Var}(X_B)
=
E[X_B^2]
-
(E[X_B])^2.
$$

Thus,

$$
\mathrm{Var}(X_B)
\approx
0.8913244025.
$$

# How Much Should Be Bet on Blackjack?

Suppose the player's bankroll is $B$ and the player chooses to bet

$$
b_B
$$

dollars on blackjack, where

$$
0\le b_B\le B.
$$

The total blackjack profit is

$$
\Pi_B
=
b_BX_B.
$$

Therefore,

$$
E[\Pi_B]
=
b_BE[X_B].
$$

Since

$$
E[X_B]
=
-0.0800917801,
$$

we obtain

$$
E[\Pi_B]
=
-0.0800917801b_B.
$$

Because the coefficient is negative, expected profit decreases as the amount bet increases.

Therefore, the optimization problem

$$
\max_{0\le b_B\le B}E[\Pi_B]
$$

has solution

$$
b_B^*=0.
$$

Thus, if the goal is to maximize expected profit, the optimal strategy under this blackjack model is

$$
\boxed{\text{Do not bet on blackjack.}}
$$

# Important LLM Simplifications

Although the LLM gave the correct result for the specified model, several assumptions had to be checked carefully.

First, a real blackjack game often pays a natural blackjack differently from an ordinary win. The simplified model used here treats all wins as $+1$.

Second, the exact rule for soft 17 must be specified. This model assumes that the dealer stands whenever the total is at least 17.

Third, the LLM must not treat successive card draws as independent.

Cards are drawn without replacement, so the probability distribution changes throughout the hand.

The recursive calculation and Monte Carlo simulation correctly account for this dependence.

These examples show why an LLM answer must be checked against the exact assumptions of the probability model.
