# Homework 2.1: Casino Probability, Sigma-Algebras, and Bad Decisions

## Introduction

This homework uses roulette and blackjack to study probability spaces, sigma-algebras, events, random variables, expectation, variance, and decision-making under uncertainty.

The main goal is not to determine how to gamble successfully. Instead, the casino provides a simple setting in which mathematical probability models can be constructed and checked.

The analysis has four main components:

1. construct a probability space for European roulette;
2. construct a probability space for a simplified blackjack model;
3. use expected value to decide how a fixed bankroll should be allocated;
4. formulate and verify two additional probability questions.

An LLM was used to help formulate and analyze the models, but its predictions were checked using mathematical derivations and Monte Carlo simulations.

---

# Part I: European Roulette

## Probability Space

For European roulette, the sample space is

$$
\Omega_R
=
\{0,1,2,\ldots,36\}.
$$

There are 37 possible outcomes.

Each outcome is equally likely, so

$$
P(\{\omega\})
=
\frac{1}{37},
\qquad
\omega\in\Omega_R.
$$

Since the sample space is finite, the natural sigma-algebra is the full power set

$$
\mathcal F_R
=
2^{\Omega_R}.
$$

Therefore, every subset of roulette outcomes is an event.

---

## Explicitly Defined Roulette Events

The event that the roulette result is red is

$$
A_{\mathrm{red}}
=
\{1,3,5,7,9,12,14,16,18,19,21,23,25,27,30,32,34,36\}.
$$

There are 18 red numbers, so

$$
P(A_{\mathrm{red}})
=
\frac{18}{37}.
$$

The event that the roulette result is even is

$$
A_{\mathrm{even}}
=
\{2,4,6,\ldots,36\}.
$$

There are also 18 even outcomes, so

$$
P(A_{\mathrm{even}})
=
\frac{18}{37}.
$$

The event that the ball lands on zero is

$$
A_{\mathrm{zero}}
=
\{0\}.
$$

Therefore,

$$
P(A_{\mathrm{zero}})
=
\frac{1}{37}.
$$

These provide three explicitly defined measurable events in the roulette probability space.

---

# Roulette Payoff Random Variable

For a one-unit bet on red, define

$$
X_R(\omega)
=
\begin{cases}
1, & \omega\in A_{\mathrm{red}},\\
-1, & \omega\notin A_{\mathrm{red}}.
\end{cases}
$$

Thus,

$$
P(X_R=1)
=
\frac{18}{37}
$$

and

$$
P(X_R=-1)
=
\frac{19}{37}.
$$

The 19 losing outcomes are the 18 black numbers together with zero.

---

## Expected Value

The expected payoff is

$$
E[X_R]
=
1\left(\frac{18}{37}\right)
+
(-1)\left(\frac{19}{37}\right).
$$

Therefore,

$$
E[X_R]
=
-\frac{1}{37}.
$$

Numerically,

$$
E[X_R]
\approx
-0.027027.
$$

Thus a one-dollar red bet loses approximately 2.70 cents on average.

---

## Variance

Since

$$
X_R^2=1
$$

for every outcome,

$$
E[X_R^2]=1.
$$

Therefore,

$$
\mathrm{Var}(X_R)
=
E[X_R^2]
-
(E[X_R])^2.
$$

Hence,

$$
\mathrm{Var}(X_R)
=
1-
\left(-\frac{1}{37}\right)^2
=
\frac{1368}{1369}.
$$

Numerically,

$$
\mathrm{Var}(X_R)
\approx
0.999270.
$$

---

# Roulette Bankroll Decision

Suppose the player has bankroll $B$ and chooses to bet $b_R$ dollars on roulette.

Then

$$
0\le b_R\le B.
$$

The total roulette profit is

$$
\Pi_R=b_RX_R.
$$

Therefore,

$$
E[\Pi_R]
=
b_RE[X_R].
$$

Since

$$
E[X_R]
=
-\frac{1}{37},
$$

we obtain

$$
E[\Pi_R]
=
-\frac{b_R}{37}.
$$

This decreases as $b_R$ increases.

Therefore, the expected-value maximizing decision is

$$
b_R^*=0.
$$

For a bankroll of

$$
B=100,
$$

the theoretical and simulated results were:

| Roulette Bet | Theoretical Expected Profit | Simulated Mean Profit |
| ---: | ---: | ---: |
| $0$ | $0.0000$ | $0.0000$ |
| $25$ | $-0.6757$ | $-0.7065$ |
| $50$ | $-1.3514$ | $-1.4129$ |
| $75$ | $-2.0270$ | $-2.1194$ |
| $100$ | $-2.7027$ | $-2.8258$ |

![Expected profit versus roulette bet size](roulette_bet_strategy_verification.png)

Both theory and simulation show that increasing the roulette bet decreases expected profit.

Therefore, the optimal roulette decision is to bet nothing.

# Part II: Simplified Blackjack

## Probability Space

For blackjack, the sample space is

$$
\Omega_B
=
\{\text{all permutations of a standard 52-card deck}\}.
$$

The cards can be regarded as 52 distinct physical cards, so

$$
|\Omega_B|=52!.
$$

Assuming a perfectly shuffled deck,

$$
P(\{\omega\})
=
\frac{1}{52!}.
$$

The natural sigma-algebra is

$$
\mathcal F_B
=
2^{\Omega_B}.
$$

Once the deck ordering and player strategy are fixed, the complete hand is determined.

---

# Simplified Blackjack Rules

The model used in this report has the following rules:

- one standard 52-card deck;
- no splitting;
- no doubling;
- no surrender;
- player hits below 17;
- player stands on 17 or greater;
- dealer hits below 17;
- dealer stands on 17 or greater;
- an ace counts as 11 unless this would cause a bust, in which case it counts as 1;
- win gives payoff $+1$;
- loss gives payoff $-1$;
- push gives payoff $0$.

The model treats all totals of 17 as standing totals, including soft 17.

The homework does not specify a special payout for a natural blackjack. Therefore, a natural blackjack is treated as an ordinary total of 21 rather than receiving a special $3:2$ payoff.

This is a modeling assumption and must be stated explicitly.

---

# Fixed Player Strategy

The fixed strategy is

$$
S=
\begin{cases}
\text{hit}, & \text{total}<17,\\
\text{stand}, & \text{total}\ge17.
\end{cases}
$$

This strategy was chosen for simplicity and is not claimed to be optimal blackjack strategy.

---

# Blackjack Events

Define the blackjack win event

$$
A_{\mathrm{win}}
=
\{\omega\in\Omega_B:
\text{the player wins under strategy }S\}.
$$

Similarly, define

$$
A_{\mathrm{loss}}
=
\{\omega\in\Omega_B:
\text{the player loses}\}
$$

and

$$
A_{\mathrm{push}}
=
\{\omega\in\Omega_B:
\text{the final result is a tie}\}.
$$

The exact recursive calculation gave

$$
P(A_{\mathrm{win}})
=
0.4088236578,
$$

$$
P(A_{\mathrm{loss}})
=
0.4889154379,
$$

and

$$
P(A_{\mathrm{push}})
=
0.1022609043.
$$

These probabilities satisfy

$$
P(A_{\mathrm{win}})
+
P(A_{\mathrm{loss}})
+
P(A_{\mathrm{push}})
=
1.
$$

---

# Blackjack Payoff Random Variable

Define

$$
X_B:\Omega_B\rightarrow\mathbb R
$$

by

$$
X_B(\omega)
=
\begin{cases}
1, & \omega\in A_{\mathrm{win}},\\
0, & \omega\in A_{\mathrm{push}},\\
-1, & \omega\in A_{\mathrm{loss}}.
\end{cases}
$$

This is the second main payoff random variable used in the report.

---

## Expected Value

The expected payoff is

$$
E[X_B]
=
1P(A_{\mathrm{win}})
-
1P(A_{\mathrm{loss}}).
$$

Therefore,

$$
E[X_B]
=
0.4088236578
-
0.4889154379.
$$

Hence,

$$
E[X_B]
=
-0.0800917801.
$$

Thus this simplified blackjack strategy loses approximately

$$
8.01\%
$$

of the amount bet on average.

---

## Variance

Since

$$
X_B^2
=
\begin{cases}
1, & \text{win or loss},\\
0, & \text{push},
\end{cases}
$$

we have

$$
E[X_B^2]
=
P(A_{\mathrm{win}})
+
P(A_{\mathrm{loss}}).
$$

Thus,

$$
E[X_B^2]
=
0.8977390957.
$$

Therefore,

$$
\mathrm{Var}(X_B)
=
E[X_B^2]
-
(E[X_B])^2.
$$

Numerically,

$$
\mathrm{Var}(X_B)
\approx
0.8913244025.
$$

# Blackjack Bankroll Decision

Then

$$
0\le b_B\le B.
$$

The blackjack profit is

$$
\Pi_B=b_BX_B.
$$

Therefore,

$$
E[\Pi_B]
=
b_BE[X_B].

$$

# Part III: Expected Value and Bankroll Allocation

The player now chooses both

$$
b_R\ge0
$$

and

$$
b_B\ge0
$$

subject to

$$
b_R+b_B\le B.
$$

The expected total profit is

$$
E[\Pi]
=
b_Re_R+b_Be_B.
$$

The optimization problem is

$$
\max_{b_R,b_B\ge0,\ b_R+b_B\le B}
\left(
b_Re_R+b_Be_B
\right).
$$

---

## Case 1: $e_R<0$ and $e_B<0$

Both games lose money on average.

Any positive $b_R$ or $b_B$ reduces expected profit.

Since the player may choose not to bet,

$$
b_R^*=0,
\qquad
b_B^*=0.
$$

---

## Case 2: $e_R<e_B<0$

Blackjack is better than roulette in this hypothetical case, but both expected values remain negative.

Because not betting has expected return zero,

$$
0>e_B>e_R.
$$

Therefore the optimum is still

$$
b_R^*=0,
\qquad
b_B^*=0.
$$

Being less bad than another game does not make a game a good expected-value decision.

---

## Case 3: $e_R<0<e_B$

Roulette has negative expectation while blackjack has positive expectation.

Therefore no money should be placed into roulette:

$$
b_R^*=0.
$$

Since each dollar placed into blackjack increases expected profit,

$$
b_B^*=B.
$$

Thus,

$$
b_R^*=0,
\qquad
b_B^*=B.
$$

# Part IV: Two Additional Probability Questions

## Question 1: Probability of Losing Five Roulette Bets in a Row

Suppose a player bets on red for five independent roulette spins.

For five spins, the sample space is

$$
\Omega_5
=
\Omega_R^5.
$$

Define

$$
L_5
=
\{\text{all five red bets lose}\}.
$$

Since a red bet loses with probability

$$
\frac{19}{37},
$$

independence gives

$$
P(L_5)
=
\left(\frac{19}{37}\right)^5.
$$

Numerically,

$$
P(L_5)
=
0.0357074950.
$$

Thus,

$$
P(L_5)
\approx
3.57075\%.
$$

A Monte Carlo simulation of

$$
500000
$$

five-spin experiments gave

$$
P_{\mathrm{sim}}(L_5)
=
0.035462.
$$

The difference from theory was approximately

$$
-0.94
$$

Monte Carlo standard errors.

Therefore, the simulation agrees with the theoretical result.

---

## Question 2: Expected Profit After 100 Roulette Bets

Let

$$
X_i
$$

be the payoff from the $i$-th one-unit red bet.

Define total profit

$$
S_{100}
=
X_1+\cdots+X_{100}.
$$

Since

$$
E[X_i]
=
-\frac{1}{37},
$$

linearity of expectation gives

$$
E[S_{100}]
=
100\left(-\frac{1}{37}\right).
$$

Therefore,

$$
E[S_{100}]
=
-\frac{100}{37}
\approx
-2.70270.
$$

The variance is

$$
\mathrm{Var}(S_{100})
=
100\mathrm{Var}(X_R).
$$

Therefore,

$$
\mathrm{Var}(S_{100})
=
100\left(\frac{1368}{1369}\right)
\approx
99.92695.
$$

A simulation of

$$
200000
$$

independent 100-spin sessions produced mean profit

$$
-2.67761
$$

and variance

$$
99.68430.
$$

The simulated mean was approximately

$$
1.12
$$

Monte Carlo standard errors from the theoretical value.

Therefore, the simulation supports both the expected value and variance calculations.

# Discussion of the LLM

The LLM was useful for organizing the probability models and suggesting ways to verify the calculations.

For roulette, it correctly identified the finite sample space

$$
\Omega_R=\{0,\ldots,36\}
$$

and the power-set sigma-algebra

$$
\mathcal F_R=2^{\Omega_R}.
$$

It also correctly calculated

$$
E[X_R]
=
-\frac{1}{37}
$$

and suggested using Monte Carlo simulation as an independent check.

For blackjack, the LLM was useful in recognizing that an ordered deck can serve as the underlying sample point. Once the deck order and strategy are fixed, the hand becomes deterministic.

The LLM was also useful in suggesting a recursive calculation rather than attempting to explicitly enumerate all

$$
52!
$$

possible ordered decks.

However, the LLM could be mathematically careless when details of the model were not specified.

For example, the original blackjack problem did not completely specify:

- whether a natural blackjack pays $3:2$;
- whether the dealer hits or stands on soft 17;
- how pushes are treated;
- how aces are valued;
- the exact player strategy.

A numerical expected value cannot be meaningfully calculated until these assumptions are fixed.

For this report I explicitly chose a simplified model in which all wins pay $+1$, pushes pay $0$, losses pay $-1$, and both the player and dealer stand on every total of 17 or greater.

Another possible error would be treating successive blackjack card draws as independent. This would be incorrect because cards are drawn without replacement.

The recursive calculation and simulation used here update the remaining deck after every card, so this dependence is correctly represented.

The bankroll problem also illustrates why an LLM answer must be interpreted carefully. Saying that one game is better than another does not imply that it should be played.

For example, in my model,

$$
e_R
\approx
-0.0270
$$

is better than

$$
e_B
\approx
-0.0801.
$$

However,

$$
0>e_R>e_B,
$$

so keeping the bankroll is better than either game.

The LLM correctly identified this once the constraint

$$
b_R+b_B\le B
$$

was taken into account.

Overall, the LLM was useful as a tool for constructing models, proposing calculations, writing simulation code, and checking reasoning. However, its answers still required mathematical verification because small unstated assumptions can substantially change a probability model.

(\Omega_R,\mathcal F_R,P),

where

$$
\Omega_R=\{0,\ldots,36\}
$$

and

$$
\mathcal F_R=2^{\Omega_R}.
$$

For a one-unit red bet,

$$
E[X_R]
=
-\frac{1}{37}
\approx
-0.027027
$$

and

$$
\mathrm{Var}(X_R)
=
\frac{1368}{1369}
\approx

$$
E[X_B]
\approx
-0.0800918
$$

and

$$
\mathrm{Var}(X_B)
\approx
0.8913244.

Both expected values are negative.

Therefore, for the actual models used in this homework,

$$
E[\Pi]
=
-0.027027b_R
-
0.0800918b_B.
$$

The expected-value maximizing bankroll allocation is

$$
b_R^*=0,
\qquad
b_B^*=0.
$$

Thus the entire bankroll should be kept rather than bet.

The analytical calculations, recursive blackjack computation, grid-search optimization, and Monte Carlo simulations all support the same conclusion.

The homework also demonstrates an important lesson about using LLMs in mathematics: an LLM can be very useful for generating models and calculations, but its assumptions and conclusions should always be independently checked.
