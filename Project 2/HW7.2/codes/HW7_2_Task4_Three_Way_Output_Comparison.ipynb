## Compare Unconditioned, Conditioned, and Direct LLM Outputs

This notebook compares three outputs for each fixed Marcus theme:

1. unconditioned Odyssey-style statistical generation;
2. Marcus-theme-conditioned Odyssey-style statistical generation;
3. direct LLM-generated output.

The two statistical examples are selected using the fixed display rule from Task 1.

The LLM receives only the theme label, not the full Marcus passage. This keeps the information supplied to the LLM closer to the information supplied to the statistical conditioned generator.

## Direct LLM prompt

The following prompt was used separately for each theme:

> You are given only the theme: `{THEME}`. Write one original sentence of 15 to 30 words that evokes the vocabulary and prose register of Samuel Butler's public-domain English prose translation of the Odyssey. Preserve the ethical direction of the theme where possible. Do not quote or reproduce any specific sentence from the Odyssey or Meditations. Return one sentence only.

The five responses below were generated by ChatGPT during preparation of this homework notebook.

They are illustrative examples, not a matched statistical benchmark, because there is only one LLM sentence per theme while the statistical experiment contains hundreds of draws.

```python
from pathlib import Path
import json

import pandas as pd

cwd = Path.cwd()
PROJECT_ROOT = cwd.parent if cwd.name.lower() == "codes" else cwd

DATA_DIR = PROJECT_ROOT / "data" / "hw7_2"

display_examples = pd.read_csv(
    DATA_DIR
    / "task3_fixed_display_examples.csv"
)

passages = pd.read_csv(
    DATA_DIR
    / "task1_selected_marcus_passages.csv"
)

design = json.loads(
    (
        DATA_DIR
        / "task1_fixed_experiment_design.json"
    ).read_text(
        encoding="utf-8"
    )
)

THEMES = design[
    "themes"
]
```

```python
PROMPT_TEMPLATE = (
    "You are given only the theme: {theme}. "
    "Write one original sentence of 15 to 30 words that evokes the vocabulary "
    "and prose register of Samuel Butler's public-domain English prose translation "
    "of the Odyssey. Preserve the ethical direction of the theme where possible. "
    "Do not quote or reproduce any specific sentence from the Odyssey or Meditations. "
    "Return one sentence only."
)

DIRECT_LLM_RESPONSES = {
    "discipline": (
        "Let the man endure his labour without complaint, keeping his purpose firm "
        "though the road be long and the gods send hardship upon him."
    ),
    "mortality": (
        "No man, though rich in cattle and dear to the gods, may escape the day "
        "when death comes and his house is left behind."
    ),
    "anger": (
        "He checked the anger in his heart, for he knew that a rash word may bring "
        "more sorrow than the wrong that first provoked it."
    ),
    "duty": (
        "Then he rose and did the work laid upon him, thinking first of his house "
        "and companions rather than of ease or honour."
    ),
    "self-command": (
        "He mastered his heart and held his tongue, waiting until reason showed him "
        "when to speak and when to keep silence."
    ),
}

llm_rows = []

for theme in THEMES:
    llm_rows.append({
        "theme": theme,
        "prompt": (
            PROMPT_TEMPLATE.format(
                theme=theme
            )
        ),
        "response": (
            DIRECT_LLM_RESPONSES[
                theme
            ]
        ),
        "information_given": (
            "theme label only"
        ),
        "llm_source": (
            "ChatGPT response generated during notebook preparation"
        ),
    })


llm_outputs = pd.DataFrame(
    llm_rows
)

llm_path = (
    DATA_DIR
    / "task4_direct_llm_outputs.csv"
)

llm_outputs.to_csv(
    llm_path,
    index=False,
)

print(
    llm_outputs[
        [
            "theme",
            "information_given",
            "response",
        ]
    ].to_string(
        index=False
    )
)
```

## Build the three-way comparison table

```python
comparison_rows = []

for theme in THEMES:
    unconditioned = display_examples[
        (
            display_examples[
                "theme"
            ] == theme
        )
        & (
            display_examples[
                "mode"
            ] == "unconditioned"
        )
    ].iloc[0]

    conditioned = display_examples[
        (
            display_examples[
                "theme"
            ] == theme
        )
        & (
            display_examples[
                "mode"
            ] == "conditioned"
        )
    ].iloc[0]

    llm = llm_outputs[
        llm_outputs[
            "theme"
        ] == theme
    ].iloc[0]

    marcus = passages[
        passages[
            "theme"
        ] == theme
    ].iloc[0]

    comparison_rows.append({
        "theme": theme,
        "marcus_passage": (
            marcus[
                "passage"
            ]
        ),
        "unconditioned_seed": int(
            unconditioned[
                "seed"
            ]
        ),
        "unconditioned_text": (
            unconditioned[
                "text"
            ]
        ),
        "conditioned_seed": int(
            conditioned[
                "seed"
            ]
        ),
        "conditioned_text": (
            conditioned[
                "text"
            ]
        ),
        "llm_prompt": (
            llm[
                "prompt"
            ]
        ),
        "llm_text": (
            llm[
                "response"
            ]
        ),
    })


comparison = pd.DataFrame(
    comparison_rows
)

comparison_path = (
    DATA_DIR
    / "task4_three_way_comparison.csv"
)

comparison.to_csv(
    comparison_path,
    index=False,
)

for row in comparison.itertuples():
    print()
    print("=" * 80)
    print("THEME:", row.theme)
    print()
    print("Marcus passage:")
    print(row.marcus_passage)
    print()
    print(
        "Unconditioned statistical output "
        f"(seed={row.unconditioned_seed}):"
    )
    print(row.unconditioned_text)
    print()
    print(
        "Conditioned statistical output "
        f"(seed={row.conditioned_seed}):"
    )
    print(row.conditioned_text)
    print()
    print("Direct LLM output:")
    print(row.llm_text)
```

## Interpretation boundary

The three displayed sentences are useful for qualitative inspection, but they do not establish a distributional effect.

The statistical comparison in Task 5 uses all 500 conditioned and all 500 unconditioned draws for each theme.

The direct LLM comparison is explicitly illustrative rather than a matched statistical benchmark. There is only one LLM output per theme, and the pretrained LLM has linguistic and literary knowledge that the n-gram model does not. The prompt also asks the LLM to preserve an ethical direction when possible, even though the LLM receives only the theme label rather than the full Marcus proposition. Therefore fluency or moral coherence in the LLM examples must not be interpreted as a controlled accuracy advantage.
