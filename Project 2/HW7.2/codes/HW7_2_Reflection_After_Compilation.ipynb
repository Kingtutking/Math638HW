## 2. Reflection question

A theme condition does not determine one unique sentence. The experiment therefore studies a conditional distribution

$$
P(S | C),
$$

not a deterministic rule from one condition to one sentence.

The final reflection answers:

- Did conditioning measurably change the generated distribution?
- How much of the measured change was guaranteed by the explicit reward rule?
- Did unboosted thematic probes also change?
- Did conditioning damage target-corpus fluency or local word-pattern compatibility?
- Did the generated text preserve the ethical direction of the Marcus passage, or only its surface topic?
- How did the direct LLM output differ from the statistical generator?
- Which conclusions are supported by many draws, and which are only qualitative observations?

## 3. Final reflection

The experiment shows that conditioning changes the generated distribution in a real but limited way. Each theme used 500 unconditioned and 500 conditioned statistical draws with fixed seeds. The explicitly rewarded-word rate increased for all five themes, and every 95 percent bootstrap interval for that increase was above zero. The increases were about 0.078, 0.343, 0.151, 0.244, and 0.443 rewarded words per 100 words for discipline, mortality, anger, duty, and self-command. Thus the fixed probability bias is measurably operating.

That result should not be confused with preservation of Marcus Aurelius' meaning. The unboosted probe-word changes were small and every corresponding bootstrap interval included zero. The broad unigram Jensen-Shannon distances also do not provide separate evidence of a global lexical shift: the conditioned-versus-unconditioned distances were about 0.341 to 0.344, while pairwise distances between unconditioned samples from the same base generator ranged from about 0.336 to 0.345. At this vocabulary size and sample size, empirical JS distance is strongly affected by sampling sparsity. The clearest demonstrated effect is therefore the direct increase in rewarded vocabulary, not a verified semantic transformation.

The failure analysis makes the distinction between local style and global quality especially clear. About 47.9 percent of unconditioned outputs and 46.6 percent of conditioned outputs reached the 30-word limit before producing the learned end marker. At the same time, mean Odyssey bigram coverage was about 95.96 percent in both modes and repeated-bigram failures were essentially absent. Many fixed-rule examples are still awkward or incoherent. This shows that matching local Odyssey word transitions does not guarantee grammatical, meaningful, or literary-quality sentences.

The source proposition is also underdetermined by the conditioning information. The statistical generator receives only a theme label. For example, the label `anger` does not say whether the Marcus passage praises anger or recommends restraint. Increasing words such as `anger`, `wrath`, and `rage` can therefore strengthen topical resemblance while moving away from the ethical direction of the original passage. No semantic accuracy score is justified from this setup.

The direct LLM examples are qualitatively more fluent and usually express a coherent moral stance, but they are not a matched benchmark. There is only one LLM response per theme, the LLM has extensive pretrained linguistic knowledge, and the prompt asks it to preserve an ethical direction when possible. The exact-corpus check found no displayed statistical or LLM sentence as an exact normalized sentence in the Butler Odyssey, so all displayed outputs should be treated as generated inventions rather than quotations.

The mathematical object in this project is a conditional distribution $P(S | C)$, not a deterministic map from one theme to one sentence. The experiment succeeds at demonstrating a controlled, reproducible change in that distribution through an explicit lexical bias. It also demonstrates an important failure: a measurable conditioning effect can be shallow. Surface vocabulary can change without evidence that the model preserves the proposition, negation, or ethical stance of the source passage. For conditioned generation, measurable distributional change and meaningful semantic preservation must therefore be evaluated separately.
