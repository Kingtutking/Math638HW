# Appendix H. Interpretation Boundaries

The report deliberately avoids several claims that the experiments do not establish.

- It does not claim that Pyphen boundaries are phonetically correct syllables.
- It does not claim that the neural decoder independently deciphered the substitution; the Republic-informed system is explicitly two-stage.
- It does not claim that the tested context range exhausts all useful context sizes.
- It does not claim that the corpus is too short, because glyph-text length was not experimentally varied.
- It does not claim that the 3,052-type vocabulary is too large, because vocabulary size was not manipulated.
- It does not claim that *The Republic* is an inadequate reference in general; it only reports the performance obtained with this fixed reference.
- It does not treat model confidence, fluent fragments, or LLM plausibility judgments as substitutes for hidden-map evaluation.

- It does not interpret Republic-informed validation loss as hidden-map accuracy; that validation target is the inferred pseudo-label map.
- It distinguishes all-$$K$$ map recovery from recovery among observed test glyph types; test-center type coverage is 44.72%.
- It attributes improvement to the complete Republic-informed two-stage pipeline rather than claiming that the English statistics alone were isolated experimentally.
- Passage examples use an explicit tie-breaking rule and are saved in `../data/task10_decoded_passage_examples.csv`.

## I.2 Nonsyllabic artifact audit

The normalized *Meditations* tokenization text contains 31 Greek-word occurrences representing 26 distinct word forms and 231 non-ASCII alphabetic characters. They are not fed to the ASCII English syllable tokenizer and remain visible non-token spans. The count and handling are saved in `../data/task2_nonsyllabic_artifact_summary.csv`.

**CSV interoperability note.** The written syllable `nan` is a legitimate token in the data. Some dataframe readers, including pandas with its default NA rules, interpret the literal string `nan` as a missing value. The project notebooks use Python's `csv` module for these vocabulary/token files, so the experiment is unaffected. When independently checking vocabulary counts with pandas, use `keep_default_na=False`; otherwise the apparent distinct-token count can be understated by one.
