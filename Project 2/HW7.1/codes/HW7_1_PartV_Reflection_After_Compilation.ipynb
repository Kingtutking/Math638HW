# 4. Reflection question 1

## Why is this a cleaner scientific analogue of text-to-image generation?

This Lorenz experiment is a cleaner scientific analogue of conditioned image generation because every stage of the pipeline has an explicit mathematical meaning:

$$
c\longmapsto\Gamma_c\longmapsto I_c.
$$

The condition vector

$$
c=(\sigma,\rho,\beta,x_0,y_0,z_0,T)
$$

specifies the differential equation, its initial condition, and the integration horizon. For the reference case, the numerical solver produced 2501 sampled states, so

$$
\Gamma_c\in\mathbb{R}^{7503}.
$$

Unlike an ordinary image generator, the intermediate object is not judged only by appearance. Part II checked it directly against the Lorenz vector field using finite-difference derivative estimates. The relative component RMSE values are approximately

$$
0.00339,\qquad 0.00569,\qquad 0.00477
$$

for $x$, $y$, and $z$, and the relative vector RMSE is about

$$
\boxed{0.00495}.
$$

These values do not prove that the numerical trajectory is an exact solution, especially because the Lorenz system is chaotic and finite differences introduce their own truncation error. They do provide a quantitative consistency check that is much stronger than merely observing a butterfly-shaped picture.

Part III then renders the trajectory as a 256-by-256 RGB image,

$$
I_c\in\mathbb{R}^{196608}.
$$

The fact that the image vector has more coordinates than the trajectory vector does **not** mean that it contains more scientific information. The rendering keeps only the $(x,z)$ projection, rasterizes continuous coordinates to pixels, removes exact time labels and speed, and omits the $y$ coordinate entirely. Therefore the rendering map is many-to-one. A visually plausible image cannot, by itself, certify that the underlying trajectory satisfies the Lorenz equations.

Part IV makes the conditioning idea concrete at dataset scale. It stores 120 condition-trajectory-image triples, split into 84 training, 18 validation, and 18 test conditions. Rendering bounds are estimated from training trajectories only and then frozen. All validation and test trajectories in this particular sample fall inside those bounds, so no clipping was observed.

For these reasons, this experiment separates three questions that are often mixed together in image generation: **what was conditioned on, what mathematical object was generated, and how that object was visualized**. That separation makes correctness testable rather than merely visual.


# 5. Reflection question 2

## Should scientific generative AI map conditions directly to images, or to mathematical structures first?

For a scientific problem like this one, I would prefer the structure-first pipeline

$$
\boxed{c\longmapsto\widehat{\Gamma}_c\longmapsto I_c}
$$

over a direct image-only map

$$
c\longmapsto I.
$$

The main reason is **verifiability**. A direct condition-to-image model can produce a Lorenz-like butterfly without producing any state sequence that satisfies

$$
\dot{x}=\sigma(y-x),\qquad
\dot{y}=x(\rho-z)-y,\qquad
\dot{z}=xy-\beta z.
$$

By contrast, a structure-first model exposes a trajectory that can be checked before it is rendered. Part II demonstrates the kind of check that becomes possible: the stored trajectory can be differentiated numerically and compared with the Lorenz right-hand side. Part III shows why this check cannot generally be reconstructed from the final raster, because the rendering is not injective.

The Part IV dataset also supports this choice operationally. Each condition is stored alongside both its 2501-state trajectory and its 256-by-256 RGB image. The image can therefore be regenerated from the scientific object, while the scientific object remains available for quantitative analysis. The dataset does **not** train a neural generator; rather, it provides paired examples from which a future conditional model could be trained and evaluated.

This does not mean that structure-first generation is automatically correct. If a learned model predicts $\widehat{\Gamma}_c$ instead of numerically solving the Lorenz equations, that predicted trajectory must still be validated. It may violate the governing equations, have the wrong long-time statistics, or respond incorrectly to parameter changes. The advantage is that these failures are measurable in the mathematical representation.

Direct image generation can still be reasonable when the image itself is the scientific observable, when no trustworthy governing structure is available, or when the goal is visualization rather than physical prediction. But when a known mathematical structure exists and scientific validity matters, the evidence from this project favors

$$
\boxed{\text{condition}\to\text{mathematical structure}\to\text{rendering}}.
$$

The image should be treated as a visualization of a checkable scientific object, not as a substitute for that object.


# 6. Important qualifications for the final reflection

The final reflection should avoid several overclaims.

## A small ODE residual does not prove exact integration

The residual is a numerical consistency check at the chosen discretization.

It supports, rather than mathematically proves, that the stored trajectory follows the Lorenz equations.

## The image does not uniquely determine the trajectory

The selected rendering discards:

- the $$y$$ coordinate;
- exact numerical state values;
- exact time parametrization;
- trajectory speed;
- ODE residual information.

Therefore the image cannot replace the trajectory for scientific verification.

## A structure-first pipeline is not automatically perfect

If a learned model eventually predicts

$$
\widehat{\Gamma}_c
$$

rather than solving the ODE directly, the predicted structure would still need physical or mathematical validation.

The advantage is that the object being validated is explicit.

## The Part IV images are not themselves a trained generative model

Part IV constructs a conditioned dataset.

It demonstrates the representation pipeline

$$
c
\leftrightarrow
\Gamma_c
\leftrightarrow
I_c,
$$

but it does not train a neural condition-to-image generator.
