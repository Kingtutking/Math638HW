## The Lorenz System

For this project, I choose the **Lorenz system**

$$
\dot{x}=\sigma(y-x),
$$

$$
\dot{y}=x(r-z)-y,
$$

$$
\dot{z}=xy-\beta z,
$$

with the classical parameter values

$$
\sigma=10,\qquad r=28,\qquad \beta=\frac{8}{3}.
$$

The Lorenz system is a good choice for studying next-token prediction because it is a low-dimensional continuous dynamical system that produces complicated and chaotic trajectories. Although the state is described by only three variables, the trajectory evolves on the Lorenz attractor and repeatedly moves between its two lobes.

After the continuous state space is divided into discrete regions, the trajectory can be converted into a sequence of tokens. The resulting sequence contains meaningful temporal structure, but prediction remains difficult because the dynamics are chaotic.

This makes the Lorenz system a useful example for studying whether a sequence model can learn regularities in a deterministic dynamical system without simply memorizing a periodic trajectory.

## Dissipative Behavior

Let the Lorenz vector field be

$$
F(x,y,z)
=
\left(
\sigma(y-x),
x(r-z)-y,
xy-\beta z
\right).
$$

To determine whether the system is dissipative, compute the divergence:

$$
\nabla\cdot F
=
\frac{\partial \dot{x}}{\partial x}
+
\frac{\partial \dot{y}}{\partial y}
+
\frac{\partial \dot{z}}{\partial z}.
$$

For the Lorenz equations,

$$
\frac{\partial \dot{x}}{\partial x}=-\sigma,
$$

$$
\frac{\partial \dot{y}}{\partial y}=-1,
$$

and

$$
\frac{\partial \dot{z}}{\partial z}=-\beta.
$$

Therefore,

$$
\nabla\cdot F
=
-\sigma-1-\beta.
$$

Using

$$
\sigma=10,\qquad \beta=\frac{8}{3},
$$

we obtain

$$
\nabla\cdot F
=
-10-1-\frac{8}{3}
=
-\frac{41}{3}<0.
$$

Thus the Lorenz system is **dissipative**. The negative divergence means that phase-space volumes contract as time evolves. This contraction is consistent with trajectories approaching the Lorenz attractor rather than filling the entire three-dimensional phase space.

```python
sigma = 10
r = 28
beta = 8 / 3

divergence = -sigma - 1 - beta

print("sigma =", sigma)
print("r =", r)
print("beta =", beta)
print("divergence =", divergence)
```

## Chaotic Behavior

For the classical parameter values

$$
\sigma=10,\qquad r=28,\qquad \beta=\frac{8}{3},
$$

the Lorenz system exhibits chaotic dynamics.

Nearby initial conditions can eventually produce very different trajectories because the system has sensitive dependence on initial conditions. At the same time, the trajectories remain organized around the Lorenz attractor and repeatedly switch between its two lobes.

Therefore, the Lorenz system is not simply random. It combines deterministic equations with complicated and difficult-to-predict behavior.

This is useful for a next-token prediction experiment because a successful model should learn recurring local patterns in the trajectory while also dealing with the uncertainty associated with chaotic switching between the two lobes.

## Classification of the System

For the parameter values used in this project, the Lorenz system can be described as:

- **Deterministic:** the future state is determined by the differential equations and the initial condition.
- **Dissipative:** the divergence is negative,

$$
\nabla\cdot F=-\frac{41}{3}<0.
$$

- **Chaotic:** trajectories exhibit sensitive dependence on initial conditions and evolve on the Lorenz attractor.
- **Not Hamiltonian in this formulation:** Hamiltonian flows preserve phase-space volume, while the Lorenz system has strictly negative divergence and contracts phase-space volume.
- **Not treated as an integrable system in this chaotic parameter regime:** the dynamics do not reduce to a simple family of conserved quantities that determines the motion explicitly.

Therefore, the Lorenz system provides a particularly useful test case because it is **deterministic, dissipative, and chaotic**, while remaining low-dimensional and straightforward to simulate numerically.
