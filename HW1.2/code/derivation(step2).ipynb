## First-Order System

Using

$$
q=(x,y)
$$

and

$$
e_\theta=(\cos\theta,\sin\theta),
$$

the equations of motion can be written as the first-order system

$$
\dot{x}=v\cos\theta,
$$

$$
\dot{y}=v\sin\theta,
$$

$$
\dot{v}
=
-\frac{k}{M}
\frac{q\cdot e_\theta}{|q|^3},
$$

$$
\dot{\theta}=\omega,
$$

$$
\dot{\omega}=0.
$$

Since

$$
q\cdot e_\theta
=
x\cos\theta+y\sin\theta
$$

and

$$
|q|
=
\sqrt{x^2+y^2},
$$

the speed equation can also be written explicitly as

$$
\dot{v}
=
-\frac{k}{M}
\frac{x\cos\theta+y\sin\theta}
{(x^2+y^2)^{3/2}}.
$$

Therefore, the complete first-order system is

$$
\boxed{
\begin{aligned}
\dot{x} &= v\cos\theta, \\
\dot{y} &= v\sin\theta, \\
\dot{v} &=
-\frac{k}{M}
\frac{x\cos\theta+y\sin\theta}
{(x^2+y^2)^{3/2}}, \\
\dot{\theta} &= \omega, \\
\dot{\omega} &= 0.
\end{aligned}
}
$$

The last equation shows that the heading angular velocity is constant:

$$
\omega(t)=\omega_0.
$$
