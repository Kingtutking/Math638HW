# 1. Derivation of the Reduced Equations of Motion

We use the state variables

$$
(x,y,v,\theta,\omega),
$$

where

$$
\omega = \dot{\theta}.
$$

The center of the disk is

$$
q=(x,y),
$$

and the heading direction of the disk is

$$
e_\theta=(\cos\theta,\sin\theta).
$$

## Rolling Constraint

The no-slipping constraint requires the velocity of the center of the disk to remain parallel to its heading direction. Therefore,

$$
\dot q = v e_\theta.
$$

Substituting

$$
e_\theta=(\cos\theta,\sin\theta)
$$

gives

$$
(\dot x,\dot y)
=
(v\cos\theta,v\sin\theta).
$$

Therefore,

$$
\dot x = v\cos\theta
$$

and

$$
\dot y = v\sin\theta.
$$

The equivalent sideways-velocity constraint is

$$
-\dot x\sin\theta+\dot y\cos\theta=0.
$$

To check this, substitute the equations for $\dot x$ and $\dot y$:

$$
-(v\cos\theta)\sin\theta
+
(v\sin\theta)\cos\theta
=0.
$$

Thus the reduced position equations satisfy the no-slipping constraint.

## Gravitational Force

The gravitational potential is

$$
U(q)
=
-\frac{k}{|q|}.
$$

Since

$$
|q|
=
\sqrt{x^2+y^2},
$$

we can write

$$
U(x,y)
=
-\frac{k}{\sqrt{x^2+y^2}}.
$$

The force is

$$
F(q)
=
-\nabla U(q).
$$

First compute the derivatives:

$$
\frac{\partial U}{\partial x}
=
\frac{kx}{(x^2+y^2)^{3/2}},
$$

and

$$
\frac{\partial U}{\partial y}
=
\frac{ky}{(x^2+y^2)^{3/2}}.
$$

Therefore,

$$
F(q)
=
-\frac{k(x,y)}
{(x^2+y^2)^{3/2}}.
$$

Thus the force components are

$$
F_x
=
-\frac{kx}
{(x^2+y^2)^{3/2}}
$$

and

$$
F_y
=
-\frac{ky}
{(x^2+y^2)^{3/2}}.
$$

Because the rolling constraint only allows the center velocity along the heading direction, the relevant force component is the projection of gravity onto $e_\theta$:

$$
F(q)\cdot e_\theta.
$$

Using

$$
e_\theta=(\cos\theta,\sin\theta),
$$

we obtain

$$
F(q)\cdot e_\theta
=
F_x\cos\theta+F_y\sin\theta.
$$

Substituting the force components gives

$$
F(q)\cdot e_\theta
=
-\frac{k(x\cos\theta+y\sin\theta)}
{(x^2+y^2)^{3/2}}.
$$

The reduced equation for the speed is

$$
M\dot v
=
F(q)\cdot e_\theta.
$$

Therefore,

$$
\dot v
=
-\frac{k(x\cos\theta+y\sin\theta)}
{M(x^2+y^2)^{3/2}}.
$$

## Effective Mass

The effective translational mass is

$$
M
=
m+\frac{I_3}{R^2}.
$$

For a homogeneous thin disk,

$$
I_3
=
\frac{1}{2}mR^2.
$$

Therefore,

$$
M
=
m+\frac{\frac{1}{2}mR^2}{R^2}.
$$

So

$$
M
=
m+\frac{1}{2}m
=
\frac{3}{2}m.
$$

The heading moment of inertia is

$$
J
=
\frac{1}{4}mR^2.
$$

## Heading Equation

The model gives

$$
J\ddot\theta=0.
$$

Since $J$ is nonzero,

$$
\ddot\theta=0.
$$

Define

$$
\omega=\dot\theta.
$$

Then

$$
\dot\theta=\omega
$$

and

$$
\dot\omega=0.
$$

Therefore the heading angular velocity is constant:

$$
\omega(t)=\omega_0.
$$

# Final Reduced First-Order System

The five reduced equations are

$$
\dot x=v\cos\theta,
$$

$$
\dot y=v\sin\theta,
$$

$$
\dot v
=
-\frac{k(x\cos\theta+y\sin\theta)}
{M(x^2+y^2)^{3/2}},
$$

$$
\dot\theta=\omega,
$$

$$
\dot\omega=0.
$$

Thus, if the state vector is

$$
z=(x,y,v,\theta,\omega),
$$

then

$$
\dot z
=
\left(
v\cos\theta,
v\sin\theta,
-\frac{k(x\cos\theta+y\sin\theta)}
{M(x^2+y^2)^{3/2}},
\omega,
0
\right).
$$

# Mathematical and Dimensional Checks

## Check 1: Position Equations

The equations

$$
\dot x=v\cos\theta
$$

and

$$
\dot y=v\sin\theta
$$

must have units of velocity.

Since $\cos\theta$ and $\sin\theta$ are dimensionless,

$$
[v\cos\theta]
=
[v\sin\theta]
=
\frac{L}{T}.
$$

This agrees with the dimensions of $\dot x$ and $\dot y$.

Therefore the position equations are dimensionally correct.

## Check 2: Dimensions of the Gravitational Constant

The potential energy is

$$
U=-\frac{k}{r}.
$$

Potential energy has dimensions

$$
[U]
=
\frac{ML^2}{T^2}.
$$

Therefore,

$$
[k]
=
[U][r]
=
\frac{ML^3}{T^2}.
$$

## Check 3: Force Dimensions

The gravitational force is

$$
F
=
-\frac{kq}{|q|^3}.
$$

Its dimensions are

$$
[F]
=
\frac{(ML^3/T^2)(L)}
{L^3}.
$$

Therefore,

$$
[F]
=
\frac{ML}{T^2},
$$

which is the correct dimension for force.

## Check 4: Speed Equation

The speed equation is

$$
M\dot v
=
F\cdot e_\theta.
$$

The left side has dimensions

$$
[M\dot v]
=
M\frac{L}{T^2}.
$$

Since $e_\theta$ is dimensionless, the right side has the same dimensions as force:

$$
[F\cdot e_\theta]
=
\frac{ML}{T^2}.
$$

Therefore the speed equation is dimensionally consistent.

## Check 5: Effective Mass

The effective mass is

$$
M=m+\frac{I_3}{R^2}.
$$

Moment of inertia has dimensions

$$
[I_3]=ML^2.
$$

Therefore,

$$
\left[\frac{I_3}{R^2}\right]
=
\frac{ML^2}{L^2}
=
M.
$$

Thus both terms in the expression for $M$ have dimensions of mass.

## Check 6: Heading Equation

The heading equation is

$$
J\ddot\theta=0.
$$

Since $\theta$ is dimensionless,

$$
[\ddot\theta]
=
\frac{1}{T^2}.
$$

Since

$$
[J]=ML^2,
$$

we obtain

$$
[J\ddot\theta]
=
\frac{ML^2}{T^2},
$$

which has the dimensions of torque.

Therefore the heading equation is dimensionally consistent.

# Conclusion

The reduced nonholonomic equations are mathematically consistent with the rolling constraint and dimensionally consistent. The final system to be simulated is

$$
\dot x=v\cos\theta,
$$

$$
\dot y=v\sin\theta,
$$

$$
\dot v
=
-\frac{k(x\cos\theta+y\sin\theta)}
{M(x^2+y^2)^{3/2}},
$$

$$
\dot\theta=\omega,
$$

$$
\dot\omega=0.
$$

For a homogeneous thin disk,

$$
M=\frac{3}{2}m
$$

and

$$
J=\frac{1}{4}mR^2.
$$
