# 8. LLM Prediction and Comparison with Numerical Results

Before comparing with the numerical simulations, I asked the LLM to predict which quantities should be conserved.

## Prompt Given to the LLM

> Consider the flat-Earth nonholonomic Kepler model with
>
> $$
> \dot{x}=v\cos\theta,
> $$
>
> $$
> \dot{y}=v\sin\theta,
> $$
>
> $$
> \dot{v}
> =
> -\frac{k}{M}
> \frac{q\cdot e_\theta}{|q|^3},
> $$
>
> $$
> \dot{\theta}=\omega,
> $$
>
> $$
> \dot{\omega}=0.
> $$
>
> Predict whether the following quantities should be conserved:
>
> 1. Total energy
> 2. Ordinary orbital angular momentum
> 3. Heading angular momentum
>
> Explain the prediction before looking at the numerical results.

## LLM Prediction

The LLM predicted the following:

- Total energy should be conserved.
- Ordinary orbital angular momentum should not be conserved in general.
- Heading angular momentum should be conserved.

The reasoning was that the rolling constraint changes the usual central-force mechanics.

The gravitational force is central, but the nonholonomic constraint forces the velocity to remain parallel to the heading direction. Therefore, an additional constraint force is required to enforce the allowed motion.

This constraint force can produce torque about the Sun, so ordinary orbital angular momentum does not have to remain constant.

However, the simplified heading equation is

$$
\dot{\omega}=0,
$$

so the heading angular velocity is constant. Since $J$ is also constant,

$$
L_\theta=J\omega
$$

should be conserved.

The LLM also predicted that the total mechanical energy

$$
E
=
\frac{1}{2}Mv^2
+
\frac{1}{2}J\omega^2
-
\frac{k}{|q|}
$$

should remain constant.

# Comparison with Numerical Results

## Energy

The numerical simulations showed that the total energy remained constant to very high numerical accuracy.

| Case | Maximum Absolute Energy Error |
| --- | ---: |
| Case 1 | 4.96e-10 |
| Case 2 | 4.30e-11 |
| Case 3 | 9.52e-10 |
| Case 4 | 7.90e-10 |

The energy variation was approximately $10^{-9}$ or smaller.

Therefore, the numerical result agrees with the LLM prediction:

$$
E=\text{constant}.
$$

## Ordinary Orbital Angular Momentum

The ordinary orbital angular momentum is

$$
L_{\mathrm{orb}}
=
M(x\dot{y}-y\dot{x}).
$$

The numerical simulations showed:

| Case | Maximum Relative Change in Lorb |
| --- | ---: |
| Case 1 | 41.9% |
| Case 2 | approximately 0% |
| Case 3 | 125.9% |
| Case 4 | 70.6% |

Cases 1, 3, and 4 show large changes in orbital angular momentum.

Therefore,

$$
L_{\mathrm{orb}}
$$

is not conserved in general.

Case 2 is a special circular solution. Its orbital angular momentum remains constant because its initial conditions satisfy

$$
\omega=\frac{v}{r}.
$$

Thus Case 2 does not contradict the conclusion that ordinary orbital angular momentum is not conserved in general.

This agrees with the LLM prediction.

## Heading Angular Momentum

The heading angular momentum is

$$
L_\theta=J\omega.
$$

The numerical results showed zero change for every case.

| Case | Initial Ltheta | Maximum Absolute Change |
| --- | ---: | ---: |
| Case 1 | 0.05 | 0.0 |
| Case 2 | 0.10 | 0.0 |
| Case 3 | 0.05 | 0.0 |
| Case 4 | 0.05 | 0.0 |

Therefore,

$$
L_\theta=\text{constant}.
$$

This also agrees exactly with the LLM prediction.

# Overall Comparison

The LLM prediction was consistent with the numerical results.

| Quantity | LLM Prediction | Numerical Result |
| --- | --- | --- |
| Energy $E$ | Conserved | Conserved to numerical error |
| Orbital angular momentum $L_{\mathrm{orb}}$ | Not conserved in general | Not conserved in Cases 1, 3, and 4 |
| Heading angular momentum $L_\theta$ | Conserved | Conserved exactly |

The most important part of the prediction was recognizing that a central gravitational force does not automatically imply conservation of orbital angular momentum in this problem.

In the ordinary Kepler problem, a central force gives zero torque and therefore conserves orbital angular momentum. In the nonholonomic rolling-disk problem, the rolling constraint introduces an additional constraint force. This force can produce torque about the Sun.

Therefore, the LLM correctly distinguished this system from the ordinary Kepler problem.

The numerical experiment supports the final conservation picture:

- Energy is conserved.
- Ordinary orbital angular momentum is not conserved in general.
- Heading angular momentum is conserved.
