# Part V: Reflection

## 1. Was the LLM Better as a Tutor or as a Generator of Plausible Mistakes?

The LLM was more useful as a tutor, but this homework also showed that it can generate very convincing mathematical mistakes.

As a tutor, the LLM helped organize the probability spaces for European and American roulette, define the payoff random variables, derive expected values, and construct Python simulations to check the theoretical calculations.

For example, it correctly identified that for European roulette,

$$
P(\text{win})
=
\frac{18}{37}
$$

and

$$
P(\text{loss})
=
\frac{19}{37},
$$

which gives

$$
E[X]
=
-\frac{1}{37}.
$$

It was also useful for constructing the Martingale simulation and comparing theoretical probabilities with Monte Carlo results.

However, Part III showed that the LLM can also produce an incorrect argument that looks mathematically sophisticated. When explicitly asked to defend the gambler's fallacy, it was able to combine correct concepts such as expectation and the law of large numbers in an incorrect way.

Therefore, I found the LLM more useful as a tutor when its work could be checked, but the experiment showed that mathematical-looking language alone is not evidence that an argument is correct.

## 2. What Was the Most Convincing Wrong Argument?

The most convincing incorrect argument was the claim that a losing streak creates a deficit of wins that must be corrected by future spins.

The LLM began with the correct statement that the long-run fraction of wins should approach

$$
p
=
\frac{18}{37}.
$$

It then supposed that the first $n$ spins were all losses and claimed

$$
E\left[
\frac{
Y_1+\cdots+Y_{n+m}
}{
n+m
}
\middle|L_n
\right]
=
p.
$$

Because

$$
Y_1+\cdots+Y_n=0
$$

under the losing-streak event $L_n$, the LLM concluded that

$$
E[
Y_{n+1}+\cdots+Y_{n+m}
\mid L_n
]
=
(n+m)p.
$$

It then defined

$$
r
=
P(Y_{n+1}=1\mid L_n)
$$

and obtained

$$
mr=(n+m)p.
$$

Therefore,

$$
r
=
p\left(1+\frac{n}{m}\right).
$$

Since

$$
1+\frac{n}{m}>1,
$$

the argument concluded that

$$
P(Y_{n+1}=1\mid L_n)
>
P(Y_{n+1}=1).
$$

For five losses and $m=100$, this produced the false prediction

$$
P(\text{win next}\mid5\text{ losses})
\approx
0.5108,
$$

or approximately

$$
51.08\%.
$$

This argument was convincing because most of its individual mathematical objects were legitimate. The mistake was hidden inside the conditional-expectation step rather than being an obviously false arithmetic calculation.

## 3. How Did Conditional Probability Expose the Error?

Conditional probability makes the mistake clear.

Let

$$
L_n
=
\{Y_1=0,\ldots,Y_n=0\}
$$

be the event that the first $n$ spins are losses.

Because roulette spins are independent,

$$
P(Y_{n+1}=1\cap L_n)
=
P(Y_{n+1}=1)P(L_n).
$$

Therefore,

$$
P(Y_{n+1}=1\mid L_n)
=
\frac{
P(Y_{n+1}=1\cap L_n)
}{
P(L_n)
}.
$$

Substituting independence gives

$$
P(Y_{n+1}=1\mid L_n)
=
\frac{
P(Y_{n+1}=1)P(L_n)
}{
P(L_n)
}.
$$

Hence,

$$
P(Y_{n+1}=1\mid L_n)
=
P(Y_{n+1}=1).
$$

For European roulette,

$$
P(Y_{n+1}=1\mid L_n)
=
\frac{18}{37}
\approx
0.486486.
$$

The conditional probability is exactly the same as the unconditional probability.

The error in the false argument was assuming that

$$
E\left[
\frac{Y_1+\cdots+Y_{n+m}}{n+m}
\middle|L_n
\right]
=
p.
$$

The correct conditional expectation is

$$
E[
Y_{n+1}+\cdots+Y_{n+m}
\mid L_n
]
=
mp.
$$

Therefore,

$$
E\left[
\frac{Y_1+\cdots+Y_{n+m}}{n+m}
\middle|L_n
\right]
=
\frac{mp}{n+m}.
$$

As $m$ becomes large,

$$
\frac{mp}{n+m}
\to
p.
$$

Thus the long-run proportion can approach $p$ without the probability of any future spin increasing.

The Monte Carlo experiment confirmed this. After losing streaks of lengths

$$
1,2,3,5,7,\text{ and }10,
$$

the simulated probability of winning the next spin continued to fluctuate around

$$
\frac{18}{37}.
$$

There was no systematic increase following longer losing streaks.

## 4. What Does This Teach About Using LLMs for Mathematical Reasoning?

This homework showed that an LLM can be very useful for mathematical reasoning, but its output should be treated as a proposed argument rather than as a proof that is automatically correct.

The LLM was useful for:

- defining probability spaces and random variables;
- organizing calculations;
- deriving expectations and variances;
- proposing simulations;
- generating alternative ways to check a result;
- explaining the connection between theory and numerical experiments.

However, Part III demonstrated an important weakness. The LLM was able to construct a false argument that used correct mathematical terminology and several correct intermediate ideas.

The incorrect argument mentioned expectation, conditional expectation, and the law of large numbers, which made it appear rigorous. The central mistake only became obvious after checking the exact conditional-probability calculation.

The Martingale experiment provided another useful lesson. A strategy could produce a profit in approximately

$$
98.17\%
$$

of sessions and still have negative expected profit:

$$
E[\Pi]
\approx
-0.174.
$$

This shows why conclusions should be checked using the correct mathematical quantity rather than relying on an intuitive statement such as "the strategy usually wins."

Overall, I learned that an LLM is most useful when it is combined with independent mathematical verification.

A good workflow is

$$
\text{LLM suggestion}
\rightarrow
\text{identify assumptions}
\rightarrow
\text{derive the result}
\rightarrow
\text{check the logic}
\rightarrow
\text{verify numerically when appropriate}.
$$

The LLM can accelerate mathematical work, but it can also make an incorrect argument sound convincing. For that reason, definitions, assumptions, conditional probabilities, and key algebraic steps should still be checked independently.
