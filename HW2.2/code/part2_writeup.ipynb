# Part II: The Gambler's Fallacy

Suppose the player has lost $n$ times in a row.

Let

$$
L_n
=
\{X_1<0,X_2<0,\ldots,X_n<0\}
$$

denote the event that the first $n$ bets are all losses.

The question is whether a player who has already experienced this losing streak is now more likely to win the next spin.

We compare

$$
P(X_{n+1}>0\mid L_n)
$$

with

$$
P(X_{n+1}>0).
$$

The gambler's fallacy claims that after enough losses, the player becomes "due" for a win.

Under the roulette model, however, successive spins are independent.

---

# Theoretical Calculation

For European roulette,

$$
P(X_k>0)
=
\frac{18}{37}
$$

and

$$
P(X_k<0)
=
\frac{19}{37}.
$$

The probability of losing $n$ consecutive bets is

$$
P(L_n)
=
\left(\frac{19}{37}\right)^n.
$$

Now consider the event that the player loses the first $n$ bets and then wins the next bet.

By independence,

$$
P\left(
L_n\cap\{X_{n+1}>0\}
\right)
=
P(L_n)P(X_{n+1}>0).
$$

Therefore,

$$
P\left(
L_n\cap\{X_{n+1}>0\}
\right)
=
\left(\frac{19}{37}\right)^n
\left(\frac{18}{37}\right).
$$

Using the definition of conditional probability,

$$
P(X_{n+1}>0\mid L_n)
=
\frac{
P\left(
L_n\cap\{X_{n+1}>0\}
\right)
}{
P(L_n)
}.
$$

Substituting gives

$$
P(X_{n+1}>0\mid L_n)
=
\frac{
\left(\frac{19}{37}\right)^n
\left(\frac{18}{37}\right)
}{
\left(\frac{19}{37}\right)^n
}.
$$

The losing-streak probability cancels:

$$
P(X_{n+1}>0\mid L_n)
=
\frac{18}{37}.
$$

But

$$
P(X_{n+1}>0)
=
\frac{18}{37}.
$$

Therefore,

$$
P(X_{n+1}>0\mid L_n)
=
P(X_{n+1}>0).
$$

Numerically,

$$
P(X_{n+1}>0\mid L_n)
=
\frac{18}{37}
\approx
0.486486.
$$

Thus the probability of winning the next spin is approximately

$$
48.65\%
$$

regardless of how many consecutive losses occurred beforehand.

---

# Probability of the Losing Streak

Although the probability of winning the next spin does not change, the probability of observing a long losing streak does decrease as the streak becomes longer.

The theoretical probabilities are:

| Number of losses $n$ | $P(L_n)$ | $P(L_n\cap\text{win next})$ | $P(\text{win next}\mid L_n)$ |
| ---: | ---: | ---: | ---: |
| 1 | 0.513514 | 0.249817 | 0.486486 |
| 2 | 0.263696 | 0.128285 | 0.486486 |
| 3 | 0.135412 | 0.065876 | 0.486486 |
| 5 | 0.035707 | 0.017371 | 0.486486 |
| 7 | 0.009416 | 0.004581 | 0.486486 |
| 10 | 0.001275 | 0.000620 | 0.486486 |

For example, the probability of ten consecutive losses is only approximately

$$
0.001275.
$$

This is a rare event.

However, once those ten losses have already occurred,

$$
P(X_{11}>0\mid L_{10})
=
\frac{18}{37}.
$$

The rarity of the past sequence does not change the probability distribution of the next independent spin.

---

# Monte Carlo Verification

To verify the calculation numerically, I simulated

$$
1000000
$$

independent experiments for several different losing-streak lengths.

For each value of $n$, I identified experiments in which the first $n$ spins were all losses and then estimated the probability that spin $n+1$ was a win.

The results were:

| $n$ | Observed Losing Streaks | Theoretical $P(\text{win})$ | Simulated $P(\text{win after losses})$ | Z-Score |
| ---: | ---: | ---: | ---: | ---: |
| 1 | 513161 | 0.486486 | 0.486407 | -0.114 |
| 2 | 263884 | 0.486486 | 0.487718 | 1.266 |
| 3 | 135156 | 0.486486 | 0.486726 | 0.177 |
| 5 | 35573 | 0.486486 | 0.490597 | 1.551 |
| 7 | 9709 | 0.486486 | 0.483057 | -0.676 |
| 10 | 1240 | 0.486486 | 0.484677 | -0.127 |

![Probability of winning after a losing streak](gamblers_fallacy_conditional_probability.png)

The theoretical probability remains fixed at

$$
\frac{18}{37}
\approx
0.486486
$$

for every value of $n$.

The simulated estimates fluctuate slightly above and below the theoretical value.

For example, after five losses the simulation gave

$$
0.490597,
$$

which is slightly larger than

$$
0.486486.
$$

However, the difference corresponds to only about

$$
1.55
$$

standard errors.

This is ordinary random sampling variation and is not evidence that five losses make a win more likely.

Similarly, after seven consecutive losses the simulated probability decreased to

$$
0.483057.
$$

If losing streaks truly made wins more likely, we would expect the conditional probability to systematically increase with $n$.

The simulation shows no such pattern.

---

# Why the Estimates Become More Variable for Long Streaks

Long losing streaks are rare.

For example, among the $1000000$ simulated experiments, there were approximately

$$
513161
$$

one-loss streaks but only

$$
1240
$$

ten-loss streaks.

Therefore, the estimate after ten losses is based on a much smaller conditional sample.

The standard error is approximately

$$
\sqrt{
\frac{
p(1-p)
}{
N_{\mathrm{streak}}
}
}.
$$

As the number of observed streaks decreases, the standard error increases.

For $n=1$, the standard error was approximately

$$
0.000698.
$$

For $n=10$, it increased to approximately

$$
0.014194.
$$

Thus the larger visible fluctuations for long streaks are caused by a smaller sample size, not by a change in the underlying probability.

---

# The Error in the Gambler's Fallacy

The gambler's fallacy confuses two different probability statements.

It is true that

$$
P(L_n)
=
\left(\frac{19}{37}\right)^n
$$

becomes small when $n$ becomes large.

Therefore, before the spins occur, a long sequence of losses is unlikely.

However, once the losses have already occurred, they are part of the known past.

The relevant probability is then

$$
P(X_{n+1}>0\mid L_n),
$$

not the probability of the entire previous sequence.

Independence gives

$$
P(X_{n+1}>0\mid L_n)
=
P(X_{n+1}>0).
$$

Therefore, the inference

> "The last several bets lost, so the next one is more likely to win"

is mathematically incorrect.

The roulette wheel does not compensate for previous outcomes.

---

# Example: Five Losses

Before the experiment, the probability of losing five times consecutively is

$$
P(L_5)
=
\left(\frac{19}{37}\right)^5
\approx
0.035707.
$$

Thus five consecutive losses occur only about

$$
3.57\%
$$

of the time.

However, once those losses have occurred,

$$
P(X_6>0\mid L_5)
=
\frac{18}{37}
\approx
0.486486.
$$

The simulation gave

$$
0.490597,
$$

which is statistically consistent with the theoretical probability.

Thus the player is not "due" for a win.

---

# Example: Ten Losses

Ten consecutive losses are much rarer:

$$
P(L_{10})
=
\left(\frac{19}{37}\right)^{10}
\approx
0.001275.
$$

However,

$$
P(X_{11}>0\mid L_{10})
=
\frac{18}{37}
\approx
0.486486.
$$

The simulation gave

$$
0.484677.
$$

This is again consistent with the same theoretical probability.

Even after ten consecutive losses, the player is no more likely to win the next spin than before the losing streak began.

---

# Conclusion

The theoretical calculation gives

$$
P(X_{n+1}>0\mid L_n)
=
P(X_{n+1}>0)
=
\frac{18}{37}
$$

for every $n$.

The Monte Carlo simulations confirmed this result for losing streaks of

$$
n=1,2,3,5,7,10.
$$

All simulated conditional probabilities were statistically consistent with

$$
\frac{18}{37}.
$$

Therefore, there is no mathematical or numerical evidence that a losing streak makes the next roulette spin more likely to be a win.

The player is not "due" for a win.

The gambler's fallacy arises from confusing the low probability of a long sequence before it occurs with the probability of the next independent event after that sequence has already occurred.
