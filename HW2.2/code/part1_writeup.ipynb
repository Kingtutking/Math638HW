# European vs. American Roulette

European and American roulette use almost the same betting rules, but American roulette has one additional green pocket.

European roulette has

$$
37
$$

possible outcomes:

$$
\Omega_E
=
\{0,1,2,\ldots,36\}.
$$

American roulette has

$$
38
$$

possible outcomes:

$$
\Omega_A
=
\{0,00,1,2,\ldots,36\}.
$$

Both versions contain exactly

$$
18
$$

red numbers and

$$
18
$$

black numbers.

The difference is that European roulette has only the green pocket $0$, while American roulette has both $0$ and $00$.

---

## Probability of Winning a Red Bet

For European roulette,

$$
P_E(\text{win})
=
\frac{18}{37}
\approx
0.486486.
$$

For American roulette,

$$
P_A(\text{win})
=
\frac{18}{38}
=
\frac{9}{19}
\approx
0.473684.
$$

Therefore,

$$
P_E(\text{win})
>
P_A(\text{win}).
$$

A red bet is more likely to win in European roulette.

---

## Probability of Losing

For European roulette, a red bet loses on the 18 black numbers and the single zero.

Therefore,

$$
P_E(\text{loss})
=
\frac{19}{37}
\approx
0.513514.
$$

For American roulette, a red bet loses on the 18 black numbers, zero, and double zero.

Therefore,

$$
P_A(\text{loss})
=
\frac{20}{38}
=
\frac{10}{19}
\approx
0.526316.
$$

Thus,

$$
P_A(\text{loss})
>
P_E(\text{loss}).
$$

The extra $00$ pocket increases the probability that the player loses.

---

## Expected Value Comparison

For a one-unit bet on red, the payoff is

$$
X
=
\begin{cases}
1, & \text{win},\\
-1, & \text{loss}.
\end{cases}
$$

For European roulette,

$$
E[X^E]
=
\frac{18}{37}
-
\frac{19}{37}
=
-\frac{1}{37}.
$$

Therefore,

$$
E[X^E]
\approx
-0.027027.
$$

For American roulette,

$$
E[X^A]
=
\frac{18}{38}
-
\frac{20}{38}
=
-\frac{2}{38}.
$$

Thus,

$$
E[X^A]
=
-\frac{1}{19}
\approx
-0.052632.
$$

Both expected values are negative, but American roulette has the more negative expected value:

$$
-\frac{1}{19}
<
-\frac{1}{37}.
$$

Therefore, American roulette is worse for the player.

---

## Expected Loss Per Dollar

For European roulette, the expected loss per dollar is

$$
\frac{1}{37}
\approx
0.027027.
$$

This corresponds to approximately

$$
2.70\%
$$

of every dollar bet.

For American roulette, the expected loss per dollar is

$$
\frac{1}{19}
\approx
0.052632.
$$

This corresponds to approximately

$$
5.26\%
$$

of every dollar bet.

The ratio of the two expected losses is

$$
\frac{1/19}{1/37}
=
\frac{37}{19}
\approx
1.947.
$$

Thus the expected loss in American roulette is about

$$
1.95
$$

times the expected loss in European roulette.

It is nearly twice as unfavorable to the player.

---

## Variance Comparison

For European roulette,

$$
\mathrm{Var}(X^E)
=
1-
\left(-\frac{1}{37}\right)^2.
$$

Therefore,

$$
\mathrm{Var}(X^E)
=
\frac{1368}{1369}
\approx
0.999270.
$$

For American roulette,

$$
\mathrm{Var}(X^A)
=
1-
\left(-\frac{1}{19}\right)^2.
$$

Therefore,

$$
\mathrm{Var}(X^A)
=
\frac{360}{361}
\approx
0.997230.
$$

The two variances are very similar because the payoff from a single spin is still either $+1$ or $-1$.

The main difference between the two games is not the size of the individual payoff but the probability of winning.

# Why the Extra 00 Matters

The important point is that American roulette adds another losing outcome without changing the payout for a successful red bet.

In European roulette, the player wins on

$$
18
$$

out of

$$
37
$$

possible outcomes.

In American roulette, the player still wins on only

$$
18
$$

outcomes, but there are now

$$
38
$$

possible outcomes.

Therefore,

$$
\frac{18}{38}
<
\frac{18}{37}.
$$

The additional pocket benefits the casino because the player receives no additional compensation for taking on the extra losing outcome.

---

# Which Version Should the Player Choose?

If the player must choose between the two roulette games and intends to make the same even-money red bet, European roulette is mathematically better because

$$
E[X^E]
>
E[X^A].
$$

Specifically,

$$
-\frac{1}{37}
>
-\frac{1}{19}.
$$

However, both expected values are negative.

If the player is also allowed to choose not to gamble, then the three expected returns are

$$
0,
\qquad
-\frac{1}{37},
\qquad
-\frac{1}{19}.
$$

Therefore,

$$
0
>
-\frac{1}{37}
>
-\frac{1}{19}.
$$

Thus the expected-value ranking is

$$
\boxed{
\text{Do not bet}
>
\text{European roulette}
>
\text{American roulette}.
}
$$

European roulette is better than American roulette, but not betting is better than either one.

---

# Connection to the Gambler's Fallacy

The difference between European and American roulette changes the probability of winning a spin, but it does not change the independence assumption.

For European roulette,

$$
P(\text{win on next spin})
=
\frac{18}{37}.
$$

For American roulette,

$$
P(\text{win on next spin})
=
\frac{18}{38}.
$$

These probabilities remain the same regardless of previous outcomes.

For example, in European roulette,

$$
P(
X_6^E=1
\mid
X_1^E=\cdots=X_5^E=-1
)
=
\frac{18}{37}.
$$

Similarly, in American roulette,

$$
P(
X_6^A=1
\mid
X_1^A=\cdots=X_5^A=-1
)
=
\frac{18}{38}.
$$

Therefore, five previous losses do not make the sixth spin more likely to win in either version of roulette.

The wheel does not remember the previous outcomes.

---

# Conclusion

European and American roulette differ because American roulette contains the additional $00$ pocket.

For European roulette,

$$
P(\text{win})
=
\frac{18}{37}
$$

and

$$
E[X]
=
-\frac{1}{37}
\approx
-0.027027.
$$

For American roulette,

$$
P(\text{win})
=
\frac{18}{38}
$$

and

$$
E[X]
=
-\frac{1}{19}
\approx
-0.052632.
$$

Thus American roulette has an expected loss of approximately $5.26\%$, compared with approximately $2.70\%$ for European roulette.

The additional $00$ almost doubles the casino's expected advantage on an even-money bet.

However, the central idea for the gambler's fallacy remains unchanged: successive roulette spins are assumed to be independent, so past losses do not increase the probability of winning the next spin.
