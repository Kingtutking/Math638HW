# Part IV: Throwing Good Money After Bad

In this section, I studied a loss-chasing strategy known as the Martingale strategy.

The idea is to increase the next bet after every loss so that a later win recovers all previous losses and produces a small profit.

I asked the LLM to describe the strategy and then tested it analytically and with Monte Carlo simulation.

# LLM Prompt

> Describe a Martingale betting strategy for European roulette in which the player bets on red and doubles the bet after each loss. Include a finite bankroll and a maximum table bet. Explain how the strategy operates and under what circumstances it can fail.

# Martingale Strategy

The strategy used in the simulation is:

1. Start by betting one unit on red.
2. After a win, end the session.
3. After a loss, double the next bet.
4. Continue doubling after each loss.
5. Stop if the required next bet is larger than the remaining bankroll.
6. Stop if the required bet is larger than the table maximum.

The parameters used were

$$
B=100
$$

for the initial bankroll,

$$
b_0=1
$$

for the initial bet, and

$$
M=64
$$

for the maximum table bet.

The possible doubling sequence is

$$
1,2,4,8,16,32,64,\ldots
$$

However, with a bankroll of $100$, the player can only afford the first six bets:

$$
1,2,4,8,16,32.
$$

If all six bets lose, the total loss is

$$
1+2+4+8+16+32
=
63.
$$

After this sequence, only

$$
100-63=37
$$

units remain.

The next required bet would be

$$
64,
$$

which cannot be afforded.

Therefore, the maximum number of bets in one session is

$$
k=6.
$$

---

# Why a Successful Session Earns Only One Unit

Suppose the player loses the first $j$ bets and then wins the next bet.

The losses before the winning bet are

$$
1+2+4+\cdots+2^{j-1}.
$$

Using the geometric-series formula,

$$
1+2+\cdots+2^{j-1}
=
2^j-1.
$$

The next bet is

$$
2^j.
$$

If this bet wins, the net profit is

$$
2^j-(2^j-1)=1.
$$

Thus every successful Martingale session produces exactly

$$
+1
$$

unit of profit.

---

# Probability of a Small Profit

For European roulette,

$$
P(\text{win})
=
\frac{18}{37}
$$

and

$$
P(\text{loss})
=
\frac{19}{37}.
$$

Let

$$
q=\frac{19}{37}.
$$

The player earns the small profit of one unit if at least one of the six bets wins.

The only way the strategy fails is if all six bets lose.

Therefore,

$$
P(\text{large loss})
=
q^6.
$$

Hence,

$$
P(\text{small profit})
=
1-q^6.
$$

Substituting

$$
q=\frac{19}{37},
$$

gives

$$
P(\text{small profit})
=
1-
\left(\frac{19}{37}\right)^6.
$$

Numerically,

$$
P(\text{small profit})
\approx
0.9816637.
$$

Thus the strategy earns one unit in approximately

$$
98.17\%
$$

of sessions.

This extremely high winning percentage makes the Martingale strategy appear attractive.

---

# Probability of a Large Loss

A large loss occurs when all six bets lose.

Therefore,

$$
P(\text{large loss})
=
\left(\frac{19}{37}\right)^6.
$$

Numerically,

$$
P(\text{large loss})
\approx
0.0183363.
$$

Thus the large-loss event occurs only approximately

$$
1.83\%
$$

of the time.

However, when this event occurs, the player loses

$$
63
$$

units.

The theoretical session distribution is therefore approximately:

| Session Result | Probability | Profit |
| --- | ---: | ---: |
| Small profit | 98.17% | $+1$ |
| Large loss | 1.83% | $-63$ |

---

# Expected Profit

The expected profit is

$$
E[\Pi]
=
P(\text{small profit})(1)
+
P(\text{large loss})(-63).
$$

Therefore,

$$
E[\Pi]
=
(1-q^6)-63q^6.
$$

Combining terms,

$$
E[\Pi]
=
1-64q^6.
$$

Substituting

$$
q=\frac{19}{37},
$$

gives

$$
E[\Pi]
=
1-
64
\left(\frac{19}{37}\right)^6.
$$

Numerically,

$$
E[\Pi]
\approx
-0.173522.
$$

Therefore, despite winning approximately $98.17\%$ of sessions, the strategy loses approximately

$$
0.174
$$

units per session on average.

---

# Monte Carlo Verification

I simulated

$$
1000000
$$

independent Martingale sessions.

The theoretical probability of a small profit was

$$
0.9816637,
$$

while the simulated probability was

$$
0.981576.
$$

The theoretical probability of a large loss was

$$
0.0183363,
$$

while the simulated probability was

$$
0.018424.
$$

The simulated mean profit was

$$
-0.179136,
$$

compared with the theoretical expected profit

$$
-0.173522.
$$

The theoretical variance was approximately

$$
73.7283,
$$

and the Monte Carlo standard error of the estimated mean was approximately

$$
0.00859.
$$

The difference between the simulated and theoretical means was only approximately

$$
-0.654
$$

standard errors.

Therefore, the Monte Carlo result is fully consistent with the theoretical prediction.

![Martingale session outcomes](martingale_outcome_probabilities.png)

The plot shows that small profitable sessions occur overwhelmingly more often than large-loss sessions.

However, frequency alone does not determine expected value.

---

# Why the Strategy Can Win Often and Still Lose Money

The expected contribution from the successful sessions is

$$
(0.9816637)(1)
\approx
0.981664.
$$

The expected contribution from the large-loss sessions is

$$
(0.0183363)(-63)
\approx
-1.15519.
$$

Therefore,

$$
0.981664-1.15519
\approx
-0.173522.
$$

The rare losses are sufficiently large to outweigh the many one-unit profits.

Thus,

$$
P(\Pi>0)
\approx
0.9817
$$

while at the same time

$$
E[\Pi]<0.
$$

This demonstrates that

$$
P(\text{profit})>0.5
$$

does not imply

$$
E[\text{profit}]>0.
$$

A strategy can win almost every session and still lose money in the long run.

---

# Effect of a Finite Bankroll

Suppose the player can afford $k$ complete Martingale bets.

The betting sequence is

$$
1,2,4,\ldots,2^{k-1}.
$$

The total amount required to survive all $k$ losses is

$$
1+2+\cdots+2^{k-1}
=
2^k-1.
$$

Therefore, without a table limit, a bankroll $B$ allows approximately

$$
k
=
\left\lfloor
\log_2(B+1)
\right\rfloor
$$

complete Martingale bets.

For $k$ permitted bets,

$$
P(\text{small profit})
=
1-q^k
$$

and

$$
P(\text{large loss})
=
q^k.
$$

The loss after $k$ consecutive losses is

$$
2^k-1.
$$

Therefore,

$$
E[\Pi]
=
(1-q^k)
-
(2^k-1)q^k.
$$

This simplifies to

$$
E[\Pi]
=
1-2^kq^k.
$$

Hence,

$$
E[\Pi]
=
1-(2q)^k.
$$

For European roulette,

$$
2q
=
2\left(\frac{19}{37}\right)
=
\frac{38}{37}
>
1.
$$

Therefore,

$$
E[\Pi]<0
$$

for every positive value of $k$.

---

As the bankroll increases, the player can survive longer losing streaks.

$$

99.06\%.

Thus, increasing the bankroll makes the strategy appear increasingly reliable.

# Why More Bankroll Does Not Fix the Strategy

Although the probability of a small profit increases with the bankroll, the size of the possible failure also increases.

For four allowed bets, the maximum loss is

$$
15.
$$

For five allowed bets, the maximum loss is

$$
31.
$$

For six allowed bets, the maximum loss is

$$
63.
$$

For seven allowed bets, the maximum loss is

$$
127.
$$

At the same time, the expected profit becomes

$$
-0.11257,
$$

then

$$
-0.14264,
$$

then

$$
-0.17352,
$$

and finally

$$
-0.20524.
$$

![Martingale expected profit versus bankroll](martingale_expected_profit_vs_bankroll.png)

Thus, a larger bankroll increases the probability of winning the small one-unit profit but also permits the player to accumulate a larger loss before the strategy fails.

The underlying negative expectation of roulette has not disappeared.

---

# Effect of the Table Maximum

An important feature of the results is that the success probability eventually stops increasing.

The table maximum was

$$
M=64.
$$

Therefore, the largest possible bet in the doubling sequence is

$$
64.
$$

Once the player has enough bankroll to make the sequence

$$
1,2,4,8,16,32,64,
$$

the next required bet would be

$$
128,
$$

which exceeds the table maximum.

Therefore, even if the player has a much larger bankroll, the strategy cannot continue.

This explains why the results for bankrolls

$$
127,\ 255,\ 511
$$

are identical.

All three bankrolls allow exactly seven bets before the table limit prevents another doubling.

Their small-profit probability remains

$$
0.990584,
$$

and their expected profit remains approximately

$$
-0.205239.
$$

Thus the table maximum creates the plateaus visible in the bankroll plots.

---

# Does Increasing the Bankroll Help?

Increasing the bankroll helps in one limited sense:

$$
P(\text{small profit})
$$

increases.

However, it does not make the strategy profitable.

For $k$ permitted bets,

$$
E[\Pi]
=
1-
\left(\frac{38}{37}\right)^k.
$$

Since

$$
\frac{38}{37}>1,
$$

the expected value is negative.

In fact, before the table limit is reached, allowing additional doubling steps makes the expected session loss more negative.

Thus a larger bankroll does not eliminate the casino's advantage.

It simply changes the distribution from

$$
\text{frequent small wins and smaller occasional losses}
$$

to

$$
\text{even more frequent small wins and rarer but larger losses}.
$$

---

# Connection to the Gambler's Fallacy

The Martingale strategy does not make the next roulette spin more likely to win.

After any number of losses,

$$
P(\text{win next})
=
\frac{18}{37}.
$$

The strategy changes only the amount of money being risked.

For example, after five consecutive losses, the probability that the sixth spin wins is still

$$
\frac{18}{37}.
$$

The strategy responds to the losing streak by increasing the financial exposure rather than changing the probability of success.

Therefore, doubling after losses does not create a probabilistic advantage.

It places increasingly large amounts of money on the same negative expected-value bet.

---

# Conclusion

The Martingale strategy creates an illusion of reliability because it produces small profits in most sessions.

For the baseline case with

$$
B=100,
$$

initial bet $1$, and table maximum $64$,

$$
P(\text{small profit})
\approx
98.17\%
$$

while

$$
P(\text{large loss})
\approx
1.83\%.
$$

However, a successful session earns only

$$
+1,
$$

while a failed session loses

$$
-63.
$$

Consequently,

$$
E[\Pi]
\approx
-0.17352.
$$

The Monte Carlo simulation of one million sessions closely reproduced these theoretical results.

Increasing the bankroll increases the probability of a small profit, but it also increases the potential loss.

The table maximum eventually prevents further doubling, causing both the success probability and expected value to plateau.

Therefore, the Martingale strategy demonstrates that a gambling strategy can win very frequently and still have negative expected value.

The strategy does not overcome the house advantage. It transforms a sequence of ordinary negative expected-value bets into a distribution consisting of many small wins and rare large losses.
