# Homework 2.2: Throwing Good Money After Bad

## Introduction

This homework studies the gambler's fallacy and loss-chasing strategies using roulette.

The main idea behind the gambler's fallacy is the belief that after a sequence of losses, a win becomes more likely. For an independent game such as roulette, this intuition is incorrect.

The homework also investigates a Martingale loss-chasing strategy, in which the player doubles the next bet after every loss.

An LLM was used in several different roles:

- as a tutor for constructing the probability model;
- as an unreliable advocate asked to defend the gambler's fallacy;
- as a tool for proposing and simulating a loss-chasing strategy.

The mathematical claims were then checked using conditional probability, expectation, and Monte Carlo simulation.

---

# Part I: Probability Model for Roulette

## European Roulette

European roulette has sample space

$$
\Omega_E
=
\{0,1,2,\ldots,36\}.
$$

There are 37 possible outcomes.

Assuming a fair wheel, each outcome has probability

$$
P_E(\{\omega\})
=
\frac{1}{37}.
$$

Since the sample space is finite, the natural sigma-algebra is

$$
\mathcal F_E
=
2^{\Omega_E}.
$$

Thus the probability space is

$$
(\Omega_E,\mathcal F_E,P_E).
$$

The red event is

$$
A_{\mathrm{red}}
=
\{1,3,5,7,9,12,14,16,18,19,21,23,25,27,30,32,34,36\}.
$$

There are 18 red numbers.

Therefore,

$$
P_E(\mathrm{red})
=
\frac{18}{37}
\approx
0.486486.
$$

A red bet loses on the 18 black numbers and the single green zero.

Therefore,

$$
P_E(\mathrm{loss})
=
\frac{19}{37}
\approx
0.513514.
$$

---

## European Roulette Payoff Random Variable

Let

$$
X_n^E
$$

be the profit or loss from a one-unit bet on red on spin $n$.

Define

$$
X_n^E
=
\begin{cases}
1, & \text{if spin }n\text{ lands on red},\\
-1, & \text{otherwise}.
\end{cases}
$$

Then

$$
P(X_n^E=1)
=
\frac{18}{37}
$$

and

$$
P(X_n^E=-1)
=
\frac{19}{37}.
$$

The expected payoff is

$$
E[X_n^E]
=
1\left(\frac{18}{37}\right)
+
(-1)\left(\frac{19}{37}\right).
$$

Therefore,

$$
E[X_n^E]
=
-\frac{1}{37}
\approx
-0.027027.
$$

Thus the expected loss is approximately

$$
2.70\%
$$

of the amount bet.

---

# American Roulette

American roulette adds a second green pocket, $00$.

Its sample space is

$$
\Omega_A
=
\{0,00,1,2,\ldots,36\}.
$$

There are 38 possible outcomes.

Each outcome has probability

$$
P_A(\{\omega\})
=
\frac{1}{38}.
$$

The natural sigma-algebra is

$$
\mathcal F_A
=
2^{\Omega_A}.
$$

Thus the probability space is

$$
(\Omega_A,\mathcal F_A,P_A).
$$

There are still only 18 red numbers, so

$$
P_A(\mathrm{win})
=
\frac{18}{38}
=
\frac{9}{19}
\approx
0.473684.
$$

A red bet loses on the 18 black numbers, $0$, and $00$.

Therefore,

$$
P_A(\mathrm{loss})
=
\frac{20}{38}
=
\frac{10}{19}
\approx
0.526316.
$$

---

## American Roulette Payoff Random Variable

Define

$$
X_n^A
=
\begin{cases}
1, & \text{if spin }n\text{ lands on red},\\
-1, & \text{otherwise}.
\end{cases}
$$

Then

$$
E[X_n^A]
=
\frac{18}{38}
-
\frac{20}{38}.
$$

Therefore,

$$
E[X_n^A]
=
-\frac{1}{19}
\approx
-0.052632.
$$

Thus the expected loss is approximately

$$
5.26\%
$$

per dollar bet.

\frac{1/19}{1/37}

\frac{37}{19}

# Independence Assumption

The model assumes that different roulette spins are independent.

For European roulette,

$$
P(X_{n+1}^E=1)
=
\frac{18}{37}
$$

regardless of what happened on earlier spins.

For example,

$$
P(
X_6^E=1
\mid
X_1^E=\cdots=X_5^E=-1
)
=
\frac{18}{37}.
$$

Similarly, for American roulette,

$$
P(
X_{n+1}^A=1
\mid
\text{previous outcomes}
)
=
\frac{18}{38}.
$$

Thus the wheel is modeled as having no memory of previous outcomes.

This independence assumption is the key to analyzing the gambler's fallacy.

# Part II: Conditional Probability and the Gambler's Fallacy

Suppose the player has lost $n$ consecutive European roulette bets.

Define

$$
L_n
=
\{X_1<0,X_2<0,\ldots,X_n<0\}.
$$

We want to compare

$$
P(X_{n+1}>0\mid L_n)
$$

with

$$
P(X_{n+1}>0).
$$

---

## Probability of the Losing Streak

Since the spins are independent,

$$
P(L_n)
=
\left(\frac{19}{37}\right)^n.
$$

The probability of losing the first $n$ bets and then winning the next spin is

$$
P(
L_n\cap\{X_{n+1}>0\}
)
=
\left(\frac{19}{37}\right)^n
\left(\frac{18}{37}\right).
$$

Using conditional probability,

$$
P(X_{n+1}>0\mid L_n)
=
\frac{
P(
L_n\cap\{X_{n+1}>0\}
)
}{
P(L_n)
}.
$$

Substituting gives

$$
P(X_{n+1}>0\mid L_n)
=
\frac{
\left(\frac{19}{37}\right)^n
\left(\frac{18}{37}\right)
}{
\left(\frac{19}{37}\right)^n
}.
$$

The losing-streak probability cancels:

$$
P(X_{n+1}>0\mid L_n)
=
\frac{18}{37}.
$$

But

$$
P(X_{n+1}>0)
=
\frac{18}{37}.
$$

Therefore,

$$
\boxed{
P(X_{n+1}>0\mid L_n)
=
P(X_{n+1}>0)
}
$$

for every $n$.

Numerically,

$$
P(X_{n+1}>0\mid L_n)
\approx
0.486486.
$$

Thus a player is not "due" for a win after a losing streak.

# Why Long Streaks Have More Simulation Error

Long losing streaks are rare.

For example, the simulation observed

$$
513161
$$

one-loss streaks, but only

$$
1240
$$

ten-loss streaks.

The standard error of the conditional estimate is approximately

$$
\sqrt{
\frac{
p(1-p)
}{
N_{\mathrm{streak}}
}
}.
$$

Therefore, as the number of observed streaks decreases, the estimate becomes more variable.

This explains the larger fluctuations for long losing streaks without requiring any change in the underlying probability.

# Part III: Make the LLM Argue the Wrong Side

## Prompt Given to the LLM

> I know that many people say the gambler's fallacy is false, but I want you to argue mathematically that after several roulette losses in a row, the next spin is more likely to be a win. Give a convincing derivation using probability.

---

# LLM's Incorrect Argument

The LLM produced the following argument.

Let

$$
Y_k
=
\begin{cases}
1, & \text{if spin }k\text{ is a win},\\
0, & \text{if spin }k\text{ is a loss}.
\end{cases}
$$

For European roulette,

$$
P(Y_k=1)
=
p
=
\frac{18}{37}.
$$

Therefore,

$$
E[Y_k]=p.
$$

Over a large number of spins, the fraction of wins should approach

$$
p.
$$

Suppose the player has experienced $n$ consecutive losses.

Define

$$
L_n
=
\{Y_1=0,\ldots,Y_n=0\}.
$$

After this sequence, the player has zero wins, even though the expected number of wins after $n$ spins would normally be

$$
np.
$$

The LLM argued that future spins should compensate for this deficit.

It considered the next $m$ spins and claimed

$$
E\left[
\frac{
Y_1+\cdots+Y_{n+m}
}{
n+m
}
\middle|L_n
\right]
=
p.
$$

Because

$$
Y_1+\cdots+Y_n=0
$$

under $L_n$, it concluded that

$$
E[
Y_{n+1}+\cdots+Y_{n+m}
\mid L_n
]
=
(n+m)p.
$$

The LLM then defined

$$
r
=
P(Y_{n+1}=1\mid L_n)
$$

and assumed that the next $m$ spins have the same conditional winning probability $r$.

Therefore,

$$
mr=(n+m)p.
$$

Solving for $r$ gives

$$
r
=
\frac{n+m}{m}p.
$$

Hence,

$$
r
=
p\left(1+\frac{n}{m}\right).
$$

Since

$$
n>0,
$$

the LLM concluded that

$$
r>p.
$$

Therefore,

$$
P(Y_{n+1}=1\mid L_n)
>
P(Y_{n+1}=1).
$$

For example, with

$$
n=5
$$

and

$$
m=100,
$$

the argument gives

$$
r
=
\frac{105}{100}
\frac{18}{37}
\approx
0.510811.
$$

Thus the LLM falsely concluded that after five losses,

$$
P(\text{win next}\mid5\text{ losses})
\approx
51.08\%.
$$

The argument claimed that this increase occurs because roulette outcomes must eventually balance toward the long-run frequency

$$
\frac{18}{37}.
$$

---

# Diagnosis of the Mathematical Error

The LLM argument sounds convincing because it uses valid concepts such as expectation and the law of large numbers.

However, the central conditional-expectation equation is false.

The LLM claims

$$
E\left[
\frac{
Y_1+\cdots+Y_{n+m}
}{
n+m
}
\middle|L_n
\right]
=
p.
$$

Before conditioning, it is true that

$$
E\left[
\frac{
Y_1+\cdots+Y_{n+m}
}{
n+m
}
\right]
=
p.
$$

However, after conditioning on $L_n$, the first $n$ outcomes are already known to be losses.

Thus,

$$
Y_1+\cdots+Y_n=0.
$$

The future $m$ spins remain independent and each has expected value $p$.

Therefore,

$$
E[
Y_{n+1}+\cdots+Y_{n+m}
\mid L_n
]
=
mp.
$$

Hence,

$$
E\left[
\frac{
Y_1+\cdots+Y_{n+m}
}{
n+m
}
\middle|L_n
\right]
=
\frac{mp}{n+m}.
$$

This is not equal to $p$ for finite $m$.

---

# Correct Conditional Expectation

The correct relation is

$$
E[
Y_{n+1}+\cdots+Y_{n+m}
\mid L_n
]
=
mp.
$$

If

$$
r
=
P(Y_{n+1}=1\mid L_n),
$$

then independence gives

$$
mr=mp.
$$

Therefore,

$$
r=p.
$$

Hence,

$$
P(Y_{n+1}=1\mid L_n)
=
\frac{18}{37}.
$$

There is no increase in the probability of winning.

---

# Misuse of the Law of Large Numbers

The incorrect argument also misinterprets the law of large numbers.

The law of large numbers says that

$$
\frac{
Y_1+\cdots+Y_N
}{
N
}
\to p
$$

as $N$ becomes large.

It does not say that future probabilities must increase in order to compensate for past outcomes.

Suppose the first five spins are all losses.

If the next $m$ spins behave normally, their expected number of wins is

$$
mp.
$$

The expected proportion over all $m+5$ spins is

$$
\frac{mp}{m+5}.
$$

As

$$
m\to\infty,
$$

we have

$$
\frac{mp}{m+5}
=
p\frac{m}{m+5}
\to p.
$$

Thus the overall proportion approaches $p$ simply because the five initial losses become negligible relative to the growing number of later spins.

No short-run correction is required.

---

# Impossible Probability Check

Another indication that the false formula cannot be correct is that

$$
r
=
p\left(1+\frac{n}{m}\right)
$$

can produce probabilities larger than $1$.

For example, if

$$
n=200
$$

and

$$
m=100,
$$

then

$$
r
=
\frac{18}{37}
\left(
1+\frac{200}{100}
\right).
$$

Therefore,

$$
r
=
\frac{54}{37}
\approx
1.459.
$$

This would imply a probability of approximately

$$
145.9\%,
$$

which is impossible.

Thus the formula cannot represent a valid probability.

---

# Check Against the Five Possible Errors

## Conditional Probability Versus Intuition

The argument does not correctly compute

$$
P(Y_{n+1}=1\mid L_n).
$$

Instead, it replaces the conditional-probability calculation with the intuition that a deficit of wins must be corrected.

The correct calculation gives

$$
P(Y_{n+1}=1\mid L_n)=p.
$$

## Long-Run Frequency Versus Short-Run Correction

The argument confuses long-run convergence with an immediate correction mechanism.

The law of large numbers describes averages over many trials. It does not imply that the next trial must compensate for previous losses.

## Immediate Balancing

The argument assumes that the roulette wheel must quickly balance its past results.

Under the independent-spin model, no such mechanism exists.

## Changing the Probability Model

The argument effectively changes the probability of the next spin from

$$
p
$$

to a larger value

$$
r
$$

without any physical or probabilistic change to the roulette wheel.

The next spin still has the same sample space and probability distribution.

## Large Numbers Used to Make a False Next-Spin Claim

The statement

$$
\frac{Y_1+\cdots+Y_N}{N}\to p
$$

is a valid long-run statement.

It does not imply

$$
P(Y_{n+1}=1\mid L_n)>p.
$$

The false proof incorrectly uses a true asymptotic statement to justify a false statement about one future spin.

# Part IV: Throwing Good Money After Bad

For the loss-chasing experiment, I studied the classic Martingale strategy.

## LLM Prompt

> Describe a Martingale betting strategy for European roulette in which the player bets on red and doubles the bet after each loss. Include a finite bankroll and a maximum table bet. Explain how the strategy operates and under what circumstances it can fail.

---

# Martingale Strategy

The strategy is:

1. Start with a one-unit bet on red.
2. If the bet wins, end the session.
3. If the bet loses, double the next bet.
4. Continue doubling after each loss.
5. Stop if the required next bet cannot be funded by the bankroll.
6. Stop if the required bet exceeds the table maximum.

The simulation used

$$
B=100
$$

as the initial bankroll,

$$
b_0=1
$$

as the base bet, and

$$
M=64
$$

as the maximum table bet.

The possible sequence is

$$
1,2,4,8,16,32,64,\ldots
$$

With a bankroll of $100$, the player can afford the sequence

$$
1,2,4,8,16,32.
$$

The total loss after these six bets is

$$
1+2+4+8+16+32
=
63.
$$

The remaining bankroll is only

$$
37,
$$

so the next required bet of $64$ cannot be placed.

Thus the session permits at most

$$
k=6
$$

bets.

---

# Why a Successful Session Earns One Unit

If the player loses the first $j$ bets, the accumulated loss is

$$
1+2+\cdots+2^{j-1}
=
2^j-1.
$$

The next bet is

$$
2^j.
$$

If this bet wins, the net session profit is

$$
2^j-(2^j-1)
=
1.
$$

Therefore every successful Martingale session earns exactly

$$
+1.
$$

---

# Probability of a Small Profit

Let

$$
q
=
\frac{19}{37}
$$

be the probability of losing one European roulette red bet.

The only way to fail is to lose all six permitted bets.

Therefore,

$$
P(\text{large loss})
=
q^6.
$$

Thus,

$$
P(\text{small profit})
=
1-q^6.
$$

Numerically,

$$
P(\text{small profit})
=
1-
\left(\frac{19}{37}\right)^6
\approx
0.981664.
$$

Therefore, the player wins one unit in approximately

$$
98.17\%
$$

of sessions.

---

# Probability of a Large Loss

The failure probability is

$$
P(\text{large loss})
=
\left(\frac{19}{37}\right)^6
\approx
0.018336.
$$

Thus the failure occurs only about

$$
1.83\%
$$

of the time.

However, the failure produces a loss of

$$
63
$$

units.

The session distribution is therefore approximately:

| Outcome | Probability | Session Profit |
| --- | ---: | ---: |
| Small profit | 98.17% | $+1$ |
| Large loss | 1.83% | $-63$ |

---

# Expected Profit of the Martingale Strategy

The expected profit is

$$
E[\Pi]
=
(1-q^6)(1)
+
q^6(-63).
$$

Therefore,

$$
E[\Pi]
=
1-64q^6.
$$

Substituting

$$
q=\frac{19}{37}
$$

gives

$$
E[\Pi]
=
1-
64
\left(\frac{19}{37}\right)^6.
$$

Numerically,

$$
E[\Pi]
\approx
-0.173522.
$$

Thus the strategy loses money on average even though approximately $98.17\%$ of sessions are profitable.

# Why Winning Often Can Still Produce Negative Expected Value

The expected contribution of successful sessions is approximately

$$
(0.981664)(1)
=
0.981664.
$$

The expected contribution of failed sessions is approximately

$$
(0.018336)(-63)
\approx
-1.15519.
$$

Therefore,

$$
0.981664-1.15519
\approx
-0.173522.
$$

The rare large losses outweigh the many small wins.

Thus,

$$
P(\Pi>0)
\approx
0.9817
$$

while

$$
E[\Pi]<0.
$$

This demonstrates that a high probability of winning does not imply positive expected value.

# Effect of the Table Maximum

The table maximum in the experiment was

$$
64.
$$

Once the bankroll is large enough to fund

$$
1,2,4,8,16,32,64,
$$

the next required Martingale bet would be

$$
128.
$$

This exceeds the table maximum.

Therefore, additional bankroll cannot create another doubling step.

This explains why bankrolls

$$
127,\ 255,\text{ and }511
$$

produce identical results.

Their probability of a small profit remains approximately

$$
0.990584
$$

and their expected profit remains approximately

$$
-0.205239.
$$

# General Martingale Expected Value

If the player can make $k$ complete doubling bets, then

$$
P(\text{failure})
=
q^k.
$$

The loss after $k$ consecutive losses is

$$
2^k-1.
$$

Therefore,

$$
E[\Pi]
=
(1-q^k)
-
(2^k-1)q^k.
$$

Simplifying,

$$
E[\Pi]
=
1-2^kq^k.
$$

Hence,

$$
E[\Pi]
=
1-(2q)^k.
$$

For European roulette,

$$
2q
=
2\left(\frac{19}{37}\right)
=
\frac{38}{37}
>
1.
$$

Therefore,

$$
E[\Pi]<0.
$$

Thus the Martingale strategy does not remove the casino advantage.

It changes the distribution of outcomes into many small wins and rare large losses.

\frac{
Y_1+\cdots+Y_{n+m}
}{
n+m
}
\middle|L_n
\right]
=
p.
$$

From this, it derived

$$
P(Y_{n+1}=1\mid L_n)
=
p\left(1+\frac{n}{m}\right).
$$

For five losses and $m=100$, it obtained the false prediction

$$
P(\text{win next}\mid5\text{ losses})
\approx
51.08\%.
$$

The argument was convincing because most of its mathematical notation and terminology were valid. The error was hidden inside the conditional-expectation assumption.

## How Did Conditional Probability Expose the Error?

Conditional probability gives

$$
P(Y_{n+1}=1\mid L_n)
=
\frac{
P(Y_{n+1}=1\cap L_n)
}{
P(L_n)
}.
$$

By independence,

$$
P(Y_{n+1}=1\cap L_n)
=
P(Y_{n+1}=1)P(L_n).
$$

Therefore,

$$
P(Y_{n+1}=1\mid L_n)
=
P(Y_{n+1}=1).
$$

Hence,

$$
P(Y_{n+1}=1\mid L_n)
=
\frac{18}{37}.
$$

The previous losing streak has no effect on the probability of the next spin.

The law of large numbers does not require future spins to compensate for previous deviations. The influence of a fixed early losing streak simply becomes negligible as more independent spins occur.

## What Does This Teach About LLM Reliability?

This homework showed that an LLM can be a useful mathematical tool, but its output should be treated as a proposed argument rather than as automatically correct mathematics.

The LLM was useful for:

- organizing probability models;
- defining random variables;
- deriving equations;
- suggesting simulations;
- explaining mathematical concepts;
- producing alternative arguments.

However, it was also able to construct a false proof using correct mathematical terminology.

The most important lesson is that definitions, assumptions, conditioning, and logical implications must still be checked independently.

The Monte Carlo experiments were useful as an additional check, but numerical simulation also does not replace mathematical reasoning.

A useful workflow is

$$
\text{LLM suggestion}
\rightarrow
\text{identify assumptions}
\rightarrow
\text{derive the mathematics}
\rightarrow
\text{check critical steps}
\rightarrow
\text{simulate when appropriate}.
$$

Thus the LLM is useful for accelerating mathematical work, but it should not be treated as an unquestionable source of mathematical truth.

# Conclusion

This homework demonstrates two related mistakes in gambling intuition.

First, the gambler's fallacy incorrectly assumes that previous losses increase the probability of a future win.

For European roulette,

$$
P(X_{n+1}>0\mid L_n)
=
P(X_{n+1}>0)
=
\frac{18}{37}.
$$

The Monte Carlo experiments confirmed this for losing streaks of several different lengths.

Second, the Martingale strategy attempts to respond to losses by increasing the amount wagered.

For the baseline experiment,

$$
P(\text{small profit})
\approx
98.17\%,
$$

but

$$
E[\Pi]
\approx
-0.17352.
$$

Thus a strategy can win very frequently while still having negative expected value.

The LLM was useful for building the models and simulations, but Part III demonstrated that it can also produce a mathematically plausible false argument.

The main lesson is that probabilistic reasoning should be based on explicit probability models, conditional probabilities, and expected values rather than intuition or persuasive mathematical language.
