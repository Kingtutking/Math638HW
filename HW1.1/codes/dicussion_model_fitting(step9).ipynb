## Discussion: Kepler's Problem as Model Fitting from Data

This problem is an early example of model fitting from data because we begin with observations and ask which mathematical model best explains them.

Instead of assuming the correct orbit in advance, we compare several candidate models:

- Sun-centered circle
- Shifted circle
- Ellipse

We then compare the residual errors of each model and determine which one best matches the reconstructed Mars positions.

Kepler did something conceptually similar when he analyzed Tycho Brahe's observations of Mars. He did not simply accept the traditional idea that planetary motion had to be circular. Instead, he compared mathematical models with observational data and looked for a model that explained the measurements more accurately.

This is similar to modern approximation and machine learning. In many modern problems, the true mathematical relationship is unknown. We observe data, choose a family of possible models, fit the parameters, and compare how well the models explain the data.

The general process is:

1. Collect observations
2. Choose candidate models
3. Fit the model parameters
4. Compute the errors
5. Compare the models
6. Select the model best supported by the data

Our experiment also showed an important limitation of model fitting. The Sun-centered circle had a much larger error than the other models, so the data clearly rejected that model. However, the shifted circle and ellipse had almost the same error.

Therefore, our data were strong enough to reject a Sun-centered circular orbit, but they were not strong enough to clearly prove that the ellipse was better than a shifted circle.

This demonstrates an important idea in scientific modeling: conclusions should come from what the data actually support, rather than from what we already know the correct historical answer to be.
