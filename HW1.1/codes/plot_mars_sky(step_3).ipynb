```python
"""
plot_mars_sky.py

Homework 1.1 - Step 3

Plot the apparent position of Mars in the sky from Uraniborg
using the synthetic observation tables produced in Step 1.

Produces:
    - Right Ascension vs Declination plots
    - Azimuth vs Altitude plots

for noise levels:
    1 arcmin
    2 arcmin
    5 arcmin
"""

import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD THE SYNTHETIC OBSERVATIONS
# ============================================================

noise_levels = [1, 2, 5]

data = {}

for noise in noise_levels:

    filename = (
        f"synthetic_mars_{noise}arcmin.csv"
    )

    df = pd.read_csv(filename)

    data[noise] = df

    print(
        f"Loaded {filename}: "
        f"{len(df)} observations"
    )


# ============================================================
# 2. RA / DECLINATION PLOTS
# ============================================================

def plot_radec(df, noise_arcmin):

    plt.figure(
        figsize=(10, 6)
    )

    # Scatter plot is better than connecting observations
    # because RA wraps from 360 degrees back to 0 degrees.
    plt.scatter(
        df["ra_deg"],
        df["dec_deg"],
        s=12,
        alpha=0.7
    )

    plt.xlabel(
        "Right Ascension (degrees)"
    )

    plt.ylabel(
        "Declination (degrees)"
    )

    plt.title(
        "Apparent Position of Mars from Uraniborg\n"
        f"RA-Dec, noise = {noise_arcmin} arcmin"
    )

    plt.grid(True)

    plt.tight_layout()

    filename = (
        f"mars_radec_{noise_arcmin}arcmin.png"
    )

    plt.savefig(
        filename,
        dpi=300
    )

    plt.show()


# ============================================================
# 3. ALTITUDE / AZIMUTH PLOTS
# ============================================================

def plot_altaz(df, noise_arcmin):

    plt.figure(
        figsize=(10, 6)
    )

    plt.scatter(
        df["az_deg"],
        df["alt_deg"],
        s=12,
        alpha=0.7
    )

    plt.xlabel(
        "Azimuth (degrees)"
    )

    plt.ylabel(
        "Altitude (degrees)"
    )

    plt.title(
        "Apparent Position of Mars in Tycho's Sky\n"
        f"Uraniborg, noise = {noise_arcmin} arcmin"
    )

    # Standard horizon limits.
    plt.xlim(
        0,
        360
    )

    plt.ylim(
        0,
        90
    )

    plt.grid(True)

    plt.tight_layout()

    filename = (
        f"mars_altaz_{noise_arcmin}arcmin.png"
    )

    plt.savefig(
        filename,
        dpi=300
    )

    plt.show()


# ============================================================
# 4. PLOT ALL THREE NOISE LEVELS
# ============================================================

for noise in noise_levels:

    df = data[noise]

    plot_radec(
        df,
        noise
    )

    plot_altaz(
        df,
        noise
    )


# ============================================================
# 5. PRINT BASIC INFORMATION
# ============================================================

for noise in noise_levels:

    df = data[noise]

    print(
        "\n================================="
    )

    print(
        f"Noise level: {noise} arcmin"
    )

    print(
        "================================="
    )

    print(
        "Number of observations:",
        len(df)
    )

    print(
        "RA range:",
        df["ra_deg"].min(),
        "to",
        df["ra_deg"].max()
    )

    print(
        "Dec range:",
        df["dec_deg"].min(),
        "to",
        df["dec_deg"].max()
    )

    print(
        "Altitude range:",
        df["alt_deg"].min(),
        "to",
        df["alt_deg"].max()
    )

    print(
        "Azimuth range:",
        df["az_deg"].min(),
        "to",
        df["az_deg"].max()
    )
```
