# Part IV: Finite-Speed Propagation

The Barenblatt solution has compact support because

$$
\rho(t,x)
=
t^{-\alpha}
\left(
C-k\frac{|x|^2}{t^{2\beta}}
\right)_+^{1/(m-1)}.
$$

The solution is positive only when

$$
C-k\frac{|x|^2}{t^{2\beta}}>0.
$$

Therefore,

$$
k\frac{|x|^2}{t^{2\beta}}<C,
$$

so

$$
|x|^2
<
\frac{C}{k}t^{2\beta}.
$$

Taking square roots gives

$$
|x|
<
\left(
\frac{C}{k}
\right)^{1/2}
t^\beta.
$$

Thus the support radius is

$$
\boxed{
R(t)
=
\left(
\frac{C}{k}
\right)^{1/2}
t^\beta
}.
$$

Outside this radius,

$$
\rho(t,x)=0.
$$

Therefore, for every finite time

$$
t>0,
$$

the solution remains supported inside a finite region.

This is the finite-speed propagation property of the porous medium equation.

# Example: $d=1$, $m=2$, $C=1$

From the previous parts,

$$
\beta
=
\frac13
$$

and

$$
k
=
\frac1{12}.
$$

Therefore,

$$
R(t)
=
\left(
\frac{1}{1/12}
\right)^{1/2}
t^{1/3}.
$$

Hence,

$$
\boxed{
R(t)
=
\sqrt{12}\,t^{1/3}
}.
$$

The numerical support radii are approximately:

| $t$ | $R(t)$ |
| ---: | ---: |
| $0.25$ | $2.182247$ |
| $0.5$ | $2.749459$ |
| $1$ | $3.464102$ |
| $2$ | $4.364495$ |
| $4$ | $5.498919$ |

![Finite-speed expansion of the support](hw41_part4_support_radius.png)

The plot shows that the support radius increases with time.

However, the radius remains finite for every finite positive time.

Thus probability mass does not appear arbitrarily far away immediately.

# Speed of the Free Boundary

Differentiating

$$
R(t)
=
\sqrt{12}\,t^{1/3}
$$

gives

$$
R'(t)
=
\frac{\sqrt{12}}{3}t^{-2/3}.
$$

Thus,

$$
\boxed{
R'(t)
=
\frac{\sqrt{12}}{3}t^{-2/3}
}.
$$

For the times used in the numerical experiment, the boundary speed is finite and positive.

![Speed of the Barenblatt free boundary](hw41_part4_boundary_speed.png)

The graph shows that the boundary moves fastest at early times and then slows down.

Since

$$
t^{-2/3}
$$

decreases as $t$ increases,

$$
R'(t)
$$

also decreases.

Therefore, the support continues to expand, but its rate of expansion becomes slower with time.

# Important Point Near $t=0$

The formula

$$
R'(t)
=
\frac{\sqrt{12}}{3}t^{-2/3}
$$

shows that

$$
R'(t)\to\infty
$$

as

$$
t\to0^+.
$$

This does not contradict finite-speed propagation.

Finite-speed propagation means that for every fixed

$$
t>0,
$$

the support radius

$$
R(t)
$$

is finite.

It does not require the derivative

$$
R'(t)
$$

to remain uniformly bounded all the way down to the singular initial time

$$
t=0.
$$

For every positive time,

$$
R(t)<\infty.
$$

Thus the solution never instantly develops support over all of space.

# Comparison with the Heat Equation

For the ordinary heat equation,

$$
\partial_t\rho
=
\Delta\rho,
$$

a compactly supported initial condition immediately becomes positive everywhere.

Thus, for any

$$
t>0,
$$

the heat-equation solution generally has support over all of

$$
\mathbb R^d.
$$

The porous medium equation behaves differently.

For

$$
\partial_t\rho
=
\Delta(\rho^m),
\qquad
m>1,
$$

the Barenblatt solution satisfies

$$
\rho(t,x)=0
$$

whenever

$$
|x|>R(t).
$$

Since

$$
R(t)
=
\left(
\frac{C}{k}
\right)^{1/2}
t^\beta
$$

is finite for every finite

$$
t>0,
$$

the disturbance cannot reach arbitrarily distant points instantaneously.

# Numerical Verification

The Python program checked that:

- every tested support radius was finite;
- every support radius was positive;
- the support radius increased with time;
- every tested boundary speed was finite;
- every tested boundary speed was positive.

The numerical experiment therefore agrees with the analytic formula

$$
R(t)
=
\sqrt{12}\,t^{1/3}.
$$

# Conclusion

The Barenblatt solution has support

$$
|x|
\le
R(t),
$$

where

$$
\boxed{
R(t)
=
\left(
\frac{C}{k}
\right)^{1/2}
t^\beta
}.
$$

For the case

$$
d=1,
\qquad
m=2,
\qquad
C=1,
$$

this becomes

$$
\boxed{
R(t)
=
\sqrt{12}\,t^{1/3}
}.
$$

The support expands with time but remains finite for every finite positive time.

Therefore, the porous medium equation exhibits finite-speed propagation, in contrast with the ordinary heat equation, which has instantaneous propagation.
