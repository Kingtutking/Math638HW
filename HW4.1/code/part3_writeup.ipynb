# Part III: Plotting the Barenblatt Solution

For this experiment, I choose

$$
d=1,
\qquad
m=2,
$$

and

$$
C=1.
$$

From Parts I and II,

$$
\alpha
=
\frac13,
\qquad
\beta
=
\frac13,
$$

and

$$
k
=
\frac1{12}.
$$

Therefore, the Barenblatt solution becomes

$$
\boxed{
\rho(t,x)
=
t^{-1/3}
\left(
1-
\frac{x^2}
{12t^{2/3}}
\right)_+
}.
$$

I evaluated the solution at

$$
t
=
0.25,\ 0.5,\ 1,\ 2,\ 4.
$$

# Barenblatt Profiles

![Barenblatt profiles](hw41_part3_barenblatt_profiles.png)

The plots clearly show that the solution becomes wider and lower as time increases.

At early times, the density is narrow and has a relatively large maximum.

At later times, the solution spreads over a larger interval while its maximum height decreases.

This demonstrates the self-similar spreading behavior of the Barenblatt solution.

# Maximum Height

The maximum occurs at

$$
x=0.
$$

Therefore,

$$
\rho(t,0)
=
t^{-1/3}.
$$

The numerical values were:

| $t$ | Maximum height |
| ---: | ---: |
| $0.25$ | $1.587401$ |
| $0.5$ | $1.259921$ |
| $1$ | $1.000000$ |
| $2$ | $0.793701$ |
| $4$ | $0.629961$ |

![Maximum height](hw41_part3_maximum_height.png)

The maximum height decreases strictly as time increases.

This agrees with the scaling law

$$
\rho(t,0)
\sim
t^{-1/3}.
$$

Thus the density becomes lower while spreading outward.

# Compact Support and Support Radius

The solution is positive only when

$$
1-
\frac{x^2}
{12t^{2/3}}
>
0.
$$

Therefore,

$$
x^2
<
12t^{2/3}.
$$

Hence,

$$
|x|
<
\sqrt{12}\,t^{1/3}.
$$

The support radius is

$$
\boxed{
R(t)
=
\sqrt{12}\,t^{1/3}
}.
$$

Thus the support is

$$
\boxed{
[-R(t),R(t)].
}
$$

The numerical support radii were:

| $t$ | Exact support radius | Numerical support radius |
| ---: | ---: | ---: |
| $0.25$ | $2.182247$ | $2.181696$ |
| $0.5$ | $2.749459$ | $2.749253$ |
| $1$ | $3.464102$ | $3.463838$ |
| $2$ | $4.364495$ | $4.363392$ |
| $4$ | $5.498919$ | $5.498506$ |

The small differences between the exact and numerical radii are caused by the finite spatial grid.

![Support radius](hw41_part3_support_radius.png)

The support radius increases strictly with time.

However, for every finite

$$
t>0,
$$

the radius is finite.

Therefore, the Barenblatt solution remains compactly supported at every finite positive time.

This is an example of finite-speed propagation.

# Mass Conservation

Although I chose

$$
C=1,
$$

the total mass is not equal to $1$.

For this choice of parameters,

$$
M
=
\int_{\mathbb R}
\rho(t,x)\,dx.
$$

Using

$$
x=t^{1/3}\xi,
$$

we obtain

$$
M
=
\int_{-\sqrt{12}}^{\sqrt{12}}
\left(
1-\frac{\xi^2}{12}
\right)
d\xi.
$$

Evaluating the integral gives

$$
\boxed{
M
=
\frac{8\sqrt3}{3}
}.
$$

Numerically,

$$
M
\approx
4.6188021535.
$$

The numerical calculation gave approximately

$$
4.618802
$$

at every tested time.

The maximum change in numerical mass was

$$
3.42\times10^{-7},
$$

and the maximum relative change was only

$$
7.40\times10^{-8}.
$$

The maximum difference between the numerical mass and the exact value was approximately

$$
2.20\times10^{-7}.
$$

![Mass conservation](hw41_part3_mass_conservation.png)

The plot uses an axis offset near

$$
4.618802,
$$

so the visible fluctuations are actually only on the order of

$$
10^{-7}.
$$

Thus the total mass is numerically conserved to very high accuracy.

# Numerical Verification

The numerical experiment verified all three properties requested in the problem.

First, the solution spreads in space because

$$
R(t)
=
\sqrt{12}\,t^{1/3}
$$

increases with time.

Second, the maximum height decreases because

$$
\rho(t,0)
=
t^{-1/3}.
$$

Third, the support remains compact because

$$
\rho(t,x)=0
$$

whenever

$$
|x|
\ge
\sqrt{12}\,t^{1/3}.
$$

The numerical simulation also provided an additional consistency check through conservation of mass.

# Conclusion

For

$$
d=1,
\qquad
m=2,
\qquad
C=1,
$$

the Barenblatt solution is

$$
\boxed{
\rho(t,x)
=
t^{-1/3}
\left(
1-
\frac{x^2}
{12t^{2/3}}
\right)_+
}.
$$

The numerical plots confirm that:

- the solution spreads outward with time;
- its maximum height decreases with time;
- its support remains compact for every finite $t>0$;
- its total mass remains constant.

Therefore, the experiment clearly demonstrates the characteristic finite-speed diffusion of the porous medium equation.
