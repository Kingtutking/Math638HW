```python
# ============================================================
# HOMEWORK 4.1 - PART I
# SELF-SIMILAR EXPONENT VERIFICATION
# ============================================================

import numpy as np
import pandas as pd


# ============================================================
# FUNCTIONS FOR THE EXPONENTS
# ============================================================

def similarity_exponents(
    d,
    m
):

    denominator = (
        d * (m - 1.0)
        +
        2.0
    )

    beta = (
        1.0
        /
        denominator
    )

    alpha = (
        d
        /
        denominator
    )

    return (
        alpha,
        beta
    )


# ============================================================
# TEST CASES
# ============================================================

test_cases = [

    (1, 2.0),
    (1, 3.0),
    (2, 2.0),
    (2, 3.0),
    (3, 2.0),
    (3, 4.0),
    (5, 2.5)

]


results = []


for d, m in test_cases:

    alpha, beta = similarity_exponents(
        d,
        m
    )


    # ========================================================
    # MASS CONSERVATION CHECK
    #
    # alpha = d beta
    # ========================================================

    mass_left = alpha

    mass_right = (
        d * beta
    )

    mass_error = (

        mass_left

        -

        mass_right
    )


    # ========================================================
    # PDE SCALING CHECK
    #
    # alpha + 1
    #
    # =
    #
    # m alpha + 2 beta
    # ========================================================

    scaling_left = (
        alpha
        +
        1.0
    )

    scaling_right = (
        m * alpha
        +
        2.0 * beta
    )

    scaling_error = (

        scaling_left

        -

        scaling_right
    )


    results.append({

        "d":
            d,

        "m":
            m,

        "alpha":
            alpha,

        "beta":
            beta,

        "d_beta":
            mass_right,

        "mass_error":
            mass_error,

        "alpha_plus_1":
            scaling_left,

        "m_alpha_plus_2_beta":
            scaling_right,

        "scaling_error":
            scaling_error

    })


results_df = pd.DataFrame(
    results
)


# ============================================================
# PRINT RESULTS
# ============================================================

print("=" * 75)

print(
    "SELF-SIMILAR EXPONENT VERIFICATION"
)

print("=" * 75)

print()

print(
    results_df.to_string(
        index=False
    )
)


# ============================================================
# AUTOMATIC TESTS
# ============================================================

max_mass_error = np.max(

    np.abs(

        results_df[
            "mass_error"
        ]
    )
)


max_scaling_error = np.max(

    np.abs(

        results_df[
            "scaling_error"
        ]
    )
)


print()

print("=" * 75)

print(
    "AUTOMATIC VERIFICATION"
)

print("=" * 75)

print(
    "Maximum mass-conservation error =",
    max_mass_error
)

print(
    "Maximum PDE-scaling error =",
    max_scaling_error
)


assert np.allclose(

    results_df[
        "alpha"
    ],

    results_df[
        "d_beta"
    ]
)


assert np.allclose(

    results_df[
        "alpha_plus_1"
    ],

    results_df[
        "m_alpha_plus_2_beta"
    ]
)


print()

print(
    "PASS: alpha = d beta for every test case."
)

print(
    "PASS: alpha + 1 = m alpha + 2 beta "
    "for every test case."
)


# ============================================================
# EXAMPLE: d = 1, m = 2
# ============================================================

d = 1
m = 2.0


alpha, beta = similarity_exponents(
    d,
    m
)


print()

print("=" * 75)

print(
    "EXAMPLE: d = 1, m = 2"
)

print("=" * 75)

print(
    "alpha =",
    alpha
)

print(
    "beta =",
    beta
)


print()

print(
    "Expected alpha = 1/3 =",
    1.0 / 3.0
)

print(
    "Expected beta = 1/3 =",
    1.0 / 3.0
)


# ============================================================
# FINAL RESULT
# ============================================================

print()

print("=" * 75)

print(
    "FINAL RESULT"
)

print("=" * 75)

print()

print(
    "alpha = d / [d(m-1) + 2]"
)

print(
    "beta  = 1 / [d(m-1) + 2]"
)

print()

print(
    "PASS: The LLM scaling derivation is verified."
)
```
