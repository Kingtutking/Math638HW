# Part V: Relation to Otto Calculus

The porous medium equation is

$$
\partial_t\rho
=
\Delta(\rho^m),
\qquad
m>1.
$$

It can be interpreted as the Wasserstein gradient flow of the energy

$$
\mathcal E(\rho)
=
\frac{1}{m-1}
\int_{\mathbb R^d}
\rho(x)^m\,dx.
$$

The goal is to compute the first variation

$$
\frac{\delta\mathcal E}{\delta\rho}
$$

and verify that Otto's formula

$$
\partial_t\rho
=
\nabla\cdot
\left(
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
\right)
$$

reproduces the porous medium equation.

# 1. First Variation of the Energy

Consider a perturbation

$$
\rho_\varepsilon
=
\rho+\varepsilon\eta,
$$

where $\eta$ is a smooth test function.

Then

$$
\mathcal E(\rho_\varepsilon)
=
\frac{1}{m-1}
\int_{\mathbb R^d}
(\rho+\varepsilon\eta)^m\,dx.
$$

Differentiate with respect to $\varepsilon$:

$$
\frac{d}{d\varepsilon}
\mathcal E(\rho_\varepsilon)
=
\frac{m}{m-1}
\int_{\mathbb R^d}
(\rho+\varepsilon\eta)^{m-1}
\eta\,dx.
$$

Evaluating at

$$
\varepsilon=0
$$

gives

$$
\left.
\frac{d}{d\varepsilon}
\mathcal E(\rho+\varepsilon\eta)
\right|_{\varepsilon=0}
=
\frac{m}{m-1}
\int_{\mathbb R^d}
\rho^{m-1}\eta\,dx.
$$

By the definition of the first variation,

$$
\left.
\frac{d}{d\varepsilon}
\mathcal E(\rho+\varepsilon\eta)
\right|_{\varepsilon=0}
=
\int_{\mathbb R^d}
\frac{\delta\mathcal E}{\delta\rho}
\eta\,dx.
$$

Therefore,

$$
\boxed{
\frac{\delta\mathcal E}{\delta\rho}
=
\frac{m}{m-1}\rho^{m-1}
}.
$$

# 2. Substitution into Otto's Formula

Take the gradient:

$$
\nabla
\frac{\delta\mathcal E}{\delta\rho}
=
\nabla
\left(
\frac{m}{m-1}
\rho^{m-1}
\right).
$$

Using the chain rule,

$$
\nabla(\rho^{m-1})
=
(m-1)\rho^{m-2}\nabla\rho.
$$

Hence,

$$
\nabla
\frac{\delta\mathcal E}{\delta\rho}
=
m\rho^{m-2}\nabla\rho.
$$

Multiplying by $\rho$ gives

$$
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
=
m\rho^{m-1}\nabla\rho.
$$

But

$$
\nabla(\rho^m)
=
m\rho^{m-1}\nabla\rho.
$$

Therefore,

$$
\boxed{
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
=
\nabla(\rho^m)
}.
$$

Substituting this identity into Otto's formula gives

$$
\partial_t\rho
=
\nabla\cdot
\left(
\nabla(\rho^m)
\right).
$$

Thus,

$$
\boxed{
\partial_t\rho
=
\Delta(\rho^m)
}.
$$

Therefore, the Wasserstein gradient flow of

$$
\mathcal E(\rho)
=
\frac{1}{m-1}
\int\rho^m\,dx
$$

is exactly the porous medium equation.

# 3. Numerical Verification of the First Variation

To verify the first variation numerically, I used the positive periodic test density

$$
\rho(x)
=
1+0.2\cos x
$$

with

$$
m=2.5.
$$

A smooth perturbation $\eta$ was introduced and the directional derivative of the energy was approximated using the central difference

$$
\frac{
\mathcal E(\rho+\varepsilon\eta)
-
\mathcal E(\rho-\varepsilon\eta)
}
{2\varepsilon}.
$$

This was compared with the theoretical value

$$
\int
\frac{m}{m-1}
\rho^{m-1}
\eta\,dx.
$$

The numerical errors were:

| $\varepsilon$ | Absolute error |
| ---: | ---: |
| $10^{-2}$ | $2.47\times10^{-9}$ |
| $10^{-3}$ | $2.49\times10^{-11}$ |
| $10^{-4}$ | $9.11\times10^{-13}$ |
| $10^{-5}$ | $9.79\times10^{-12}$ |
| $10^{-6}$ | $3.46\times10^{-11}$ |

The error decreases rapidly at first and reaches approximately

$$
9.11\times10^{-13}
$$

for

$$
\varepsilon=10^{-4}.
$$

For still smaller $\varepsilon$, the error begins to increase slightly because floating-point cancellation becomes important.

This is the expected behavior of a finite-difference derivative calculation.

Therefore, the numerical experiment strongly supports

$$
\boxed{
\frac{\delta\mathcal E}{\delta\rho}
=
\frac{m}{m-1}\rho^{m-1}
}.
$$

# 4. Numerical Verification of Otto's Formula

The two quantities

$$
\nabla\cdot
\left(
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
\right)
$$

and

$$
\Delta(\rho^m)
$$

were computed independently using periodic finite differences.

![Otto formula compared with porous medium equation](hw41_part5_otto_vs_pme.png)

The two curves are visually almost indistinguishable.

The maximum absolute difference was

$$
9.45\times10^{-8},
$$

while the RMS difference was

$$
4.51\times10^{-8}.
$$

The maximum relative difference was only

$$
1.44\times10^{-7}.
$$

Therefore,

$$
\nabla\cdot
\left(
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
\right)
\approx
\Delta(\rho^m)
$$

to very high numerical accuracy.

# 5. Flux Identity Check

The intermediate identity

$$
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
=
\nabla(\rho^m)
$$

was also checked independently.

The maximum numerical difference between the two fluxes was

$$
1.02\times10^{-8}.
$$

This provides another verification of the derivation.

# Numerical Error

![Numerical error in Otto formula](hw41_part5_otto_error.png)

The error remains on the order of

$$
10^{-7}
$$

or smaller throughout the spatial domain.

The error is not exactly zero because the derivatives were approximated using finite differences.

A discrete finite-difference operator does not satisfy the continuous product rule and chain rule exactly.

Thus the observed error is a numerical discretization effect rather than a disagreement with Otto's formula.

# Interpretation

The energy

$$
\mathcal E(\rho)
=
\frac{1}{m-1}
\int\rho^m\,dx
$$

drives the density in Wasserstein space.

Its first variation is

$$
\frac{\delta\mathcal E}{\delta\rho}
=
\frac{m}{m-1}\rho^{m-1}.
$$

The corresponding flux is

$$
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
=
\nabla(\rho^m).
$$

Taking its divergence produces

$$
\partial_t\rho
=
\Delta(\rho^m).
$$

Thus the nonlinear diffusion in the porous medium equation can be interpreted as gradient descent of the energy $\mathcal E$ with respect to the Wasserstein geometry of probability densities.

# Conclusion

The first variation is

$$
\boxed{
\frac{\delta\mathcal E}{\delta\rho}
=
\frac{m}{m-1}\rho^{m-1}
}.
$$

Using Otto's formula,

$$
\partial_t\rho
=
\nabla\cdot
\left(
\rho
\nabla
\frac{\delta\mathcal E}{\delta\rho}
\right),
$$

we obtain

$$
\boxed{
\partial_t\rho
=
\Delta(\rho^m)
}.
$$

The numerical calculation verifies this identity with relative error approximately

$$
1.44\times10^{-7}.
$$

Therefore, both the analytic derivation and the numerical experiment confirm that the porous medium equation is the Wasserstein gradient flow of

$$
\boxed{
\mathcal E(\rho)
=
\frac{1}{m-1}
\int_{\mathbb R^d}
\rho^m\,dx
}.
$$
