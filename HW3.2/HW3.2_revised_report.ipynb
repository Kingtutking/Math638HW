## Homework 3.2: Benamou-Brenier in One Dimension
# Introduction
This homework studies the dynamic optimal transport interpretation of the Benamou-Brenier formula in one dimension.

Given two probability densities

$$
\rho_0
$$

and

$$
\rho_1,
$$

with cumulative distribution functions

$$
F_0
$$

and

$$
F_1,
$$

the monotone one-dimensional optimal transport map is

$$
T(x) = F_1^{-1}(F_0(x)).
$$

The corresponding Wasserstein displacement interpolation is

$$
X_t(x) = (1-t)x+tT(x), \qquad 0\le t\le1,
$$

and the intermediate probability density is

$$
\rho_t = (X_t)_\#\rho_0.
$$

The goals of the experiment were to:

- compute and verify the one-dimensional optimal transport map;

- visualize the displacement interpolation;

- estimate the squared Wasserstein distance;

- numerically evaluate the Benamou-Brenier action;

- compare compactly supported and Gaussian target distributions;

- evaluate the mathematical reliability of LLM-generated reasoning and code.

---

# Part I: LLM Plan and Preliminary Verification
## Prompt Given to the LLM
> Given two one-dimensional probability densities $\rho_0$ and $\rho_1$, how can I compute the optimal transport map using CDFs and quantile functions? How can I use this map to animate the Benamou-Brenier displacement interpolation?

## LLM Plan
The LLM proposed computing

$$
F_0(x) = \int_{-\infty}^{x} \rho_0(s)\\,ds
$$

and

$$
F_1(x) = \int_{-\infty}^{x} \rho_1(s)\\,ds.
$$

The optimal transport map is then

$$
T(x) = F_1^{-1}(F_0(x)).
$$

Once the map is known, particles move according to

$$
X_t(x) = (1-t)x+tT(x).
$$

The density is transported through

$$
\rho_t = (X_t)_\#\rho_0.
$$

The proposed numerical procedure was:

1. evaluate the source and target densities on a grid;

2. normalize the densities;

3. compute their CDFs;

4. construct the target quantile function;

5. calculate $T(x)$;

6. calculate $X_t(x)$;

7. reconstruct the intermediate densities;

8. plot the resulting transport.

---

## Preliminary Sanity Check
Before applying the method to the homework densities, I tested the implementation on two Gaussian distributions:

$$
\rho_0\approx N(-2,1)
$$

and

$$
\rho_1\approx N(2,1).
$$

Since the target is exactly a translation of the source by four units, the exact optimal map is

$$
T(x)=x+4.
$$

The numerical calculation gave an average displacement

$$
3.9999999642,
$$

with maximum interior error approximately

$$
2.93\times10^{-7}.
$$

The defining quantile identity

$$
F_1(T(x)) = F_0(x)
$$

had maximum numerical error

$$
1.11\times10^{-16}.
$$

The squared Wasserstein distance should be

$$
W_2^2=16,
$$

and the numerical calculation gave

$$
W_2^2 = 15.9999999028.
$$

Thus the CDF, quantile, and displacement-interpolation implementation was verified before being used for the main experiment.

# Part II: Half-Circle Density to Uniform Density
The initial density is

$$
\rho_0(x) = \frac{2}{\pi}\sqrt{1-x^2}, \qquad -1\le x\le1,
$$

and zero outside $[-1,1]$.

The target density is uniform:

$$
\rho_1(x) = \frac12, \qquad -1\le x\le1,
$$

and zero outside $[-1,1]$.

## Source and Target Densities
![Half-circle and uniform densities](hw32_part2_source_target_densities.png)

The half-circle density is concentrated more strongly around the origin, whereas the uniform target distributes the probability mass evenly across the interval.

---

## CDFs
For the source,

$$
F_0(x) = \frac12 + \frac{1}{\pi} \left( \arcsin x + x\sqrt{1-x^2} \right), \qquad -1\le x\le1.
$$

For the target,

$$
F_1(x) = \frac{x+1}{2}, \qquad -1\le x\le1.
$$

Thus

$$
F_1^{-1}(u) = 2u-1.
$$

![CDFs of half-circle and uniform densities](hw32_part2_cdfs.png)

The numerical and analytic CDFs agreed closely.

The maximum error for the numerical source CDF was approximately

$$
2.03\times10^{-6}.
$$

---

## Optimal Transport Map
The optimal map is

$$
T(x) = F_1^{-1}(F_0(x)).
$$

Therefore,

$$
T(x) = 2F_0(x)-1,
$$

which gives

$$
T(x) = \frac{2}{\pi} \left( \arcsin x + x\sqrt{1-x^2} \right) .
$$

The map satisfies

$$
T(-1)=-1,
$$

$$
T(0)=0,
$$

and

$$
T(1)=1.
$$

The numerical and analytic transport maps differed by at most approximately

$$
4.05\times10^{-6}.
$$

![Optimal transport map versus identity](hw32_part2_transport_map.png)

For

$$
x>0,
$$

we have

$$
T(x)>x,
$$

so particles move outward to the right.

For

$$
x<0,
$$

we have

$$
T(x)<x,
$$

so particles move outward to the left.

This is exactly what is required to flatten the initially peaked density.

---

## Displacement Interpolation
The Wasserstein geodesic is

$$
X_t(x) = (1-t)x+tT(x).
$$

Substituting the transport map,

$$
X_t(x) = (1-t)x + \frac{2t}{\pi} \left( \arcsin x + x\sqrt{1-x^2} \right).
$$

The maps were evaluated at

$$
t=0,\ 0.25,\ 0.5,\ 0.75,\ 1.
$$

![Displacement interpolation maps](hw32_part2_displacement_maps.png)

The interpolation begins with the identity map and continuously approaches the optimal transport map.

---

## Interpolated Densities
Differentiating the transport map gives

$$
T'(x) = \frac{4}{\pi}\sqrt{1-x^2}.
$$

Therefore,

$$
X_t'(x) = (1-t) + \frac{4t}{\pi}\sqrt{1-x^2}.
$$

Conservation of mass gives

$$
\rho_t(X_t(x)) X_t'(x) = \rho_0(x).
$$

Hence,

$$
\rho_t(X_t(x)) = \frac{ \frac{2}{\pi}\sqrt{1-x^2} }{ (1-t) + \frac{4t}{\pi}\sqrt{1-x^2} } .
$$

![Wasserstein displacement interpolation](hw32_part2_intermediate_densities.png)

At

$$
t=0,
$$

the density is the half-circle.

As $t$ increases, the central peak decreases and probability mass moves outward.

At

$$
t=1,
$$

the density becomes

$$
\rho_1(x)=\frac12.
$$

The total mass remained approximately one for every tested time.

---

## Velocity Along Particle Paths
The particle trajectory is

$$
X_t(x) = (1-t)x+tT(x).
$$

Differentiating with respect to time gives

$$
\dot X_t(x) = T(x)-x.
$$

Therefore,

$$
\dot X_t(x) = \frac{2}{\pi} \left( \arcsin x + x\sqrt{1-x^2} \right) -x .
$$

![Velocity along particle paths](hw32_part2_particle_velocity.png)

The velocity is negative on the left side and positive on the right side.

Thus particles move outward from the center.

The velocity is zero at

$$
x=-1,\quad 0,\quad 1.
$$

Each particle has constant **Lagrangian** velocity along its own displacement-interpolation path. The corresponding Eulerian velocity field is

$$
v_t(y)=T(x)-x \quad\text{where }y=X_t(x),
$$

so at a fixed spatial location $y$ the velocity field can depend on time through the inverse map $x=X_t^{-1}(y)$.

# Part III: Numerical Wasserstein Distance and Benamou-Brenier Action
The squared Wasserstein distance is

$$
W_2^2(\rho_0,\rho_1) = \int_{-1}^{1} |T(x)-x|^2 \rho_0(x)\\,dx.
$$

For this problem,

$$
W_2^2 = \int_{-1}^{1} \left[ \frac{2}{\pi} \left( \arcsin x+x\sqrt{1-x^2} \right) -x \right]^2 \frac{2}{\pi}\sqrt{1-x^2} \\,dx.
$$

The numerical estimate was approximately

$$
W_2^2 \approx 0.00692838 .
$$

As an additional check, the analytic benchmark is

$$
W_2^2 = \frac{7}{12} - \frac{256}{45\pi^2},
$$

which is also approximately

$$
0.00692838.
$$

---

## Benamou-Brenier Action
The dynamic action is

$$
\mathcal A = \int_0^1 \int_{\mathbb R} |v_t(y)|^2 \rho_t(y) \\,dy\\,dt.
$$

Along the particle path,

$$
v_t(X_t(x)) = T(x)-x.
$$

Using

$$
y=X_t(x)
$$

and conservation of probability mass,

$$
\rho_t(X_t(x)) X_t'(x) = \rho_0(x),
$$

we obtain

$$
\int_{\mathbb R} |v_t(y)|^2 \rho_t(y)\\,dy = \int_{-1}^{1} |T(x)-x|^2 \rho_0(x)\\,dx.
$$

Therefore,

$$
A(t) = W_2^2
$$

for every $t$ theoretically.

Numerically, the instantaneous action remained approximately

$$
6.92837\times10^{-3}.
$$

The difference

$$
A(t)-W_2^2
$$

remained only on the scale of

$$
10^{-9}.
$$

![Numerical error in Benamou-Brenier action](hw32_part3_action_error.png)

At the largest observed discrepancy, the relative error was only on the order of

$$
10^{-7}.
$$

Therefore, the numerical dynamic action satisfies

$$
\mathcal A \approx 0.00692838
$$

and agrees with

$$
W_2^2 \approx 0.00692838 .
$$

Thus,

$$
\mathcal A \approx W_2^2
$$

to very high numerical accuracy.

This shows that the numerical implementation is consistent with the Benamou-Brenier identity to high accuracy. The equality itself is an analytical theorem; the computation is a verification of the discretization and implementation.

# Part IV: Compact Support to Gaussian
The source remains

$$
\rho_0(x) = \frac{2}{\pi}\sqrt{1-x^2}, \qquad -1\le x\le1.
$$

The target is now the standard Gaussian:

$$
\rho_1(y) = \frac{1}{\sqrt{2\pi}} e^{-y^2/2}.
$$

Let

$$
\Phi(y)
$$

be its CDF.

The transport map is

$$
T(x) = \Phi^{-1}(F_0(x)) .
$$

## Compact Source and Gaussian Target
![Compact source and Gaussian target](hw32_part4_source_gaussian_target.png)

The source has support

$$
[-1,1],
$$

whereas the Gaussian has support

$$
\mathbb R.
$$

This difference produces a major change in the optimal transport map.

---

## Gaussian Transport Map
Since

$$
F_0(-1)=0
$$

and

$$
F_0(1)=1,
$$

we have

$$
T(-1) = \Phi^{-1}(0) = -\infty,
$$

and

$$
T(1) = \Phi^{-1}(1) = +\infty.
$$

Therefore,

$$
T(x)\to-\infty \quad \text{as }x\to-1^+
$$

and

$$
T(x)\to+\infty \quad \text{as }x\to1^-.
$$

![Optimal map from half-circle to Gaussian](hw32_part4_transport_map.png)

The numerical endpoint experiment gave:

| $\delta$ | $T(-1+\delta)$ | $T(1-\delta)$ |
| ---: | ---: | ---: |
| $10^{-1}$ | $-2.0815$ | $2.0815$ |
| $10^{-2}$ | $-3.2392$ | $3.2392$ |
| $10^{-3}$ | $-4.1196$ | $4.1196$ |
| $10^{-4}$ | $-4.8556$ | $4.8556$ |
| $10^{-6}$ | $-6.0802$ | $6.0802$ |
| $10^{-8}$ | $-7.1025$ | $7.1025$ |

This clearly demonstrates the divergence of the transport map.

---

## Interpolated Gaussian Densities
Differentiating

$$
\Phi(T(x)) = F_0(x)
$$

gives

$$
\phi(T(x))T'(x) = \rho_0(x),
$$

where

$$
\phi(y) = \frac{1}{\sqrt{2\pi}}e^{-y^2/2}.
$$

Hence,

$$
T'(x) = \frac{\rho_0(x)} {\phi(T(x))}.
$$

The intermediate density satisfies

$$
\rho_t(X_t(x)) = \frac{\rho_0(x)} { (1-t)+tT'(x) }.
$$

![Displacement interpolation from half-circle to Gaussian](hw32_part4_interpolated_densities.png)

The compact half-circle density gradually spreads into distributions with longer tails.

At

$$
t=1,
$$

the resulting density is the standard Gaussian.

The maximum final density error compared with the exact Gaussian was approximately

$$
5.55\times10^{-17}.
$$

Probability mass was also preserved to approximately

$$
10^{-6}.
$$

---

## Why the Mass Spreads to All of $\mathbb R$
Initially,

$$
\operatorname{supp}(\rho_0) = [-1,1].
$$

The Gaussian satisfies

$$
\rho_1(y)>0
$$

for every finite $y$.

Thus,

$$
\operatorname{supp}(\rho_1) = \mathbb R.
$$

The quantile map sends probability level $u$ to

$$
\Phi^{-1}(u).
$$

Since

$$
\Phi^{-1}(u)\to-\infty
$$

as

$$
u\to0,
$$

and

$$
\Phi^{-1}(u)\to+\infty
$$

as

$$
u\to1,
$$

source particles arbitrarily close to $-1$ and $1$ must move arbitrarily far into the tails.

Furthermore, for every fixed

$$
t>0,
$$

$$
X_t(x) = (1-t)x+tT(x).
$$

Since $T(x)$ diverges at the source endpoints,

$$
X_t(x)
$$

also has unbounded range for every positive time.

Thus the Wasserstein geodesic develops arbitrarily small tails across all of $\mathbb R$ immediately after $t=0$.

---

## Gaussian Truncation
A computer cannot represent the entire real line, so the Gaussian can be approximated on

$$
[-L,L].
$$

The omitted tail probability is

$$
P(|Z|>L) = 2(1-\Phi(L)).
$$

The numerical results were:

| $L$ | Gaussian mass outside $[-L,L]$ |
| ---: | ---: |
| $2$ | $4.55\times10^{-2}$ |
| $3$ | $2.70\times10^{-3}$ |
| $4$ | $6.33\times10^{-5}$ |
| $5$ | $5.73\times10^{-7}$ |
| $6$ | $1.97\times10^{-9}$ |

![Gaussian truncation error](hw32_part4_truncation_error.png)

The effect on the transport map is shown below.

![Truncated Gaussian transport maps](hw32_part4_truncated_transport_maps.png)

For

$$
L=2,
$$

the approximation is visibly poor.

At

$$
L=3,
$$

the approximation is substantially better.

At

$$
L=4,
$$

the omitted mass is only about

$$
6.33\times10^{-5},
$$

so the approximation is visually very good.

At

$$
L=5,
$$

only

$$
5.73\times10^{-7}
$$

of the Gaussian probability is omitted.

Therefore,

$$
L\approx4
$$

is a reasonable choice for visualization, while

$$
L\approx5
$$

provides a very accurate finite-domain approximation.

The source density was

$$
\rho_0(x) = \frac{2}{\pi}\sqrt{1-x^2}, \qquad -1\le x\le1.

Analytically,

$$

The error was therefore approximately

The optimal map was computed using

## 4. Compact Support
The source density is supported only on

$$
[-1,1].
$$

At the endpoints,

$$
F_0(-1)=0
$$

and

$$

$$

Therefore,

$$
T(-1) = \Phi^{-1}(0) = -\infty
$$

and

$$
T(1) = \Phi^{-1}(1) = +\infty.
$$

A numerical program cannot directly use these infinite values.

The LLM-generated code therefore excluded the exact endpoints from the particle grid.

This is a reasonable numerical choice, provided that it is understood as a computational approximation rather than a change in the mathematics.

The code separately examined points increasingly close to the endpoints.

For example,

Therefore, the compact support was handled correctly.

One minor weakness is that one final diagnostic line saying that endpoint divergence had been demonstrated was hard-coded as `True`.

## Where the LLM Could Be Mathematically Careless
Even though the code performed well, several aspects could have been misleading if they were accepted without inspection.

First, the program prints its own `PASS` statements.

A printed `PASS` statement does not independently prove that the test itself is mathematically valid.

Second, the code clips probability levels away from exactly

$$
0
$$

and

$$
1
$$

before applying the Gaussian quantile function.

This is necessary numerically, but it could hide the true mathematical endpoint divergence if the user did not examine the limiting behavior separately.

Third, finite-grid calculations only approximate the full continuous problem.

For example, mass conservation was accurate only to about

$$
10^{-6},
$$

not exact.

Fourth, a finite plot of the Gaussian cannot demonstrate that its support is all of

$$
\mathbb R.
$$

That conclusion comes from the mathematical form of the Gaussian density, not from the plot.

Finally, one endpoint-divergence diagnostic was hard-coded rather than computed automatically.

Therefore, the generated program still required mathematical inspection.

# Final Conclusion
The experiments numerically demonstrate the one-dimensional Benamou-Brenier picture of optimal transport.

For the half-circle to uniform problem, the optimal transport map is

$$
T(x) = \frac{2}{\pi} \left( \arcsin x+x\sqrt{1-x^2} \right).
$$

The displacement interpolation continuously spreads probability mass away from the center until the density becomes uniform.

The squared Wasserstein distance was approximately

$$
W_2^2 \approx 0.00692838,
$$

while the numerical Benamou-Brenier action was also approximately

$$
0.00692838.
$$

Their agreement numerically verifies

$$
W_2^2(\rho_0,\rho_1) = \int_0^1 \int_{\mathbb R} |v_t(y)|^2 \rho_t(y) \\,dy\\,dt
$$

along the Wasserstein geodesic.

When the target was replaced by a Gaussian, the optimal transport map became unbounded near the endpoints of the compact source interval.

This illustrates how optimal transport can move a compactly supported distribution into one with support over the entire real line.

Overall, the analytic formulas, numerical transport maps, interpolated densities, Wasserstein calculation, Benamou-Brenier action, mass-conservation checks, and Gaussian truncation study all agreed with the theoretical predictions.
