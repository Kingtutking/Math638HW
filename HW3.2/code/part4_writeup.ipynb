# Part IV: Compact Support to Gaussian

The source distribution remains the half-circle density

$$
\rho_0(x)
=
\frac{2}{\pi}\sqrt{1-x^2},
\qquad
-1\le x\le1.
$$

The new target distribution is the standard Gaussian

$$
\rho_1(y)
=
\frac{1}{\sqrt{2\pi}}
e^{-y^2/2},
\qquad
y\in\mathbb R.
$$

Let

$$
\Phi(y)
$$

denote the standard Gaussian CDF.

Then

$$
F_1(y)=\Phi(y).
$$

The source CDF from the previous section is

$$
F_0(x)
=
\frac12
+
\frac{1}{\pi}
\left(
\arcsin x
+
x\sqrt{1-x^2}
\right),
\qquad
-1\le x\le1.
$$

# 1. Optimal Transport Map

The one-dimensional optimal transport map is

$$
T(x)
=
F_1^{-1}(F_0(x)).
$$

Since the target CDF is $\Phi$,

$$
\boxed{
T(x)
=
\Phi^{-1}(F_0(x))
}.
$$

Unlike the uniform-target example, the Gaussian inverse CDF does not have an elementary closed form, so it was evaluated numerically.

The numerical map was verified to be nondecreasing.

The defining pushforward identity is

$$
\Phi(T(x))
=
F_0(x).
$$

The maximum numerical error was

$$
2.22\times10^{-16},
$$

which is essentially floating-point precision.

Therefore, the computed map accurately pushes the half-circle distribution to the Gaussian distribution.

![Optimal transport map from half-circle to Gaussian](hw32_part4_transport_map.png)

# Behavior Near $x=-1$ and $x=1$

At the endpoints of the source support,

$$
F_0(-1)=0
$$

and

$$
F_0(1)=1.
$$

Therefore,

$$
T(-1)
=
\Phi^{-1}(0)
=
-\infty,
$$

while

$$
T(1)
=
\Phi^{-1}(1)
=
+\infty.
$$

Thus,

$$
\boxed{
T(x)\to-\infty
\quad\text{as}\quad
x\to-1^+
}
$$

and

$$
\boxed{
T(x)\to+\infty
\quad\text{as}\quad
x\to1^-.
}
$$

This behavior was clearly observed numerically.

| $\delta$ | $T(-1+\delta)$ | $T(1-\delta)$ |
| ---: | ---: | ---: |
| $10^{-1}$ | $-2.0815$ | $2.0815$ |
| $10^{-2}$ | $-3.2392$ | $3.2392$ |
| $10^{-3}$ | $-4.1196$ | $4.1196$ |
| $10^{-4}$ | $-4.8556$ | $4.8556$ |
| $10^{-6}$ | $-6.0802$ | $6.0802$ |
| $10^{-8}$ | $-7.1025$ | $7.1025$ |

As

$$
\delta\to0,
$$

the magnitude of the transported position continues to increase.

This divergence is necessary because the source density has compact support

$$
[-1,1],
$$

while the Gaussian target has support

$$
\mathbb R.
$$

The exact endpoints themselves have zero probability mass, so the infinite endpoint values do not create a problem for the transport measure.

# Source and Target Densities

![Compact source and Gaussian target](hw32_part4_source_gaussian_target.png)

The source density vanishes outside

$$
[-1,1].
$$

In contrast, the Gaussian satisfies

$$
\rho_1(y)>0
$$

for every finite

$$
y\in\mathbb R.
$$

The target therefore has nonzero tails extending indefinitely in both directions.

# 2. Displacement Interpolation

The Wasserstein displacement interpolation is

$$
X_t(x)
=
(1-t)x+tT(x),
\qquad
0\le t\le1.
$$

At

$$
t=0,
$$

we recover

$$
X_0(x)=x.
$$

At

$$
t=1,
$$

we obtain

$$
X_1(x)=T(x).
$$

The numerical interpolation was evaluated at

$$
t=0,\ 0.25,\ 0.5,\ 0.75,\ 1.
$$

![Particle maps from compact support to Gaussian](hw32_part4_displacement_maps.png)

The curves near the endpoints rapidly spread outward as $t$ increases.

This is caused by the fact that

$$
T(x)\to\pm\infty
$$

near the source endpoints.

# Intermediate Densities

Differentiating

$$
\Phi(T(x))
=
F_0(x)
$$

gives

$$
\phi(T(x))T'(x)
=
\rho_0(x),
$$

where

$$
\phi(y)
=
\frac{1}{\sqrt{2\pi}}e^{-y^2/2}.
$$

Therefore,

$$
T'(x)
=
\frac{\rho_0(x)}
{\phi(T(x))}.
$$

The derivative of the displacement map is

$$
X_t'(x)
=
(1-t)+tT'(x).
$$

By conservation of probability mass,

$$
\rho_t(X_t(x))X_t'(x)
=
\rho_0(x).
$$

Therefore,

$$
\boxed{
\rho_t(X_t(x))
=
\frac{\rho_0(x)}
{(1-t)+tT'(x)}
}.
$$

Substituting the expression for $T'(x)$ gives

$$
\rho_t(X_t(x))
=
\frac{\rho_0(x)}
{
(1-t)
+
t\frac{\rho_0(x)}{\phi(T(x))}
}.
$$

The resulting densities are shown below.

![Displacement interpolation from half-circle to Gaussian](hw32_part4_interpolated_densities.png)

Initially, the density is confined to

$$
[-1,1].
$$

As time increases, the central peak decreases and the density develops increasingly long tails.

At

$$
t=1,
$$

the density becomes the standard Gaussian.

# Mass Conservation

The numerical masses of the interpolated densities were

$$
0.99999963
$$

at

$$
t=0,
$$

$$
1.00000613
$$

at

$$
t=0.25,
$$

$$
1.00000615
$$

at

$$
t=0.5,
$$

$$
1.00000616
$$

at

$$
t=0.75,
$$

and

$$
1.00000616
$$

at

$$
t=1.
$$

All values differ from $1$ by only a few parts in

$$
10^6.
$$

Thus,

$$
\int_{\mathbb R}\rho_t(y)\,dy
\approx1
$$

throughout the interpolation.

The small discrepancies are caused by finite-grid numerical quadrature.

# Final Gaussian Verification

At

$$
t=1,
$$

the computed density was directly compared with

$$
\phi(y)
=
\frac{1}{\sqrt{2\pi}}e^{-y^2/2}.
$$

The maximum difference was

$$
5.55\times10^{-17}.
$$

Therefore,

$$
\boxed{
\rho_{t=1}(y)
=
\rho_1(y)
}
$$

to essentially machine precision.

# 3. Why the Mass Must Spread to All of $\mathbb R$

Initially,

$$
\operatorname{supp}(\rho_0)
=
[-1,1].
$$

The Gaussian target satisfies

$$
\rho_1(y)>0
$$

for every

$$
y\in\mathbb R.
$$

Therefore,

$$
\operatorname{supp}(\rho_1)
=
\mathbb R.
$$

The quantile construction matches each cumulative probability level

$$
u\in(0,1)
$$

with the Gaussian quantile

$$
\Phi^{-1}(u).
$$

As

$$
u\to0,
$$

$$
\Phi^{-1}(u)\to-\infty,
$$

and as

$$
u\to1,
$$

$$
\Phi^{-1}(u)\to+\infty.
$$

Therefore, particles that begin arbitrarily close to

$$
x=-1
$$

and

$$
x=1
$$

must travel arbitrarily far into the Gaussian tails.

There is an even stronger conclusion.

For any fixed

$$
t>0,
$$

we have

$$
X_t(x)
=
(1-t)x+tT(x).
$$

Since

$$
T(x)\to\pm\infty
$$

near the source endpoints,

$$
X_t(x)\to\pm\infty
$$

for every

$$
t>0.
$$

Therefore, mathematically the interpolated density already has support extending through all of

$$
\mathbb R
$$

for every positive time.

At very small $t$, however, the far tails contain extremely little mass and may not be visible numerically.

# 4. Truncating the Gaussian to $[-L,L]$

A numerical computer cannot represent an infinite spatial interval.

Therefore, a natural approximation is to truncate the Gaussian to

$$
[-L,L].
$$

The Gaussian probability retained inside the interval is

$$
P(|Z|\le L)
=
\Phi(L)-\Phi(-L).
$$

The omitted probability is

$$
P(|Z|>L)
=
1-
\left[
\Phi(L)-\Phi(-L)
\right].
$$

Because the standard Gaussian is symmetric,

$$
P(|Z|>L)
=
2(1-\Phi(L)).
$$

The numerical truncation study gave approximately:

| $L$ | Mass inside $[-L,L]$ | Mass omitted |
| ---: | ---: | ---: |
| $2$ | $0.9544997$ | $4.5500\times10^{-2}$ |
| $3$ | $0.9973002$ | $2.6998\times10^{-3}$ |
| $4$ | $0.9999367$ | $6.3342\times10^{-5}$ |
| $5$ | $0.9999994$ | $5.7330\times10^{-7}$ |
| $6$ | $0.9999999980$ | $1.9732\times10^{-9}$ |

![Gaussian truncation error](hw32_part4_truncation_error.png)

The truncation error decreases extremely rapidly as $L$ increases.

# Effect on the Transport Map

For the Gaussian conditioned on

$$
[-L,L],
$$

the normalized CDF is

$$
F_L(y)
=
\frac{
\Phi(y)-\Phi(-L)
}{
\Phi(L)-\Phi(-L)
}.
$$

Therefore,

$$
F_L^{-1}(u)
=
\Phi^{-1}
\left(
\Phi(-L)
+
u
\left[
\Phi(L)-\Phi(-L)
\right]
\right).
$$

This gives a truncated transport map

$$
T_L(x)
=
F_L^{-1}(F_0(x)).
$$

The full and truncated maps are shown below.

![Effect of Gaussian truncation on the transport map](hw32_part4_truncated_transport_maps.png)

For

$$
L=2,
$$

the difference is visually substantial because about

$$
4.55\%
$$

of Gaussian probability is removed.

For

$$
L=3,
$$

the approximation is much better, but the tails are still visibly limited.

For

$$
L=4,
$$

the truncated map closely follows the full Gaussian map except extremely near the source endpoints.

For

$$
L=5,
$$

the two maps are nearly indistinguishable over almost the entire source probability mass.

# Quantitative Transport-Map Error

Restricting attention to the central

$$
99.8\%
$$

of probability,

$$
0.001\le F_0(x)\le0.999,
$$

the truncation experiment gave approximately:

| $L$ | Maximum transport-map error | Mean transport-map error |
| ---: | ---: | ---: |
| $2$ | $1.1068$ | $0.1135$ |
| $3$ | $0.2625$ | $0.01094$ |
| $4$ | $0.00923$ | $2.86\times10^{-4}$ |
| $5$ | $8.47\times10^{-5}$ | $2.60\times10^{-6}$ |
| $6$ | $2.92\times10^{-7}$ | $8.95\times10^{-9}$ |

Thus the approximation improves very rapidly with $L$.

# How Large Should $L$ Be?

The numerical results suggest that

$$
L=2
$$

is too small because a noticeable fraction of the Gaussian tails is missing.

At

$$
L=3,
$$

the approximation is reasonable but still loses approximately

$$
0.27\%
$$

of the total probability.

At

$$
L=4,
$$

the missing probability is only

$$
6.33\times10^{-5},
$$

or approximately

$$
0.0063\%.
$$

Thus

$$
\boxed{L\approx4}
$$

is sufficient for a visually reasonable approximation.

At

$$
L=5,
$$

only about

$$
5.73\times10^{-7}
$$

of the Gaussian probability is omitted.

Therefore,

$$
\boxed{L\approx5}
$$

provides a very accurate numerical approximation.

# Conclusion

The optimal map from the compact half-circle distribution to the standard Gaussian is

$$
\boxed{
T(x)
=
\Phi^{-1}(F_0(x))
}.
$$

Because

$$
F_0(x)\to0
$$

as

$$
x\to-1^+,
$$

and

$$
F_0(x)\to1
$$

as

$$
x\to1^-,
$$

the map satisfies

$$
T(x)\to-\infty
$$

and

$$
T(x)\to+\infty
$$

at the two endpoints.

This divergence is necessary to transport a compactly supported probability distribution to one with support on all of

$$
\mathbb R.
$$

The numerical displacement interpolation preserved total mass and reproduced the standard Gaussian at

$$
t=1
$$

with maximum error approximately

$$
5.55\times10^{-17}.
$$

The truncation study showed that

$$
L=4
$$

is already visually accurate, while

$$
L=5
$$

gives a very high-accuracy finite-domain approximation.

Thus the analytic behavior of the transport map, the numerical density evolution, and the Gaussian truncation experiment are all consistent with the one-dimensional optimal transport theory.
