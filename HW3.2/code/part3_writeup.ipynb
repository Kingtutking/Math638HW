# Numerical Comparison

The instantaneous kinetic action is

$$
A(t)
=
\int_{\mathbb R}
|v_t(y)|^2\rho_t(y)\,dy.
$$

For the Wasserstein displacement interpolation, theory predicts

$$
A(t)=W_2^2(\rho_0,\rho_1)
$$

for every

$$
0\le t\le1.
$$

The numerical plot appears to show a slight decrease in the action, but the vertical axis uses an offset of approximately

$$
6.92837\times10^{-3}
$$

and displays variations only on the scale of

$$
10^{-9}.
$$

Therefore, the apparent slope is extremely small.

The numerical variation is caused by finite-grid and trapezoidal quadrature error when integrating on the nonuniform transported grid

$$
y=X_t(x).
$$

Thus the calculation confirms that

$$
\int_{\mathbb R}
|v_t(y)|^2\rho_t(y)\,dy
$$

is constant in time up to numerical error.

Consequently,

$$
\int_0^1
\int_{\mathbb R}
|v_t(y)|^2\rho_t(y)\,dy\,dt
\approx
W_2^2(\rho_0,\rho_1).
$$

This numerically verifies the Benamou-Brenier identity for the computed Wasserstein geodesic.
