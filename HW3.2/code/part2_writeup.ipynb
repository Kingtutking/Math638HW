# Part II: Half-Circle Density to Uniform Density

The initial probability density is

$$
\rho_0(x)
=
\frac{2}{\pi}\sqrt{1-x^2},
\qquad
-1\le x\le1,
$$

and

$$
\rho_0(x)=0
$$

outside $[-1,1]$.

The target density is uniform on the same interval:

$$
\rho_1(x)
=
\frac12,
\qquad
-1\le x\le1,
$$

with

$$
\rho_1(x)=0
$$

outside $[-1,1]$.

# 1. CDFs of the Two Densities

For the source distribution,

$$
F_0(x)
=
\int_{-1}^{x}
\frac{2}{\pi}\sqrt{1-s^2}\,ds.
$$

Using

$$
\int \sqrt{1-s^2}\,ds
=
\frac12
\left(
s\sqrt{1-s^2}
+
\arcsin s
\right),
$$

we obtain

$$
F_0(x)
=
\frac12
+
\frac{1}{\pi}
\left(
\arcsin x
+
x\sqrt{1-x^2}
\right)
$$

for

$$
-1\le x\le1.
$$

Thus,

$$
F_0(x)
=
\begin{cases}
0, & x<-1,\\
\frac12+
\frac{1}{\pi}
\left(
\arcsin x+x\sqrt{1-x^2}
\right),
& -1\le x\le1,\\
1, & x>1.
\end{cases}
$$

For the uniform target density,

$$
F_1(x)
=
\frac{x+1}{2},
\qquad
-1\le x\le1.
$$

Therefore,

$$
F_1(x)
=
\begin{cases}
0, & x<-1,\\
\frac{x+1}{2}, & -1\le x\le1,\\
1, & x>1.
\end{cases}
$$

The numerical CDFs were also computed using cumulative trapezoidal integration.

The maximum error between the numerical and analytic source CDFs was approximately

$$
2.03\times10^{-6},
$$

while the numerical target CDF agreed with the exact linear CDF to numerical precision.

![CDFs of half-circle and uniform densities](hw32_part2_cdfs.png)

The two CDFs meet at

$$
x=0,
$$

where both have cumulative probability

$$
\frac12.
$$

# 2. Optimal Transport Map

The one-dimensional optimal transport map is

$$
T(x)
=
F_1^{-1}(F_0(x)).
$$

Since

$$
F_1(x)
=
\frac{x+1}{2},
$$

its inverse is

$$
F_1^{-1}(u)
=
2u-1.
$$

Therefore,

$$
T(x)
=
2F_0(x)-1.
$$

Substituting the source CDF gives

$$
T(x)
=
\frac{2}{\pi}
\left(
\arcsin x
+
x\sqrt{1-x^2}
\right).
$$

Thus,

$$
T(-1)=-1,
$$

$$
T(0)=0,
$$

and

$$
T(1)=1.
$$

The numerical transport map agreed with this exact formula with maximum error approximately

$$
4.05\times10^{-6}.
$$

The pushforward identity was also verified:

$$
F_1(T(x))
=
F_0(x).
$$

The maximum numerical error in this identity was approximately

$$
1.11\times10^{-16}.
$$

Thus the computed map correctly transports the half-circle distribution to the uniform distribution.

# 3. Transport Map Compared with the Identity

The transport map was plotted together with

$$
I(x)=x.
$$

![Optimal transport map versus identity](hw32_part2_transport_map.png)

For

$$
x>0,
$$

the graph satisfies

$$
T(x)>x.
$$

Therefore particles on the right half of the interval move farther to the right.

For

$$
x<0,
$$

we have

$$
T(x)<x.
$$

Thus particles on the left half move farther to the left.

This makes physical sense because the initial half-circle density contains too much mass near the center relative to the uniform target.

The optimal transport therefore pushes mass outward from the center toward the edges.

# Derivative of the Transport Map

Differentiating

$$
T(x)
=
\frac{2}{\pi}
\left(
\arcsin x
+
x\sqrt{1-x^2}
\right)
$$

gives

$$
T'(x)
=
\frac{4}{\pi}\sqrt{1-x^2}.
$$

Since

$$
T'(x)\ge0,
$$

the transport map is monotone increasing, as required for the one-dimensional optimal transport map.

The numerical derivative agreed with the analytic derivative to approximately

$$
3.02\times10^{-6}
$$

away from the endpoints.

# 4. Displacement Interpolation

The Wasserstein displacement interpolation is

$$
X_t(x)
=
(1-t)x+tT(x).
$$

Substituting the exact transport map gives

$$
X_t(x)
=
(1-t)x
+
\frac{2t}{\pi}
\left(
\arcsin x
+
x\sqrt{1-x^2}
\right).
$$

At

$$
t=0,
$$

we recover

$$
X_0(x)=x.
$$

At

$$
t=1,
$$

we obtain

$$
X_1(x)=T(x).
$$

The maps were evaluated at

$$
t=0,\ 0.25,\ 0.5,\ 0.75,\ 1.
$$

![Displacement interpolation maps](hw32_part2_displacement_maps.png)

The curves evolve continuously from the identity map to the optimal transport map.

Because every particle follows

$$
X_t(x)
=
x+t(T(x)-x),
$$

each individual particle moves along a straight path in time.

# 5. Intermediate Densities

The density at time $t$ is the pushforward

$$
\rho_t
=
(X_t)_\#\rho_0.
$$

Since the interpolation map is monotone,

$$
\rho_t(X_t(x))
=
\frac{\rho_0(x)}
{X_t'(x)}.
$$

Since

$$
X_t'(x)
=
(1-t)+tT'(x),
$$

we obtain

$$
X_t'(x)
=
(1-t)
+
\frac{4t}{\pi}\sqrt{1-x^2}.
$$

Therefore,

$$
\rho_t(X_t(x))
=
\frac{
\frac{2}{\pi}\sqrt{1-x^2}
}{
(1-t)
+
\frac{4t}{\pi}\sqrt{1-x^2}
}.
$$

The densities were plotted for

$$
t=0,\ 0.25,\ 0.5,\ 0.75,\ 1.
$$

![Wasserstein displacement interpolation](hw32_part2_intermediate_densities.png)

At

$$
t=0,
$$

the density is the original half-circle:

$$
\rho_t=\rho_0.
$$

As $t$ increases, the peak near the center gradually decreases while mass moves toward the edges.

At

$$
t=1,
$$

the density becomes

$$
\rho_1(x)=\frac12.
$$

Thus the displacement interpolation continuously transforms the curved half-circle density into the flat uniform density.

# Conservation of Probability Mass

The numerical integral of each intermediate density remained approximately equal to $1$.

For the tested values of $t$, the numerical masses were all within approximately

$$
10^{-6}
$$

of one.

Therefore,

$$
\int_{-1}^{1}\rho_t(x)\,dx
\approx1
$$

throughout the interpolation.

This verifies numerical conservation of probability mass.

# Endpoint Density Verification

At

$$
t=0,
$$

the numerical density agreed exactly with the initial half-circle density on the particle grid.

At

$$
t=1,
$$

the transformed density agreed with

$$
\rho_1(x)=\frac12
$$

to numerical precision.

At the exact endpoints $x=\pm1$, the change-of-variables expression at $t=1$ formally has the form

$$
\frac{0}{0},
$$

because both $\rho_0(x)$ and $T'(x)$ vanish there.

For this reason, the numerical density calculation used points just inside the interval.

This does not affect the probability density, since individual endpoints have zero measure.

# 6. Velocity Along Particle Paths

Differentiating

$$
X_t(x)
=
(1-t)x+tT(x)
$$

with respect to time gives

$$
\dot X_t(x)
=
T(x)-x.
$$

Therefore,

$$
\dot X_t(x)
=
\frac{2}{\pi}
\left(
\arcsin x
+
x\sqrt{1-x^2}
\right)
-
x.
$$

The velocity does not depend on time.

Thus each particle moves at constant velocity along its straight transport path.

![Velocity along particle paths](hw32_part2_particle_velocity.png)

The velocity is negative for

$$
-1<x<0,
$$

so particles on the left move leftward.

The velocity is positive for

$$
0<x<1,
$$

so particles on the right move rightward.

The velocity vanishes at

$$
x=-1,\quad x=0,\quad x=1.
$$

Numerically, the maximum velocity was approximately

$$
0.11542,
$$

and the minimum velocity was approximately

$$
-0.11542.
$$

The largest outward displacement occurs near

$$
x\approx\pm0.619.
$$

The velocity is antisymmetric:

$$
v(-x)=-v(x).
$$

This reflects the symmetry of both probability distributions.

# Interpretation

The half-circle density has maximum density at the center:

$$
\rho_0(0)
=
\frac{2}{\pi}
\approx
0.6366.
$$

The target density is only

$$
\rho_1(x)=0.5.
$$

Thus the source has excess mass near the center.

Near the endpoints, however,

$$
\rho_0(\pm1)=0,
$$

while the uniform target has density

$$
0.5.
$$

The optimal transport therefore redistributes probability mass outward.

The left half moves toward $-1$, while the right half moves toward $1$.

The transport is symmetric and monotone, so particles never cross.

# Numerical Verification Summary

The Python program verified:

- the source density has total mass approximately $1$;
- the uniform target has total mass exactly $1$ numerically;
- the numerical and analytic CDFs agree;
- the numerical and analytic transport maps agree;
- $T(x)$ is monotone;
- the identity

$$
F_1(T(x))=F_0(x)
$$

holds;

- the analytic derivative

$$
T'(x)
=
\frac{4}{\pi}\sqrt{1-x^2}
$$

agrees with numerical differentiation;

- every intermediate density preserves total probability mass;
- the $t=0$ density is the half-circle density;
- the $t=1$ density is the uniform density;
- the particle velocity is

$$
T(x)-x.
$$

# Conclusion

The one-dimensional optimal transport map from the half-circle density to the uniform density is

$$
T(x)
=
\frac{2}{\pi}
\left(
\arcsin x
+
x\sqrt{1-x^2}
\right).
$$

The corresponding Wasserstein geodesic is

$$
X_t(x)
=
(1-t)x+tT(x).
$$

The numerical results show probability mass moving continuously outward from the center of the interval.

The initially peaked half-circle density gradually becomes flatter until it reaches the uniform density at

$$
t=1.
$$

The particle velocity is

$$
\dot X_t(x)=T(x)-x,
$$

so every particle follows a straight line with constant velocity.

The analytic derivation, numerical CDF calculation, transport map, mass-conservation test, intermediate densities, and velocity plots all agree.
