# Homework 3.1: Bakeries, Cafés, and the Difference Between Monge and Kantorovich

# Introduction

This homework studies discrete optimal transport using a bakery-café model.

Bakeries represent sources of bread supply, cafés represent destinations with bread demand, and transportation has a cost determined by the distance between locations.

The main objective is to compare two formulations of optimal transport:

- Monge transport, where each bakery must send all of its supply to exactly one café;
- Kantorovich transport, where a bakery may divide its supply among multiple cafés.

An LLM was used to help generate the examples, formulate the optimization problems, and generate Python code. The mathematical claims and computational results were then independently checked.

# Part II: Transportation Cost Matrix

The first transportation cost is Euclidean distance:

$$
c_{ij}
=
\|x_i-y_j\|.
$$

For points

$$
x_i=(x_i^{(1)},x_i^{(2)})
$$

and

$$
y_j=(y_j^{(1)},y_j^{(2)}),
$$

this is

$$
c_{ij}
=
\sqrt{
(x_i^{(1)}-y_j^{(1)})^2
+
(x_i^{(2)}-y_j^{(2)})^2
}.
$$

The Euclidean cost matrix is

$$
C
=

\end{pmatrix}.
$$

For example,

$$
c_{11}
=
\|(1,1)-(2,3)\|.
$$

Therefore,

$$
c_{11}
=
\sqrt{(1-2)^2+(1-3)^2}

2.2361.
$$

=

# Part III: Kantorovich Transport

In the Kantorovich formulation, define

$$
\gamma_{ij}\ge0
$$

to be the amount of bread sent from Bakery $i$ to Café $j$.

The transport matrix is

$$
\gamma
=
\begin{pmatrix}
\gamma_{11}&\gamma_{12}&\gamma_{13}&\gamma_{14}\\
\gamma_{21}&\gamma_{22}&\gamma_{23}&\gamma_{24}\\
\gamma_{31}&\gamma_{32}&\gamma_{33}&\gamma_{34}
\end{pmatrix}.
$$

# Supply Constraints

Each bakery must ship exactly all of its supply:

$$
\sum_{j=1}^{4}\gamma_{ij}=a_i.
$$

Therefore,

$$
\gamma_{11}+\gamma_{12}+\gamma_{13}+\gamma_{14}=5,
$$

$$
\gamma_{21}+\gamma_{22}+\gamma_{23}+\gamma_{24}=6,
$$

and

$$
\gamma_{31}+\gamma_{32}+\gamma_{33}+\gamma_{34}=4.
$$

# Demand Constraints

Each café must receive exactly its demand:

$$
\sum_{i=1}^{3}\gamma_{ij}=b_j.
$$

Therefore,

$$
\gamma_{11}+\gamma_{21}+\gamma_{31}=3,
$$

$$
\gamma_{12}+\gamma_{22}+\gamma_{32}=4,
$$

$$
\gamma_{13}+\gamma_{23}+\gamma_{33}=5,
$$

and

$$
\gamma_{14}+\gamma_{24}+\gamma_{34}=3.
$$

# Objective

The Kantorovich problem is

$$
\min_{\gamma_{ij}\ge0}
\sum_{i=1}^{3}
\sum_{j=1}^{4}
c_{ij}\gamma_{ij}.
$$

This is a linear programming problem.

I solved it using `scipy.optimize.linprog`.

---

$$

}.
$$

This means:

- Bakery 2 sends $4$ units to Café 2 and $2$ units to Café 4;
- Bakery 3 sends $3$ units to Café 3 and $1$ unit to Café 4.

$$

$$

$$

$$

The program additionally verified nonnegativity and independently recomputed the objective value.

# Part IV: The Monge Problem

In the Monge formulation, every bakery must send all of its supply to exactly one café.

A Monge map has the form

$$
T:\{1,2,3\}\to\{1,2,3,4\}.
$$

The café constraints require

$$
\sum_{i:T(i)=j}a_i=b_j.
$$

The Monge cost would be

$$
J_M(T)
=
\sum_{i=1}^{3}a_i c_{i,T(i)}.
$$

# Number of Possible Maps

Each of the three bakeries has four possible destinations.

Therefore,

$$
4^3=64
$$

Monge maps are possible.

The Python program exhaustively searched all $64$ maps.

It found

$$
\boxed{0}
$$

admissible Monge maps.

Thus,

$$
\boxed{\text{the Monge problem is infeasible for this city}.}
$$

---

# Why No Monge Map Exists

There are

$$
N=3
$$

bakeries but

$$
M=4
$$

cafés.

Every café has positive demand.

Because one bakery can serve only one café under Monge transport, three bakeries can serve at most three distinct cafés.

But four cafés require positive mass.

Therefore, at least one café would necessarily receive zero bread.

Thus no Monge map can satisfy the demand constraints.

There is also a second obstruction.

The possible sums formed from the indivisible bakery supplies

$$
5,\ 6,\ 4
$$

are

$$
0,\ 4,\ 5,\ 6,\ 9,\ 10,\ 11,\ 15.
$$

However, two cafés require only

$$
3
$$

units.

No subset of whole bakery supplies sums to $3$.

Thus the specific supply and demand values also prevent a Monge solution.

---

# Comparison with Kantorovich

The Kantorovich solution was

$$
\gamma^*
=
\begin{pmatrix}
3&0&2&0\\
0&4&0&2\\
0&0&3&1
\end{pmatrix}.
$$

Each bakery splits its supply.

For example,

$$
5=3+2,
$$

$$
6=4+2,
$$

and

$$
4=3+1.
$$

This splitting is exactly what allows Kantorovich transport to satisfy all four cafés.

Therefore,

$$
\boxed{\text{Kantorovich is feasible}}
$$

while

$$
\boxed{\text{Monge is infeasible}}.
$$

---

# 1. Supply Constraints

For Kantorovich transport, the LLM correctly imposed

$$
\sum_jx_{ij}=a_i.
$$

Its code constructs one equality constraint for each bakery.

For Monge transport, it introduces binary variables

$$
z_{ij}\in\{0,1\},
$$

with

$$
\sum_jz_{ij}=1.
$$

The resulting shipment is

$$
x_{ij}=a_i z_{ij}.
$$

Thus every bakery sends all of its supply to exactly one café.

The supply constraints are correct.

---

# 2. Demand Constraints

For Kantorovich transport, the LLM correctly imposed

$$
\sum_ix_{ij}=b_j.
$$

For Monge transport, it correctly imposed

$$
\sum_i a_i z_{ij}=b_j.
$$

Thus the demand constraints are also mathematically correct.

There was one notation typo in the written explanation.

It wrote

$$
\sum_{i(i)=j}a_i=b_j,
$$

when it should have written

$$
\sum_{i:T(i)=j}a_i=b_j.
$$

The implementation itself used the correct condition.

---

# 3. Transport Map Versus Transport Plan

The LLM correctly distinguished a Monge transport map from a Kantorovich transport plan.

Monge uses

$$
T(i)=j,
$$

meaning that the entire source mass $a_i$ is assigned to one destination.

Kantorovich instead uses the matrix

$$
(x_{ij}),
$$

and multiple entries in one row may be positive.

Every feasible Monge map induces a Kantorovich plan through

$$
x_{ij}
=
\begin{cases}
a_i, & j=T(i),\\
0, & j\ne T(i).
\end{cases}
$$

However, not every Kantorovich plan comes from a Monge map.

Thus,

$$
\{\text{Monge-induced plans}\}
\subseteq
\{\text{Kantorovich plans}\}.
$$

The LLM handled this distinction correctly.

---

# 4. Does a Monge Map Always Exist?

The LLM correctly stated that equal total supply and demand do not guarantee a Monge solution.

Its second example used

$$
a=(5,5)
$$

and

$$
b=(6,4).
$$

Total supply and demand are both

$$
10,
$$

but no bakery can split its supply.

Therefore, the demands $6$ and $4$ cannot be formed from two indivisible masses of size $5$.

The LLM correctly reported that the Monge problem is infeasible.

Kantorovich can instead use

$$
X=
\begin{pmatrix}
5&0\\
1&4
\end{pmatrix},
$$

which has the correct row and column sums.

The transportation cost is

$$
5(2)+1(3)+4(4)
=
29.
$$

Thus this example is mathematically correct.

# Final Conclusion

The experiments demonstrate the central difference between Monge and Kantorovich optimal transport.

Monge transport requires a deterministic map:

$$
T:\{1,\ldots,N\}
\to
\{1,\ldots,M\},
$$

where each bakery sends its entire mass to one café.

Kantorovich transport instead allows a transport matrix

$$
\gamma=(\gamma_{ij}),
$$

where mass may be divided among multiple destinations.

Every Monge solution induces a Kantorovich plan, so whenever Monge is feasible,

$$
\boxed{
C_K^*
\le
C_M^*
}.
$$

In the $N>M$ example, a compatible Monge map existed and both formulations had the same optimal cost:

$$
\boxed{
C_M^*
=
C_K^*
=
16
}.
$$

In the $N<M$ example, no Monge map existed, while Kantorovich remained feasible through mass splitting:

$$
\boxed{
C_K^*
\approx
43.827724
}.
$$

Therefore, Kantorovich transport is more general than Monge transport.

The computational experiments also show why LLM-generated mathematics should be verified. The LLM's formulation and code were largely correct, but it initially failed to demonstrate one of the specific cases requested by the assignment.
