# Part IV: Compare the Cases $N>M$ and $N<M$

The goal of this section is to compare two discrete optimal transport problems:

- Case 1 has more bakeries than cafés, $N>M$.
- Case 2 has fewer bakeries than cafés, $N<M$.

For each case, I compare the Monge and Kantorovich formulations.

---

# Case 1: More Bakeries Than Cafés

For the first experiment, I chose

$$
N=4
$$

bakeries and

$$
M=3
$$

cafés.

Thus,

$$
N>M.
$$

The bakery supplies are

$$
a=(2,2,2,2),
$$

and the café demands are

$$
b=(4,2,2).
$$

The total supply is

$$
2+2+2+2=8,
$$

and the total demand is

$$
4+2+2=8.
$$

Therefore,

$$
\sum_{i=1}^{4}a_i
=
\sum_{j=1}^{3}b_j
=
8.
$$

The problem is balanced.

# Locations

The bakery locations are

$$
x_1=(0,0),
$$

$$
x_2=(0,6),
$$

$$
x_3=(8,0),
$$

and

$$
x_4=(8,6).
$$

The café locations are

$$
y_1=(0,3),
$$

$$
y_2=(7,0),
$$

and

$$
y_3=(7,6).
$$

# Cost Matrix

Using Euclidean distance,

$$
c_{ij}
=
\|x_i-y_j\|,
$$

the cost matrix is

$$
C
\approx
\begin{pmatrix}
3.0000 & 7.0000 & 9.2195\\
3.0000 & 9.2195 & 7.0000\\
8.5440 & 1.0000 & 6.0828\\
8.5440 & 6.0828 & 1.0000
\end{pmatrix}.
$$

---

# Monge Solution for Case 1

A Monge transport map is

$$
T:\{1,2,3,4\}
\rightarrow
\{1,2,3\}.
$$

Each of the four bakeries has three possible destinations.

Therefore, there are

$$
3^4=81
$$

possible Monge maps.

The exhaustive search found

$$
12
$$

admissible maps satisfying all café demands.

The minimum-cost map returned by the search was

$$
(0,0,1,2)
$$

using Python's zero-based indexing.

In mathematical indexing, this means

$$
T(1)=1,
$$

$$
T(2)=1,
$$

$$
T(3)=2,
$$

and

$$
T(4)=3.
$$

Thus:

- Bakery 1 sends its $2$ units to Café 1.
- Bakery 2 sends its $2$ units to Café 1.
- Bakery 3 sends its $2$ units to Café 2.
- Bakery 4 sends its $2$ units to Café 3.

The resulting café totals are

$$
2+2=4,
$$

$$
2,
$$

and

$$
2.
$$

Therefore, the demands

$$
(4,2,2)
$$

are satisfied exactly.

# Optimal Monge Cost

The cost is

$$
J_M
=
2c_{11}
+
2c_{21}
+
2c_{32}
+
2c_{43}.
$$

Using

$$
c_{11}=3,
\qquad
c_{21}=3,
\qquad
c_{32}=1,
\qquad
c_{43}=1,
$$

we obtain

$$
J_M
=
2(3)+2(3)+2(1)+2(1).
$$

Therefore,

$$
J_M
=
16.
$$

Thus,

$$
\boxed{J_M^*=16}.
$$

---

# Kantorovich Solution for Case 1

The optimal Kantorovich transport matrix was

$$
\gamma^*
=
\begin{pmatrix}
2&0&0\\
2&0&0\\
0&2&0\\
0&0&2
\end{pmatrix}.
$$

Its row sums are

$$
2,\ 2,\ 2,\ 2,
$$

which equal the bakery supplies.

Its column sums are

$$
4,\ 2,\ 2,
$$

which equal the café demands.

The optimal Kantorovich cost was

$$
J_K^*
=
16.
$$

Therefore,

$$
\boxed{
J_K^*
=
J_M^*
=
16
}.
$$

# No Mass Splitting in Case 1

The number of positive destinations for each bakery was

$$
(1,1,1,1).
$$

Thus every bakery sends bread to exactly one café.

The optimal Kantorovich plan is therefore induced by the same Monge map.

No mass splitting is required.

![Case 1 optimal transport plan](case1_N_greater_M_plan.png)

The figure shows two bakeries supplying Café 1, while the other two bakeries supply Café 2 and Café 3 individually.

---

# Why Monge Works in Case 1

The fact that

$$
N>M
$$

does not by itself guarantee that a Monge map exists.

The supply values must also be compatible with the café demands.

In this example, every bakery supplies exactly

$$
2
$$

units.

The café demands are

$$
4,\ 2,\ 2.
$$

These can be formed from whole bakery supplies:

$$
4=2+2,
$$

$$
2=2,
$$

and

$$
2=2.
$$

Therefore the no-splitting condition is compatible with the demand constraints.

This allows an admissible Monge map to exist.

# Case 2: Fewer Bakeries Than Cafés

For the second experiment, I returned to the original synthetic city from Parts I--III.

There are

$$
N=3
$$

bakeries and

$$
M=4
$$

cafés.

Thus,

$$
N<M.
$$

The bakery supplies are

$$
a=(5,6,4),
$$

and the café demands are

$$
b=(3,4,5,3).
$$

The total supply and demand are both

$$
15.
$$

Every café has positive demand.

---

# Monge Solution for Case 2

Each bakery must choose exactly one café.

Since there are only three bakeries, at most three distinct cafés can receive bread.

However, there are four cafés with positive demand.

Therefore, a Monge map cannot satisfy all four cafés.

The exhaustive search tested all

$$
4^3=64
$$

possible Monge maps.

It found

$$
0
$$

admissible maps.

Thus,

$$
\boxed{\text{no Monge solution exists in Case 2}.}
$$

There is therefore no minimum Monge cost.

---

# Kantorovich Solution for Case 2

The Kantorovich problem remains feasible because bakery supplies may be split.

The optimal plan was

$$
\gamma^*
=
\begin{pmatrix}
3&0&2&0\\
0&4&0&2\\
0&0&3&1
\end{pmatrix}.
$$

The row sums are

$$
3+2=5,
$$

$$
4+2=6,
$$

and

$$
3+1=4.
$$

Thus the bakery supplies are satisfied.

The column sums are

$$
3,
$$

$$
4,
$$

$$
2+3=5,
$$

and

$$
2+1=3.
$$

Thus the café demands are also satisfied.

The optimal Euclidean transportation cost was

$$
J_K^*
\approx
43.827724.
$$

Therefore,

$$
\boxed{
J_K^*
\approx
43.8277
}.
$$

---

# Mass Splitting in Case 2

The number of positive destinations for the three bakeries was

$$
(2,2,2).
$$

Therefore every bakery splits its supply between two cafés.

For Bakery 1,

$$
5=3+2.
$$

It sends $3$ units to Café 1 and $2$ units to Café 3.

For Bakery 2,

$$
6=4+2.
$$

It sends $4$ units to Café 2 and $2$ units to Café 4.

For Bakery 3,

$$
4=3+1.
$$

It sends $3$ units to Café 3 and $1$ unit to Café 4.

This splitting is forbidden in the Monge formulation but allowed in the Kantorovich formulation.

The computational results can be summarized as follows.

| Number of cafés | 3 | 4 |

| Kantorovich feasible? | Yes | Yes |
| Optimal Kantorovich cost | $16$ | $43.8277$ |

# Relationship Between Monge and Kantorovich

Every admissible Monge map can be converted into a Kantorovich plan.

Given a Monge map $T$, define

$$
\gamma_{ij}
=
\begin{cases}
a_i, & j=T(i),\\
0, & j\ne T(i).
\end{cases}
$$

Therefore, whenever a Monge solution exists, it is also a valid Kantorovich plan.

The Kantorovich formulation has a larger feasible set because it additionally allows

$$
\gamma_{ij}>0
$$

for several values of $j$ in the same row.

Therefore, whenever both problems are feasible,

$$
J_K^*
\le
J_M^*.
$$

In Case 1,

$$
J_K^*
=
J_M^*
=
16.
$$

Thus allowing splitting gives no improvement for that particular geometry.

In Case 2, there is no Monge feasible point at all, while Kantorovich still has an optimal solution.

---

# What $N>M$ Tells Us

The condition

$$
N>M
$$

makes a Monge solution possible in principle because there are enough bakery source points to serve all cafés.

However, it does not guarantee existence.

The indivisible bakery supplies must still be capable of producing the exact café demands.

In Case 1, the supplies

$$
2,2,2,2
$$

can be grouped as

$$
4,\ 2,\ 2.
$$

Therefore a Monge map exists.

---

# What $N<M$ Tells Us

Suppose

$$
N<M
$$

and every café has positive demand.

A Monge map assigns each of the $N$ bakeries to only one destination.

Therefore at most

$$
N
$$

cafés can receive positive mass.

But there are

$$
M>N
$$

cafés that all require positive mass.

Therefore at least one café must receive zero bread.

Hence no discrete Monge map can satisfy all demands.

Thus, under these assumptions,

$$
\boxed{
N<M
\text{ and all }b_j>0
\Longrightarrow
\text{no Monge map exists}.
}
$$

Kantorovich avoids this problem because one bakery may send positive mass to multiple cafés.

---

# Computational Verification

The Python program confirmed all of the theoretical claims.

For Case 1:

- $N=4>M=3$;
- all $81$ possible Monge maps were considered;
- $12$ maps were feasible;
- the optimal Monge cost was $16$;
- the optimal Kantorovich cost was also $16$;
- each bakery had exactly one positive destination.

For Case 2:

- $N=3<M=4$;
- all $64$ possible Monge maps were considered;
- no Monge map was feasible;
- the Kantorovich problem was feasible;
- the optimal Kantorovich cost was approximately $43.8277$;
- every bakery split its supply between two cafés.

Therefore the mathematical reasoning and computational results agree.

# Conclusion

The experiments demonstrate the main structural difference between Monge and Kantorovich transport.

In Case 1,

$$
N>M,
$$

and the supply masses can be grouped exactly to satisfy café demands.

A Monge solution exists, and the optimal Kantorovich solution happens to coincide with it:

$$
\boxed{
J_M^*=J_K^*=16.
}
$$

In Case 2,

$$
N<M,
$$

and every café has positive demand.

The Monge formulation is infeasible because three indivisible bakery sources cannot serve four positive-demand cafés.

The Kantorovich formulation remains feasible because bakery mass can be split:

$$
\boxed{
J_K^*\approx43.8277.
}
$$

Thus Kantorovich is a relaxation of the Monge problem.

It contains Monge-type transport plans when they are feasible, but it also permits split transport plans in situations where no deterministic Monge map exists.
