# Part V: LLM Critique — Monge and Kantorovich Optimal Transport

## 1. Prompt Given to the LLM

> Explain the difference between Monge and Kantorovich optimal transport using a discrete bakery–café problem. Provide the mathematical formulations for both approaches and generate Python code to solve them using the same transportation example.

## 2. Mathematical Explanation

Suppose there are $m$ bakeries and $n$ cafés.

Define:

- $a_i$: supply of bakery $i$.
- $b_j$: demand of café $j$.
- $c_{ij}$: transportation cost per unit from bakery $i$ to café $j$.

Assume that supply and demand are nonnegative and balanced:

$$
\sum_{i=1}^{m}a_i=\sum_{j=1}^{n}b_j.
$$

### 2.1 Monge Transport

In the discrete Monge formulation, each bakery must transport all its supply to exactly one café.

The decision variable is a deterministic map

$$
T:\{1,\ldots,m\}\rightarrow\{1,\ldots,n\}.
$$

The objective is

$$
\min_T\sum_{i=1}^{m}a_i c_{i,T(i)}.
$$

Subject to

$$
\sum_{i:T(i)=j}a_i=b_j
$$

for every café $j$.

The restriction means that a bakery cannot split its supply among different cafés.

Consequently, a feasible Monge transport map may not exist even when total supply equals total demand.

### 2.2 Kantorovich Transport

The Kantorovich formulation relaxes the deterministic assignment restriction.

Instead of mapping an entire bakery's supply to one café, we introduce

$$
x_{ij}\geq0,
$$

where $x_{ij}$ represents the amount transported from bakery $i$ to café $j$.

The optimization problem becomes

$$
\min_{x_{ij}}\sum_{i=1}^{m}\sum_{j=1}^{n}c_{ij}x_{ij}.
$$

Subject to

$$
\sum_{j=1}^{n}x_{ij}=a_i
$$

for each bakery $i$, and

$$
\sum_{i=1}^{m}x_{ij}=b_j
$$

for each café $j$.

The Kantorovich formulation allows supply splitting and is a linear programming problem.

For finite nonnegative supplies and demands with equal total mass, a feasible Kantorovich plan always exists.

### 2.3 Relationship Between the Two Formulations

Every feasible Monge map generates a Kantorovich transport plan through

$$
x_{ij}=
\begin{cases}
a_i, & T(i)=j,\\
0, & T(i)\neq j.
\end{cases}
$$

However, not every Kantorovich plan corresponds to a Monge map.

Therefore, if a feasible Monge solution exists,

$$
C_{\mathrm{Kantorovich}}^*
\leq
C_{\mathrm{Monge}}^*.
$$

The Kantorovich formulation is a relaxation of the Monge formulation.

## 3. Numerical Example

Consider three bakeries with supplies

$$
a=(2,3,5)
$$

and two cafés with demands

$$
b=(5,5).
$$

The transportation cost matrix is

$$
C=
\begin{pmatrix}
2&6\\
3&4\\
6&1
\end{pmatrix}.
$$

The rows represent bakeries and the columns represent cafés.

One feasible Monge assignment is

$$
T(1)=1,\qquad T(2)=1,\qquad T(3)=2.
$$

The corresponding transport plan is

$$
X=
\begin{pmatrix}
2&0\\
3&0\\
0&5
\end{pmatrix}.
$$

The total transportation cost is

$$
2(2)+3(3)+5(1)=18.
$$

In this example, the plan is optimal for both formulations, giving

$$
C_{\mathrm{Monge}}^*
=
C_{\mathrm{Kantorovich}}^*
=
18.
$$

This equality does not imply that the formulations are equivalent.

## 4. Example Where Monge Is Infeasible

Consider two bakeries with supplies

$$
a=(5,5)
$$

and two cafés with demands

$$
b=(6,4).
$$

Total supply and demand are equal:

$$
5+5=6+4=10.
$$

However, Monge transport requires each bakery to send all five units to exactly one café.

Neither café can receive exactly six or four units through such assignments.

Therefore, no feasible Monge map exists.

Kantorovich transport allows the plan

$$
X=
\begin{pmatrix}
5&0\\
1&4
\end{pmatrix}.
$$

The first bakery sends five units to café 1, while the second bakery sends one unit to café 1 and four units to café 2.

The row sums are

$$
(5,5),
$$

and the column sums are

$$
(6,4).
$$

Thus, the plan satisfies all supply and demand constraints.

This demonstrates why Kantorovich transport is more general than Monge transport.

## 6. Critique of the LLM's Explanation

The LLM correctly identifies that Monge transport assigns each source to exactly one destination, whereas Kantorovich transport allows the source mass to be split.

The main points requiring careful interpretation are:

1. **Feasibility:** Equal total supply and demand guarantee a feasible Kantorovich transport plan, but they do not guarantee a feasible Monge map.

2. **Optimization:** The Monge problem uses deterministic assignments. Its discrete formulation generally requires combinatorial optimization, whereas Kantorovich transport is a linear program.

3. **Optimal costs:** Since every feasible Monge solution is also a feasible Kantorovich solution, the Kantorovich optimal cost cannot exceed the Monge optimal cost.

4. **Interpretation:** Kantorovich does not require bakeries to split their supply. It merely permits splitting when necessary or beneficial.

The first numerical example gives equal optimal costs, so it does not reveal the difference in feasibility. The second example provides a clearer demonstration because Kantorovich has a feasible solution while Monge does not.

## 7. Conclusion

The fundamental distinction between Monge and Kantorovich optimal transport is whether the mass at a source may be divided among multiple destinations.

Monge transport requires a deterministic mapping, while Kantorovich transport permits a more general transport plan.

Mathematically,

$$
\{\text{Monge-induced feasible plans}\}
\subseteq
\{\text{Kantorovich feasible plans}\}.
$$

Consequently,

$$
C_{\mathrm{Kantorovich}}^*
\leq
C_{\mathrm{Monge}}^*
$$

whenever the Monge problem is feasible.

Kantorovich's formulation is particularly useful because it converts the discrete transportation problem into a linear programming problem and guarantees feasibility when nonnegative total supply equals total demand.
