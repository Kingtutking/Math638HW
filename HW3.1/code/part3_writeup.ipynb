# Part III: The Monge Problem

In the Monge formulation, every bakery must send all of its bread to exactly one café.

A Monge transport rule is a map

$$
T:\{1,\ldots,N\}\to\{1,\ldots,M\}.
$$

If

$$
T(i)=j,
$$

then all of Bakery $i$'s supply is transported to Café $j$.

Unlike the Kantorovich formulation, a bakery is not allowed to divide its supply among several cafés.

# Monge Demand Constraint

For each café $j$, the total supply of all bakeries assigned to it must equal its demand:

$$
\sum_{i:T(i)=j}a_i=b_j.
$$

For the synthetic city,

$$
a=(5,6,4)
$$

and

$$
b=(3,4,5,3).
$$

Thus the three indivisible bakery supplies

$$
5,\ 6,\ 4
$$

must somehow be assigned to the four cafés so that their received amounts are exactly

$$
3,\ 4,\ 5,\ 3.
$$

# Number of Possible Monge Maps

There are

$$
N=3
$$

bakeries and

$$
M=4
$$

cafés.

Each bakery can be assigned independently to one of the four cafés.

Therefore, the total number of possible maps is

$$
M^N
=
4^3
=
64.
$$

Because this is a small finite number, I checked every possible Monge map computationally.

# Structural Nonexistence Argument

Before running the exhaustive search, there is already a simple reason to expect that no Monge map exists.

We have

$$
N=3<M=4.
$$

There are only three bakeries, and each bakery can serve only one café.

Therefore, at most three distinct cafés can receive bread.

However, all four cafés have positive demand:

$$
b_1=3,
$$

$$
b_2=4,
$$

$$
b_3=5,
$$

and

$$
b_4=3.
$$

Thus all four cafés must receive positive amounts of bread.

Three bakeries cannot supply four positive-demand cafés when each bakery is restricted to exactly one destination.

Therefore, by the pigeonhole principle, no admissible Monge map can exist.

# Additional Supply-Size Obstruction

There is also another way to see the difficulty.

Because bakery supplies are indivisible under the Monge model, a café can only receive sums of whole bakery supplies.

The possible subset sums of

$$
5,\ 6,\ 4
$$

are

$$
0,\ 4,\ 5,\ 6,\ 9,\ 10,\ 11,\ 15.
$$

The café demands are

$$
3,\ 4,\ 5,\ 3.
$$

Demand $4$ can be produced by assigning Bakery 3 alone:

$$
4.
$$

Demand $5$ can be produced by assigning Bakery 1 alone:

$$
5.
$$

However, demand $3$ cannot be produced from any combination of whole bakery supplies.

Since Café 1 and Café 4 both require

$$
3
$$

units, their demands cannot be satisfied by any collection of indivisible bakery supplies.

Thus even apart from the fact that

$$
N<M,
$$

the actual supply sizes also prevent a Monge solution.

# Exhaustive Computational Search

The Python program enumerated all

$$
64
$$

possible maps

$$
T:\{1,2,3\}\to\{1,2,3,4\}.
$$

For every candidate map, the program calculated how much bread each café would receive and compared it with

$$
(3,4,5,3).
$$

The result was

$$
\text{Number of admissible Monge maps}=0.
$$

Therefore,

$$
\boxed{\text{no admissible Monge map exists}.}
$$

The computational search agrees exactly with the structural argument.

# Monge Cost

If an admissible Monge map existed, its transportation cost would be

$$
J(T)
=
\sum_{i=1}^{N}
a_i c_{i,T(i)}.
$$

For this example,

$$
J(T)
=
5c_{1,T(1)}
+
6c_{2,T(2)}
+
4c_{3,T(3)}.
$$

However, a transport cost can only be minimized over admissible maps.

Because the feasible set is empty, there is no minimum-cost Monge map for this synthetic city.

Thus,

$$
\boxed{\text{the discrete Monge problem is infeasible}.}
$$

# Comparison with the Kantorovich Result

This result contrasts directly with Part II.

The Kantorovich problem was feasible and produced the optimal transport plan

$$
\gamma^*
=
\begin{pmatrix}
3&0&2&0\\
0&4&0&2\\
0&0&3&1
\end{pmatrix}.
$$

In this plan, Bakery 1 splits its supply:

$$
5=3+2.
$$

Bakery 2 also splits its supply:

$$
6=4+2.
$$

Bakery 3 splits its supply as well:

$$
4=3+1.
$$

These splits allow all four cafés to receive exactly their required demands:

$$
3,\ 4,\ 5,\ 3.
$$

This is allowed in the Kantorovich formulation but forbidden in the Monge formulation.

Therefore,

$$
\boxed{
\text{Kantorovich is feasible, while Monge is infeasible.}
}
$$

# Computational Verification

The exhaustive-search code confirmed:

- all $64$ possible Monge maps were checked;
- zero admissible maps were found;
- the computational result agrees with the $N<M$ structural argument;
- the demands of size $3$ cannot be formed from the indivisible bakery supplies.

Therefore, the mathematical reasoning and computational verification agree.

# Conclusion

For the current synthetic city,

$$
N=3,
\qquad
M=4,
$$

with positive demand at every café.

A Monge map can send each bakery to only one café, so three bakeries can directly serve at most three distinct cafés.

Since four cafés require positive demand, no admissible Monge map exists.

The exhaustive search over all

$$
4^3=64
$$

possible maps confirmed this result.

Thus,

$$
\boxed{\text{Monge: infeasible}}
$$

while the Kantorovich formulation remains feasible because it allows mass splitting.

This demonstrates the central difference between the two formulations: the Monge model requires a deterministic one-destination transport map, while the Kantorovich model allows a source's mass to be divided among several destinations.
