```python
# ============================================================
# HOMEWORK 3.1 - PART III
# MONGE TRANSPORT MAP SEARCH
# STANDALONE VERIFICATION
# ============================================================

import numpy as np
import pandas as pd

from itertools import product


# ============================================================
# DATA FROM PART I
# ============================================================

bakery_locations = np.array([
    [1.0, 1.0],
    [7.0, 1.0],
    [4.0, 7.0]
])


bakery_supply = np.array([
    5.0,
    6.0,
    4.0
])


cafe_locations = np.array([
    [2.0, 3.0],
    [6.0, 3.0],
    [3.0, 6.0],
    [7.0, 6.0]
])


cafe_demand = np.array([
    3.0,
    4.0,
    5.0,
    3.0
])


N = len(
    bakery_supply
)

M = len(
    cafe_demand
)


# ============================================================
# COST MATRICES
# ============================================================

difference = (

    bakery_locations[:, None, :]

    -

    cafe_locations[None, :, :]
)


euclidean_cost = np.sqrt(

    np.sum(
        difference**2,
        axis=2
    )
)


squared_cost = np.sum(
    difference**2,
    axis=2
)


# ============================================================
# BASIC CHECKS
# ============================================================

assert np.isclose(
    np.sum(bakery_supply),
    np.sum(cafe_demand)
)


assert np.all(
    bakery_supply > 0
)


assert np.all(
    cafe_demand > 0
)


print("=" * 70)

print(
    "MONGE TRANSPORT MAP SEARCH"
)

print("=" * 70)

print(
    "Number of bakeries N =",
    N
)

print(
    "Number of cafes M =",
    M
)

print(
    "Bakery supplies =",
    bakery_supply
)

print(
    "Cafe demands =",
    cafe_demand
)

print(
    "Total supply =",
    np.sum(bakery_supply)
)

print(
    "Total demand =",
    np.sum(cafe_demand)
)


# ============================================================
# STRUCTURAL CHECK
# ============================================================

print()

print("=" * 70)

print(
    "STRUCTURAL FEASIBILITY CHECK"
)

print("=" * 70)


if (
    N < M
    and
    np.all(cafe_demand > 0)
):

    print(
        "N < M and every cafe has positive demand."
    )

    print(
        "Each bakery may serve only one cafe."
    )

    print(
        "Therefore at most",
        N,
        "cafes can receive bread."
    )

    print(
        "But",
        M,
        "cafes require positive demand."
    )

    print()

    print(
        "PREDICTION: No admissible Monge map exists."
    )


# ============================================================
# ENUMERATE ALL POSSIBLE MONGE MAPS
#
# Python uses cafe indices:
#
# 0 = Cafe 1
# 1 = Cafe 2
# 2 = Cafe 3
# 3 = Cafe 4
#
# A map such as:
#
# (0, 1, 2)
#
# means:
#
# Bakery 1 -> Cafe 1
# Bakery 2 -> Cafe 2
# Bakery 3 -> Cafe 3
# ============================================================

all_maps = list(
    product(
        range(M),
        repeat=N
    )
)


print()

print("=" * 70)

print(
    "EXHAUSTIVE SEARCH"
)

print("=" * 70)

print(
    "Total possible Monge maps =",
    len(all_maps)
)


assert len(all_maps) == M**N


admissible_maps = []


for T in all_maps:

    # Amount received by each cafe
    received = np.zeros(
        M
    )


    # Send each bakery's ENTIRE supply
    # to exactly one cafe
    for i in range(N):

        cafe_index = T[i]

        received[
            cafe_index
        ] += bakery_supply[i]


    # Check whether all cafe demands
    # are exactly satisfied
    admissible = np.allclose(

        received,

        cafe_demand
    )


    if admissible:

        # Euclidean Monge cost
        euclidean_total = sum(

            bakery_supply[i]
            *
            euclidean_cost[
                i,
                T[i]
            ]

            for i in range(N)
        )


        # Squared Euclidean Monge cost
        squared_total = sum(

            bakery_supply[i]
            *
            squared_cost[
                i,
                T[i]
            ]

            for i in range(N)
        )


        admissible_maps.append({

            "Map":
                T,

            "Received":
                received.copy(),

            "Euclidean_Cost":
                euclidean_total,

            "Squared_Cost":
                squared_total

        })


# ============================================================
# RESULTS
# ============================================================

print()

print(
    "Number of admissible Monge maps =",
    len(admissible_maps)
)


if len(admissible_maps) == 0:

    print()

    print(
        "RESULT: No admissible Monge map exists."
    )


else:

    print()

    print(
        "Admissible maps found:"
    )

    for item in admissible_maps:

        print(
            item
        )


# ============================================================
# AUTOMATIC VERIFICATION
# ============================================================

print()

print("=" * 70)

print(
    "AUTOMATIC VERIFICATION"
)

print("=" * 70)


assert len(
    admissible_maps
) == 0


print(
    "PASS: All",
    M**N,
    "possible Monge maps were checked."
)

print(
    "PASS: Zero admissible Monge maps were found."
)

print(
    "PASS: The computational result agrees "
    "with the N < M structural argument."
)


# ============================================================
# ADDITIONAL CHECK:
# POSSIBLE SUBSET SUMS OF BAKERY SUPPLIES
# ============================================================

possible_sums = set()


for choice in product(
    [0, 1],
    repeat=N
):

    total = sum(

        bakery_supply[i]

        for i in range(N)

        if choice[i] == 1
    )


    possible_sums.add(
        total
    )


possible_sums = sorted(
    possible_sums
)


print()

print("=" * 70)

print(
    "POSSIBLE SUMS OF INDIVISIBLE BAKERY SUPPLIES"
)

print("=" * 70)

print(
    possible_sums
)


for demand in cafe_demand:

    print(
        f"Demand {demand:.0f} can be formed "
        f"from whole bakery supplies:",
        demand in possible_sums
    )


# ============================================================
# FINAL CONCLUSION
# ============================================================

print()

print("=" * 70)

print(
    "FINAL CONCLUSION"
)

print("=" * 70)

print(
    "The Monge problem is infeasible "
    "for this synthetic city."
)

print(
    "No minimum-cost Monge map exists "
    "because there is no admissible map."
)
```
