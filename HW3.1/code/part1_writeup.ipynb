# Homework 3.1: Bakeries, Cafés, and the Difference Between Monge and Kantorovich

# Part I: Generate a Discrete Transport Problem

The goal is to construct a small synthetic city containing bakeries that supply bread and cafés that demand bread.

Each bakery has:

- a location $x_i\in\mathbb R^2$;
- a positive supply $a_i$.

Each café has:

- a location $y_j\in\mathbb R^2$;
- a positive demand $b_j$.

The total supply must equal the total demand:

$$
\sum_{i=1}^{N}a_i
=
\sum_{j=1}^{M}b_j.
$$

I deliberately choose a different number of bakeries and cafés:

$$
N=3,
\qquad
M=4.
$$

This will make the later comparison between the Monge and Kantorovich formulations more interesting.

# LLM Prompt

I asked the LLM:

> Generate a small discrete optimal transport example representing bakeries and cafés in a two-dimensional city. Use 3 bakeries and 4 cafés, assign positive supplies and demands, and make sure total supply equals total demand. Choose simple coordinates so that Euclidean transportation costs can be calculated and visualized easily.

# Generated Bakery Locations and Supplies

The three bakeries are

$$
x_1=(1,1),
$$

$$
x_2=(7,1),
$$

and

$$
x_3=(4,7).
$$

Their supplies are

$$
a_1=5,
\qquad
a_2=6,
\qquad
a_3=4.
$$

Thus the total supply is

$$
\sum_{i=1}^{3}a_i
=
5+6+4
=
15.
$$

# Generated Café Locations and Demands

The four cafés are

$$
y_1=(2,3),
$$

$$
y_2=(6,3),
$$

$$
y_3=(3,6),
$$

and

$$
y_4=(7,6).
$$

Their demands are

$$
b_1=3,
\qquad
b_2=4,
\qquad
b_3=5,
\qquad
b_4=3.
$$

Thus the total demand is

$$
\sum_{j=1}^{4}b_j
=
3+4+5+3
=
15.
$$

Therefore,

$$
\sum_{i=1}^{3}a_i
=
\sum_{j=1}^{4}b_j
=
15.
$$

The transport problem is balanced.

# Transportation Cost

The first transportation cost is ordinary Euclidean distance:

$$
c_{ij}
=
\|x_i-y_j\|.
$$

For coordinates

$$
x_i=(x_i^{(1)},x_i^{(2)})
$$

and

$$
y_j=(y_j^{(1)},y_j^{(2)}),
$$

the cost is

$$
c_{ij}
=
\sqrt{
\left(x_i^{(1)}-y_j^{(1)}\right)^2
+
\left(x_i^{(2)}-y_j^{(2)}\right)^2
}.
$$

The second cost considered is squared Euclidean distance:

$$
c_{ij}^{(2)}
=
\|x_i-y_j\|^2.
$$

Thus,

$$
c_{ij}^{(2)}
=
\left(x_i^{(1)}-y_j^{(1)}\right)^2
+
\left(x_i^{(2)}-y_j^{(2)}\right)^2.
$$

# Euclidean Cost Matrix

Using

$$
C=(c_{ij})_{3\times4},
$$

the Euclidean transportation-cost matrix is approximately

$$
C
=
\begin{pmatrix}
2.2361 & 5.3852 & 5.3852 & 7.8102\\
5.3852 & 2.2361 & 6.4031 & 5.0000\\
4.4721 & 4.4721 & 1.4142 & 3.1623
\end{pmatrix}.
$$

For example,

$$
c_{11}
=
\|(1,1)-(2,3)\|
=
\sqrt{(1-2)^2+(1-3)^2}
=
\sqrt{5}
\approx
2.2361.
$$

# Squared Euclidean Cost Matrix

The squared-distance cost matrix is

$$
C^{(2)}
=
\begin{pmatrix}
5 & 29 & 29 & 61\\
29 & 5 & 41 & 25\\
20 & 20 & 2 & 10
\end{pmatrix}.
$$

For example,

$$
c_{11}^{(2)}
=
(1-2)^2+(1-3)^2
=
5.
$$

As a consistency check,

$$
c_{ij}^{(2)}
=
(c_{ij})^2.
$$
