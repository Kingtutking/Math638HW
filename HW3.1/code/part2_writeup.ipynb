# Part II: The Kantorovich Problem

In the discrete Kantorovich formulation, the unknown is a transport matrix

$$
\gamma=(\gamma_{ij}),
$$

where

$$
\gamma_{ij}\ge0
$$

is the amount of bread transported from bakery $i$ to café $j$.

For this problem there are

$$
N=3
$$

bakeries and

$$
M=4
$$

cafés.

Therefore,

$$
\gamma
=
\begin{pmatrix}
\gamma_{11} & \gamma_{12} & \gamma_{13} & \gamma_{14}\\
\gamma_{21} & \gamma_{22} & \gamma_{23} & \gamma_{24}\\
\gamma_{31} & \gamma_{32} & \gamma_{33} & \gamma_{34}
\end{pmatrix}.
$$

The Kantorovich formulation differs from the Monge formulation because a bakery is allowed to split its supply among multiple cafés.

# Supply Constraints

The bakery supplies are

$$
a=(5,6,4).
$$

Therefore,

$$
\gamma_{11}+\gamma_{12}+\gamma_{13}+\gamma_{14}=5,
$$

$$
\gamma_{21}+\gamma_{22}+\gamma_{23}+\gamma_{24}=6,
$$

and

$$
\gamma_{31}+\gamma_{32}+\gamma_{33}+\gamma_{34}=4.
$$

# Demand Constraints

The café demands are

$$
b=(3,4,5,3).
$$

Therefore,

$$
\gamma_{11}+\gamma_{21}+\gamma_{31}=3,
$$

$$
\gamma_{12}+\gamma_{22}+\gamma_{32}=4,
$$

$$
\gamma_{13}+\gamma_{23}+\gamma_{33}=5,
$$

and

$$
\gamma_{14}+\gamma_{24}+\gamma_{34}=3.
$$

The transport quantities must also satisfy

$$
\gamma_{ij}\ge0.
$$

# Kantorovich Optimization Problem

For a transportation cost matrix

$$
C=(c_{ij}),
$$

the total transportation cost is

$$
J(\gamma)
=
\sum_{i=1}^{3}
\sum_{j=1}^{4}
c_{ij}\gamma_{ij}.
$$

Therefore, the Kantorovich problem is

$$
\min_{\gamma_{ij}\ge0}
\sum_{i=1}^{3}
\sum_{j=1}^{4}
c_{ij}\gamma_{ij}
$$

subject to all supply and demand constraints.

Because both the objective and constraints are linear in $\gamma_{ij}$, this is a linear programming problem.

I solved it using `scipy.optimize.linprog`.

# Optimal Transport Plan

For Euclidean transportation cost, the optimization produced

$$
\gamma^*
=
\begin{pmatrix}
3&0&2&0\\
0&4&0&2\\
0&0&3&1
\end{pmatrix}.
$$

The plan has the following interpretation:

- Bakery 1 sends $3$ units to Café 1 and $2$ units to Café 3.
- Bakery 2 sends $4$ units to Café 2 and $2$ units to Café 4.
- Bakery 3 sends $3$ units to Café 3 and $1$ unit to Café 4.

Thus every bakery splits its supply between two cafés.

# Verification of the Supply Constraints

The row sums of the optimal matrix are

$$
3+0+2+0=5,
$$

$$
0+4+0+2=6,
$$

and

$$
0+0+3+1=4.
$$

Therefore,

$$
\sum_{j=1}^{4}\gamma_{1j}=5=a_1,
$$

$$
\sum_{j=1}^{4}\gamma_{2j}=6=a_2,
$$

and

$$
\sum_{j=1}^{4}\gamma_{3j}=4=a_3.
$$

Thus every bakery ships exactly its entire supply.

# Verification of the Demand Constraints

The column sums are

$$
3+0+0=3,
$$

$$
0+4+0=4,
$$

$$
2+0+3=5,
$$

and

$$
0+2+1=3.
$$

Therefore,

$$
\sum_{i=1}^{3}\gamma_{i1}=3=b_1,
$$

$$
\sum_{i=1}^{3}\gamma_{i2}=4=b_2,
$$

$$
\sum_{i=1}^{3}\gamma_{i3}=5=b_3,
$$

and

$$
\sum_{i=1}^{3}\gamma_{i4}=3=b_4.
$$

Thus every café receives exactly its required demand.

# Euclidean Transportation Cost

The Euclidean cost matrix is

$$
C
=
\begin{pmatrix}
\sqrt{5} & \sqrt{29} & \sqrt{29} & \sqrt{61}\\
\sqrt{29} & \sqrt{5} & \sqrt{41} & 5\\
\sqrt{20} & \sqrt{20} & \sqrt{2} & \sqrt{10}
\end{pmatrix}.
$$

Only six entries of the optimal transport matrix are nonzero.

Therefore, the total cost is

$$
J_E
=
3\sqrt{5}
+
2\sqrt{29}
+
4\sqrt{5}
+
2(5)
+
3\sqrt{2}
+
\sqrt{10}.
$$

Numerically,

$$
J_E
\approx
43.8277.
$$

This agrees with the value returned by the linear-programming solver.

# Squared Euclidean Cost

The squared-distance cost matrix is

$$
C^{(2)}
=
\begin{pmatrix}
5&29&29&61\\
29&5&41&25\\
20&20&2&10
\end{pmatrix}.
$$

For this particular city, minimizing squared Euclidean distance produced the same optimal transport matrix:

$$
\gamma^{*(2)}
=
\begin{pmatrix}
3&0&2&0\\
0&4&0&2\\
0&0&3&1
\end{pmatrix}.
$$

The squared-distance transportation cost is

$$
J_{E^2}
=
3(5)
+
2(29)
+
4(5)
+
2(25)
+
3(2)
+
1(10).
$$

Therefore,

$$
J_{E^2}
=
15+58+20+50+6+10
=
159.
$$

![Optimal Kantorovich plan with squared Euclidean cost](kantorovich_squared_plan.png)

Thus, for this example, Euclidean and squared Euclidean costs lead to the same transport pattern, although their objective values are different.

This agreement is specific to this example and should not be assumed to occur for every transport problem.

# Why Mass Splitting Matters

The optimal matrix contains multiple positive entries in several rows.

For Bakery 1,

$$
\gamma_{11}=3
$$

and

$$
\gamma_{13}=2.
$$

Thus Bakery 1 sends bread to both Café 1 and Café 3.

Similarly,

$$
\gamma_{22}=4,
\qquad
\gamma_{24}=2,
$$

so Bakery 2 sends bread to both Café 2 and Café 4.

Bakery 3 also splits its supply:

$$
\gamma_{33}=3,
\qquad
\gamma_{34}=1.
$$

This splitting is allowed in the Kantorovich formulation.

It is important because there are only

$$
3
$$

bakeries but

$$
4
$$

cafés with positive demand.

The transport plan can still satisfy all four cafés because a single bakery is allowed to serve more than one café.

# Computational Verification

The Python program independently checked:

- total supply equals total demand;
- all $\gamma_{ij}$ are nonnegative;
- every row sum equals the corresponding bakery supply;
- every column sum equals the corresponding café demand;
- the cost returned by the optimizer equals the cost recomputed directly from $\gamma$ and $C$.

All checks passed.

Therefore, the computed matrix is a valid feasible Kantorovich transport plan and the linear-programming solver identified it as an optimal plan.

# Conclusion

The optimal Kantorovich transport plan is

$$
\boxed{
\gamma^*
=
\begin{pmatrix}
3&0&2&0\\
0&4&0&2\\
0&0&3&1
\end{pmatrix}
}.
$$

For Euclidean cost,

$$
\boxed{
J_E\approx43.8277
}
$$

and for squared Euclidean cost,

$$
\boxed{
J_{E^2}=159
}.
$$

Both cost functions produced the same optimal transport matrix in this example.

Most importantly, the optimal solution requires mass splitting.

Each bakery can send bread to more than one café, which allows the $3$ bakeries to satisfy the positive demands of all $4$ cafés.

This flexibility will become the central difference when the same problem is examined using the Monge formulation.
