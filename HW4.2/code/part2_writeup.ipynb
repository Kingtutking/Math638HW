## Numerical Verification

The formula

$$
W_2^2
\left(
\mathcal N(0,\sigma^2),
\mathcal N(0,\sigma_k^2)
\right)
=
(\sigma-\sigma_k)^2
$$

was verified numerically in two independent ways.

### Verification Using the Optimal Transport Map

The optimal transport map from

$$
\mathcal N(0,\sigma_k^2)
$$

to

$$
\mathcal N(0,\sigma^2)
$$

is

$$
T(x)
=
\frac{\sigma}{\sigma_k}x.
$$

Therefore, the Wasserstein cost can be computed from

$$
W_2^2
=
\int_{\mathbb R}
|T(x)-x|^2
\rho_k(x)\,dx.
$$

The numerical integration was performed over a sufficiently large interval containing essentially all of the Gaussian mass.

The computed values agreed with

$$
(\sigma-\sigma_k)^2.
$$

### Verification Using the Quantile Formula

In one dimension,

$$
W_2^2(\mu,\nu)
=
\int_0^1
\left|
F_\mu^{-1}(u)
-
F_\nu^{-1}(u)
\right|^2du.
$$

For centered Gaussians,

$$
F_\sigma^{-1}(u)
=
\sigma\Phi^{-1}(u)
$$

and

$$
F_{\sigma_k}^{-1}(u)
=
\sigma_k\Phi^{-1}(u).
$$

Thus the numerical calculation used

$$
W_2^2
=
\int_0^1
\left[
(\sigma-\sigma_k)\Phi^{-1}(u)
\right]^2du.
$$

The endpoints

$$
u=0
\qquad\text{and}\qquad
u=1
$$

were excluded numerically because

$$
\Phi^{-1}(0)=-\infty
$$

and

$$
\Phi^{-1}(1)=+\infty.
$$

This introduces only a very small truncation error.

As an additional check, the computation verified

$$
\int_0^1
[\Phi^{-1}(u)]^2du
=
1,
$$

which is equivalent to

$$
E[Z^2]=1
$$

for

$$
Z\sim\mathcal N(0,1).
$$

### Wasserstein Distance Plot

![Wasserstein distance between centered Gaussians](hw42_part2_gaussian_wasserstein.png)

For the plot,

$$
\sigma_k=1,
$$

so the theoretical curve is

$$
W_2^2
=
(\sigma-1)^2.
$$

The numerical points lie on the theoretical parabola.

The minimum occurs at

$$
\sigma=1,
$$

where the two distributions are identical and

$$
W_2^2=0.
$$

As $\sigma$ moves away from $\sigma_k$, the squared Wasserstein distance increases quadratically.

Therefore, the numerical experiment confirms

$$
\boxed{
W_2^2
\left(
\mathcal N(0,\sigma^2),
\mathcal N(0,\sigma_k^2)
\right)
=
(\sigma-\sigma_k)^2
}.
$$

Equivalently,

$$
\boxed{
W_2
\left(
\mathcal N(0,\sigma^2),
\mathcal N(0,\sigma_k^2)
\right)
=
|\sigma-\sigma_k|
}.
$$
