## Numerical Verification

The explicit JKO update is

$$
\boxed{
\sigma_{k+1}
=
\frac{
\sigma_k+\sqrt{\sigma_k^2+4\tau}
}{2}
}.
$$

To verify this formula, I compared the explicit solution with an independent numerical minimization of

$$
J(\sigma)
=
-\log\sigma
+
\frac{1}{2\tau}
(\sigma-\sigma_k)^2.
$$

Several combinations of

$$
\sigma_k
$$

and

$$
\tau
$$

were tested.

The largest difference between the explicit formula and the numerical optimizer was

$$
2.39\times10^{-9}.
$$

Thus the independent numerical minimization agrees very closely with the analytic solution.

### First-Order Condition

The minimizer should satisfy

$$
J'(\sigma)
=
-\frac{1}{\sigma}
+
\frac{\sigma-\sigma_k}{\tau}
=
0.
$$

At the explicit solution, the maximum numerical value of

$$
|J'(\sigma_{k+1})|
$$

over all test cases was only

$$
1.45\times10^{-14}.
$$

This verifies the first-order optimality condition.

The equivalent quadratic equation is

$$
\sigma_{k+1}^2
-
\sigma_k\sigma_{k+1}
-
\tau
=
0.
$$

The maximum numerical residual in this equation was

$$
5.27\times10^{-16}.
$$

Therefore, the explicit formula satisfies the quadratic equation to essentially machine precision.

### Convexity

The second derivative of the objective is

$$
J''(\sigma)
=
\frac{1}{\sigma^2}
+
\frac{1}{\tau}.
$$

Since

$$
\sigma>0
$$

and

$$
\tau>0,
$$

we have

$$
J''(\sigma)>0.
$$

In the numerical experiments, the smallest value of the second derivative at the minimizer was

$$
1.09167>0.
$$

Thus the objective is strictly convex and the positive critical point is the unique global minimizer.

### Choice of Root

The two algebraic roots are

$$
\sigma_\pm
=
\frac{
\sigma_k
\pm
\sqrt{\sigma_k^2+4\tau}
}{2}.
$$

For every test case, the positive root was positive and the negative root was negative.

Since a standard deviation must satisfy

$$
\sigma>0,
$$

the negative root is not admissible.

Therefore,

$$
\boxed{
\sigma_{k+1}
=
\frac{
\sigma_k+\sqrt{\sigma_k^2+4\tau}
}{2}
}
$$

is the physically meaningful solution.

### JKO Objective

For the example

$$
\sigma_k=1,
\qquad
\tau=0.1,
$$

the explicit formula gives

$$
\sigma_{k+1}
\approx
1.091608.
$$

![JKO scalar minimization](hw42_part3_jko_minimization.png)

The plot shows that the objective is strictly convex and reaches its unique minimum at approximately

$$
\sigma=1.091608.
$$

The previous value

$$
\sigma_k=1
$$

is shown for comparison.

The minimizer lies to the right of $\sigma_k$, meaning that the Gaussian becomes wider after one JKO step.

### Dependence on the Time Step

![JKO update versus time step](hw42_part3_jko_update_vs_tau.png)

For fixed

$$
\sigma_k=1,
$$

the update

$$
\sigma_{k+1}
=
\frac{
1+\sqrt{1+4\tau}
}{2}
$$

increases as $\tau$ increases.

Thus a larger JKO time step produces a larger change in the Gaussian width.

For every

$$
\tau>0,
$$

we also have

$$
\sqrt{\sigma_k^2+4\tau}
>
\sigma_k,
$$

and therefore

$$
\sigma_{k+1}
>
\sigma_k.
$$

This is consistent with heat diffusion: the Gaussian spreads and its standard deviation increases with time.

### Final Verification

The numerical experiment confirms that:

- the positive quadratic root is admissible;
- the negative root is not physically meaningful;
- the explicit solution satisfies $J'(\sigma)=0$;
- the quadratic equation is satisfied to machine precision;
- $J''(\sigma)>0$, so the minimizer is unique;
- the numerical optimizer agrees with the explicit formula;
- every JKO step increases the Gaussian standard deviation.

Therefore,

$$
\boxed{
\sigma_{k+1}
=
\frac{
\sigma_k+\sqrt{\sigma_k^2+4\tau}
}{2}
}
$$

is verified analytically and numerically.
