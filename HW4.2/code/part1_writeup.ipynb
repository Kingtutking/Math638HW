## Numerical Verification

The Gaussian entropy formula was also checked numerically for several values of

$$
\sigma.
$$

For each value of $\sigma$, I evaluated the Gaussian density

$$
\rho(x)
=
\frac{1}{\sqrt{2\pi}\sigma}
\exp\left(
-\frac{x^2}{2\sigma^2}
\right)
$$

on a large finite interval.

The computation checked four quantities:

- normalization of the density;
- the second moment;
- the full entropy functional;
- the reduced entropy after removing the constant independent of $\sigma$.

The numerical density satisfied

$$
\int_{\mathbb R}\rho(x)\,dx
\approx1,
$$

and the numerical second moment satisfied

$$
\int_{\mathbb R}x^2\rho(x)\,dx
\approx\sigma^2.
$$

The numerical entropy was then compared with

$$
\mathcal E(\rho)
=
-\frac12\log(2\pi e\sigma^2).
$$

![Gaussian entropy verification](hw42_part1_gaussian_entropy.png)

The numerical points lie directly on the theoretical curve, confirming

$$
\boxed{
\mathcal E(\rho)
=
-\frac12\log(2\pi e\sigma^2)
}.
$$

To isolate the part that depends on $\sigma$, I added the constant

$$
\frac12\log(2\pi e).
$$

This gives

$$
\mathcal E(\rho)
+
\frac12\log(2\pi e)
=
-\log\sigma.
$$

![Reduced Gaussian entropy](hw42_part1_reduced_entropy.png)

Again, the numerical values agree with the theoretical curve.

Therefore, up to a constant independent of $\sigma$,

$$
\boxed{
\mathcal E(\rho)
=
-\log\sigma
}.
$$

The plot also shows that the functional decreases as $\sigma$ increases. This is consistent with the heat equation spreading the Gaussian distribution over time.
