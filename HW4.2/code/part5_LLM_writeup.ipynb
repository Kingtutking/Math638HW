## Part I: Entropy of a Gaussian

Let

$$
\rho(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{x^2}{2\sigma^2}
\right),
$$

so that

$$
\rho
=
\mathcal N(0,\sigma^2).
$$

The entropy functional is

$$
\mathcal E(\rho)
=
\int_{\mathbb R}
\rho(x)\log\rho(x)\,dx.
$$

First,

$$
\log\rho(x)
=
-\frac12\log(2\pi\sigma^2)
-
\frac{x^2}{2\sigma^2}.
$$

Therefore,

$$
\mathcal E(\rho)
=
-\frac12\log(2\pi\sigma^2)
\int_{\mathbb R}\rho(x)\,dx
-
\frac{1}{2\sigma^2}
\int_{\mathbb R}x^2\rho(x)\,dx.
$$

Since

$$
\int_{\mathbb R}\rho(x)\,dx=1
$$

and

$$
\int_{\mathbb R}x^2\rho(x)\,dx
=
\sigma^2,
$$

we get

$$
\mathcal E(\rho)
=
-\frac12\log(2\pi\sigma^2)
-
\frac12.
$$

Hence,

$$
\boxed{
\mathcal E(\rho)
=
-\frac12\log(2\pi e\sigma^2)
}.
$$

Since

$$
\log(\sigma^2)
=
2\log\sigma,
$$

this can also be written as

$$
\mathcal E(\rho)
=
-\frac12\log(2\pi e)
-
\log\sigma.
$$

Thus, up to a constant independent of $\sigma$,

$$
\boxed{
\mathcal E(\rho)
=
-\log\sigma.
}
$$

Also,

$$
\frac{d\mathcal E}{d\sigma}
=
-\frac1{\sigma}<0.
$$

Therefore, this entropy functional decreases as the Gaussian spreads.

Note that

$$
\int\rho\log\rho\,dx
$$

is the negative of the usual differential entropy.

It is not necessarily negative for every possible $\sigma$, although for sufficiently large $\sigma$ it is negative. Its important property here is that it decreases as $\sigma$ increases.

## Part II: Wasserstein Distance Between Centered Gaussians

Consider

$$
\mu
=
\mathcal N(0,\sigma^2)
$$

and

$$
\nu
=
\mathcal N(0,\sigma_k^2).
$$

In one dimension,

$$
W_2^2(\mu,\nu)
=
\int_0^1
\left|
F_\mu^{-1}(u)
-
F_\nu^{-1}(u)
\right|^2du.
$$

Let

$$
\Phi
$$

be the standard normal CDF.

For

$$
\mathcal N(0,\sigma^2),
$$

the quantile function is

$$
F_\mu^{-1}(u)
=
\sigma\Phi^{-1}(u).
$$

Similarly,

$$
F_\nu^{-1}(u)
=
\sigma_k\Phi^{-1}(u).
$$

Therefore,

$$
W_2^2(\mu,\nu)
=
\int_0^1
\left[
(\sigma-\sigma_k)
\Phi^{-1}(u)
\right]^2du.
$$

Thus,

$$
W_2^2(\mu,\nu)
=
(\sigma-\sigma_k)^2
\int_0^1
[\Phi^{-1}(u)]^2du.
$$

If

$$
U\sim\operatorname{Uniform}(0,1),
$$

then

$$
\Phi^{-1}(U)
\sim
\mathcal N(0,1).
$$

Hence,

$$
\int_0^1
[\Phi^{-1}(u)]^2du
=
1.
$$

Therefore,

$$
\boxed{
W_2^2
\left(
\mathcal N(0,\sigma^2),
\mathcal N(0,\sigma_k^2)
\right)
=
(\sigma-\sigma_k)^2.
}
$$

Equivalently,

$$
\boxed{
W_2
=
|\sigma-\sigma_k|.
}
$$

An important point is that the formula involves the difference of the standard deviations,

$$
\sigma-\sigma_k,
$$

not the difference of the variances

$$
\sigma^2-\sigma_k^2.
$$

## Part III: The JKO Minimization

The JKO step is

$$
\sigma_{k+1}
=
\operatorname*{arg\,min}_{\sigma>0}
\left\{
-\log\sigma
+
\frac{1}{2\tau}
(\sigma-\sigma_k)^2
\right\}.
$$

Define

$$
J(\sigma)
=
-\log\sigma
+
\frac{1}{2\tau}
(\sigma-\sigma_k)^2.
$$

Differentiate:

$$
J'(\sigma)
=
-\frac1{\sigma}
+
\frac{\sigma-\sigma_k}{\tau}.
$$

At the minimizer,

$$
J'(\sigma)=0.
$$

Thus,

$$
-\frac1{\sigma}
+
\frac{\sigma-\sigma_k}{\tau}
=
0.
$$

Multiplying by

$$
\tau\sigma
$$

gives

$$
\sigma(\sigma-\sigma_k)
=
\tau.
$$

Therefore,

$$
\sigma^2
-
\sigma_k\sigma
-
\tau
=
0.
$$

Using the quadratic formula,

$$
\sigma
=
\frac{
\sigma_k
\pm
\sqrt{\sigma_k^2+4\tau}
}{2}.
$$

Since

$$
\sqrt{\sigma_k^2+4\tau}
>
\sigma_k
$$

for

$$
\tau>0,
$$

the root

$$
\frac{
\sigma_k-
\sqrt{\sigma_k^2+4\tau}
}{2}
$$

is negative.

A standard deviation must satisfy

$$
\sigma>0,
$$

so this root must be rejected.

Therefore,

$$
\boxed{
\sigma_{k+1}
=
\frac{
\sigma_k+
\sqrt{\sigma_k^2+4\tau}
}{2}.
}
$$

To check that this is a minimum,

$$
J''(\sigma)
=
\frac1{\sigma^2}
+
\frac1{\tau}.
$$

Since

$$
\sigma>0
$$

and

$$
\tau>0,
$$

we have

$$
J''(\sigma)>0.
$$

Thus the objective is strictly convex and the positive root is the unique global minimizer.

The optimality condition can also be written as

$$
\boxed{
\sigma_{k+1}-\sigma_k
=
\frac{\tau}{\sigma_{k+1}}.
}
$$

Since the right-hand side is positive,

$$
\sigma_{k+1}>\sigma_k.
$$

Thus every JKO step spreads the Gaussian.

## Part IV: Comparison with the Heat Equation

For the heat equation

$$
\partial_t\rho
=
\partial_{xx}\rho,
$$

starting from

$$
\rho^0
=
\mathcal N(0,\sigma_0^2),
$$

the exact solution is

$$
\rho(t)
=
\mathcal N
\left(
0,\sigma_0^2+2t
\right).
$$

Thus the exact variance is

$$
\boxed{
\sigma_{\mathrm{exact}}^2(t)
=
\sigma_0^2+2t.
}
$$

At the discrete times

$$
t_k=k\tau,
$$

this becomes

$$
\boxed{
\sigma_{\mathrm{exact}}^2(t_k)
=
\sigma_0^2+2k\tau.
}
$$

The JKO scheme instead evolves the standard deviation by

$$
\sigma_{k+1}
=
\frac{
\sigma_k+
\sqrt{\sigma_k^2+4\tau}
}{2}.
$$

Using

$$
\sigma_{k+1}-\sigma_k
=
\frac{\tau}{\sigma_{k+1}},
$$

we can compute the change in variance:

$$
\sigma_{k+1}^2-\sigma_k^2
=
(\sigma_{k+1}-\sigma_k)
(\sigma_{k+1}+\sigma_k).
$$

Therefore,

$$
\sigma_{k+1}^2-\sigma_k^2
=
\frac{\tau}{\sigma_{k+1}}
(\sigma_{k+1}+\sigma_k).
$$

Since

$$
\sigma_k
=
\sigma_{k+1}
-
\frac{\tau}{\sigma_{k+1}},
$$

we get

$$
\sigma_{k+1}+\sigma_k
=
2\sigma_{k+1}
-
\frac{\tau}{\sigma_{k+1}}.
$$

Hence,

$$
\boxed{
\sigma_{k+1}^2-\sigma_k^2
=
2\tau
-
\frac{\tau^2}{\sigma_{k+1}^2}.
}
$$

The exact heat equation increases the variance by

$$
2\tau
$$

during one time interval.

The JKO variance increase is slightly smaller:

$$
\sigma_{k+1}^2-\sigma_k^2
<
2\tau.
$$

Thus, for a finite time step, the JKO scheme slightly underestimates the exact variance.

However, the one-step discrepancy is

$$
\frac{\tau^2}{\sigma_{k+1}^2},
$$

which is of order

$$
O(\tau^2).
$$

Over a fixed time interval, there are approximately

$$
O(1/\tau)
$$

steps, so the total accumulated error is expected to be

$$
O(\tau).
$$

Therefore, as

$$
\tau\to0,
$$

the JKO iteration converges to the exact heat equation evolution.

The optimality condition can also be written as

$$
\frac{
\sigma_{k+1}-\sigma_k
}{\tau}
=
\frac{1}{\sigma_{k+1}}.
$$

This is the backward Euler discretization of

$$
\frac{d\sigma}{dt}
=
\frac1{\sigma}.
$$

Multiplying by

$$
2\sigma
$$

gives

$$
\frac{d}{dt}\sigma^2
=
2.
$$

Integrating,

$$
\sigma^2(t)
=
\sigma_0^2+2t,
$$

which is exactly the variance growth law of the heat equation.

Thus,

$$
\boxed{
\text{as }\tau\to0,
\qquad
\sigma_k^2
\to
\sigma_0^2+2k\tau.
}
$$

Smaller values of

$$
\tau
$$

therefore give better agreement between the JKO iteration and the exact heat equation.

# Final Result

For a centered Gaussian

$$
\rho
=
\mathcal N(0,\sigma^2),
$$

the entropy functional is

$$
\boxed{
\mathcal E(\rho)
=
-\frac12
\log(2\pi e\sigma^2)
}
$$

or, up to a constant,

$$
\boxed{
\mathcal E(\rho)
=
-\log\sigma.
}
$$

The squared Wasserstein distance between centered Gaussians is

$$
\boxed{
W_2^2
\left(
\mathcal N(0,\sigma^2),
\mathcal N(0,\sigma_k^2)
\right)
=
(\sigma-\sigma_k)^2.
}
$$

The JKO minimization gives

$$
\boxed{
\sigma_{k+1}
=
\frac{
\sigma_k+
\sqrt{\sigma_k^2+4\tau}
}{2}.
}
$$

The positive root must be chosen because

$$
\sigma
$$

is a standard deviation and must remain positive.

The exact heat equation variance is

$$
\boxed{
\sigma^2(t)
=
\sigma_0^2+2t.
}
$$

The JKO approximation slightly underestimates this variance for finite $\tau$, but it converges to the exact heat-equation evolution as

$$
\tau\to0.
$$
