## Numerical Comparison with the Heat Equation

The exact heat equation solution starting from

$$
\rho^0
=
\mathcal N(0,\sigma_0^2)
$$

is

$$
\rho(t)
=
\mathcal N
\left(
0,\sigma_0^2+2t
\right).
$$

Therefore, the exact variance is

$$
\boxed{
\sigma_{\mathrm{exact}}^2(t)
=
\sigma_0^2+2t
}.
$$

For the numerical experiment, I chose

$$
\sigma_0=1
$$

and compared the JKO iteration with the exact heat-equation variance up to

$$
T=2.
$$

The JKO update from Part III is

$$
\sigma_{k+1}
=
\frac{
\sigma_k+\sqrt{\sigma_k^2+4\tau}
}{2}.
$$

At the discrete times

$$
t_k=k\tau,
$$

the JKO variance

$$
\sigma_k^2
$$

was compared with

$$
1+2k\tau.
$$

## Representative JKO Trajectory

For

$$
\tau=0.1,
$$

the numerical JKO variance closely follows the exact heat-equation variance.

![JKO variance versus exact heat equation](hw42_part4_jko_vs_heat.png)

The JKO curve lies slightly below the exact solution.

This agrees with the identity

$$
\sigma_{k+1}^2
-
\sigma_k^2
=
2\tau
-
\frac{\tau^2}{\sigma_{k+1}^2}.
$$

Since

$$
\frac{\tau^2}{\sigma_{k+1}^2}>0,
$$

each JKO step increases the variance by slightly less than

$$
2\tau.
$$

Therefore, for finite $\tau$,

$$
\sigma_k^2
<
\sigma_0^2+2k\tau.
$$

This demonstrates that reducing the JKO time step improves agreement with the exact heat equation.

## Convergence Rate

To estimate the convergence rate, I compared errors for successive time steps.

The observed orders were approximately

$$
0.896,\quad
0.948,\quad
0.976,\quad
0.988.
$$

These values approach

$$
1
$$

as

$$
\tau\to0.
$$

![Convergence of JKO variance](hw42_part4_convergence.png)

The numerical errors closely follow the reference line

$$
O(\tau).
$$

Therefore, the JKO variance has approximately first-order global convergence:

$$
\boxed{
\text{error}
=
O(\tau).
}
$$

## Verification of the Variance Increment

The identity

$$
\sigma_{k+1}^2
-
\sigma_k^2
=
2\tau
-
\frac{\tau^2}{\sigma_{k+1}^2}
$$

was also checked directly during every JKO step.

The maximum numerical residual was

$$
2.11\times10^{-15}.
$$

Thus the identity is satisfied to essentially machine precision.

This also explains why the JKO variance systematically underestimates the exact heat variance for finite $\tau$.

## Interpretation as a Time Discretization

The JKO optimality condition can be written as

$$
\frac{
\sigma_{k+1}-\sigma_k
}{\tau}
=
\frac{1}{\sigma_{k+1}}.
$$

This is precisely the backward Euler discretization of

$$
\frac{d\sigma}{dt}
=
\frac1{\sigma}.
$$

Multiplying the continuous equation by

$$
2\sigma
$$

gives

$$
\frac{d}{dt}\sigma^2
=
2.
$$

Therefore,

$$
\sigma^2(t)
=
\sigma_0^2+2t,
$$

which is exactly the variance law for the heat equation.

The first-order convergence observed numerically is therefore consistent with the fact that backward Euler is a first-order time discretization.

## Conclusion

The exact heat equation variance is

$$
\boxed{
\sigma_{\mathrm{exact}}^2(t)
=
\sigma_0^2+2t
}.
$$

The JKO scheme produces a discrete approximation satisfying

$$
\sigma_{k+1}
=
\frac{
\sigma_k+\sqrt{\sigma_k^2+4\tau}
}{2}.
$$

For finite $\tau$, the JKO variance grows slightly more slowly than the exact variance.

However, as

$$
\tau\to0,
$$

the JKO solution converges to the exact heat-equation solution.

The numerical convergence orders approach

$$
1,
$$

so the global error behaves approximately as

$$
\boxed{
O(\tau).
}
$$

Therefore,

$$
\boxed{
\text{smaller }\tau
\Longrightarrow
\text{better agreement with the exact heat equation}.
}
$$
