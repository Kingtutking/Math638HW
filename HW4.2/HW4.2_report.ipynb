# Homework 4.2: An Explicit JKO Step for the Heat Equation

# Introduction

In this homework, I study the Jordan-Kinderlehrer-Otto (JKO) scheme for the heat equation.

Consider the entropy functional

$$
\mathcal E(\rho)
=
\int_{\mathbb R}
\rho(x)\log\rho(x)\,dx.
$$

Its Wasserstein gradient flow is the heat equation

$$
\partial_t\rho
=
\partial_{xx}\rho.
$$

Assume that the initial density is a centered Gaussian

$$
\rho^0
=
\mathcal N(0,\sigma_0^2).
$$

Since the heat equation preserves Gaussianity, I restrict the JKO scheme to centered Gaussian densities

$$
\rho
=
\mathcal N(0,\sigma^2).
$$

This reduces the infinite-dimensional Wasserstein minimization problem to a scalar optimization problem involving the standard deviation $\sigma$.

The main goals of this homework are to:

- compute the entropy of a centered Gaussian;
- compute the Wasserstein distance between centered Gaussians;
- derive an explicit JKO update for $\sigma$;
- compare the discrete JKO evolution with the exact heat equation;
- investigate convergence as the time step $\tau$ decreases;
- evaluate the mathematical reliability of an LLM-generated solution.

# Part I: Entropy of a Gaussian

Let

$$
\rho
=
\mathcal N(0,\sigma^2).
$$

Its density is

$$
\rho(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-\frac{x^2}{2\sigma^2}
\right).
$$

The entropy functional used in the JKO scheme is

$$
\mathcal E(\rho)
=
\int_{\mathbb R}
\rho(x)\log\rho(x)\,dx.
$$

## Derivation

Taking the logarithm of the Gaussian density gives

$$
\log\rho(x)
=
-\frac12\log(2\pi\sigma^2)
-
\frac{x^2}{2\sigma^2}.
$$

Therefore,

$$
\mathcal E(\rho)
=
-\frac12\log(2\pi\sigma^2)
\int_{\mathbb R}\rho(x)\,dx
-
\frac{1}{2\sigma^2}
\int_{\mathbb R}x^2\rho(x)\,dx.
$$

Since $\rho$ is a probability density,

$$
\int_{\mathbb R}\rho(x)\,dx
=
1.
$$

Also,

$$
\int_{\mathbb R}x^2\rho(x)\,dx
=
\sigma^2.
$$

Thus,

$$
\mathcal E(\rho)
=
-\frac12\log(2\pi\sigma^2)
-
\frac12.
$$

Since

$$
\frac12
=
\frac12\log e,
$$

we obtain

$$
\boxed{
\mathcal E(\rho)
=
-\frac12\log(2\pi e\sigma^2)
}.
$$

Expanding the logarithm gives

$$
\mathcal E(\rho)
=
-\frac12\log(2\pi e)
-
\log\sigma.
$$

Therefore, up to constants independent of $\sigma$,

$$
\boxed{
\mathcal E(\rho)
=
-\log\sigma.
}
$$

## Numerical Verification

The entropy formula was verified numerically for several values of $\sigma$.

For each Gaussian, I checked:

- normalization;
- the second moment;
- the full entropy functional;
- the reduced entropy after removing the constant independent of $\sigma$.

The numerical entropy values lie on the theoretical curve

$$
-\frac12\log(2\pi e\sigma^2).
$$

![Gaussian entropy verification](hw42_part1_gaussian_entropy.png)

After removing the constant

$$
-\frac12\log(2\pi e),
$$

the remaining quantity agrees with

$$
-\log\sigma.
$$

![Reduced Gaussian entropy](hw42_part1_reduced_entropy.png)

Since

$$
\frac{d\mathcal E}{d\sigma}
=
-\frac1{\sigma}
<
0,
$$

the functional decreases as the Gaussian spreads.

The functional

$$
\int\rho\log\rho\,dx
$$

is the negative of the usual differential entropy.

# Part II: Wasserstein Distance Between Centered Gaussians

Consider

$$
\mu
=
\mathcal N(0,\sigma^2)
$$

and

$$
\nu
=
\mathcal N(0,\sigma_k^2).
$$

The goal is to compute

$$
W_2^2(\mu,\nu).
$$

## Quantile Derivation

In one dimension,

$$
W_2^2(\mu,\nu)
=
\int_0^1
\left|
F_\mu^{-1}(u)
-
F_\nu^{-1}(u)
\right|^2du.
$$

Let $\Phi$ denote the standard Gaussian CDF.

For a centered Gaussian,

$$
F_\mu^{-1}(u)
=
\sigma\Phi^{-1}(u)
$$

and

$$
F_\nu^{-1}(u)
=
\sigma_k\Phi^{-1}(u).
$$

Therefore,

$$
W_2^2(\mu,\nu)
=
(\sigma-\sigma_k)^2
\int_0^1
[\Phi^{-1}(u)]^2du.
$$

If

$$
U\sim\operatorname{Uniform}(0,1),
$$

then

$$
\Phi^{-1}(U)
\sim
\mathcal N(0,1).
$$

Hence,

$$
\int_0^1
[\Phi^{-1}(u)]^2du
=
1.
$$

Thus,

$$
\boxed{
W_2^2
\left(
\mathcal N(0,\sigma^2),
\mathcal N(0,\sigma_k^2)
\right)
=
(\sigma-\sigma_k)^2.
}
$$

Equivalently,

$$
\boxed{
W_2
=
|\sigma-\sigma_k|.
}
$$

The formula depends on the difference between the standard deviations, not the difference between the variances.

## Optimal Transport Map

The monotone optimal map from

$$
\mathcal N(0,\sigma_k^2)
$$

to

$$
\mathcal N(0,\sigma^2)
$$

is

$$
T(x)
=
\frac{\sigma}{\sigma_k}x.
$$

Then

$$
W_2^2
=
\int_{\mathbb R}
|T(x)-x|^2\rho_k(x)\,dx.
$$

Since

$$
T(x)-x
=
\left(
\frac{\sigma}{\sigma_k}-1
\right)x,
$$

we get

$$
W_2^2
=
\left(
\frac{\sigma-\sigma_k}{\sigma_k}
\right)^2
\int_{\mathbb R}
x^2\rho_k(x)\,dx.
$$

Using

$$
\int_{\mathbb R}
x^2\rho_k(x)\,dx
=
\sigma_k^2,
$$

we again obtain

$$
\boxed{
W_2^2
=
(\sigma-\sigma_k)^2.
}
$$

## Numerical Verification

The formula was checked numerically using both:

- the optimal transport map;
- the quantile representation.

For fixed

$$
\sigma_k=1,
$$

the theoretical curve is

$$
W_2^2
=
(\sigma-1)^2.
$$

![Wasserstein distance between centered Gaussians](hw42_part2_gaussian_wasserstein.png)

The numerical values lie directly on the theoretical parabola.

Thus the numerical computation confirms

$$
\boxed{
W_2^2
=
(\sigma-\sigma_k)^2.
}
$$

# Part III: The JKO Minimization

From Parts I and II, the JKO step becomes

$$
\sigma_{k+1}
=
\operatorname*{arg\,min}_{\sigma>0}
\left\{
-\log\sigma
+
\frac{1}{2\tau}
(\sigma-\sigma_k)^2
\right\}.
$$

Define

$$
J(\sigma)
=
-\log\sigma
+
\frac{1}{2\tau}
(\sigma-\sigma_k)^2.
$$

## First-Order Condition

Differentiate:

$$
J'(\sigma)
=
-\frac1{\sigma}
+
\frac{\sigma-\sigma_k}{\tau}.
$$

At the minimizer,

$$
J'(\sigma)=0.
$$

Therefore,

$$
-\frac1{\sigma}
+
\frac{\sigma-\sigma_k}{\tau}
=
0.
$$

Multiplying by $\tau\sigma$ gives

$$
\sigma(\sigma-\sigma_k)
=
\tau.
$$

Thus,

$$
\sigma^2
-
\sigma_k\sigma
-
\tau
=
0.
$$

Using the quadratic formula,

$$
\sigma
=
\frac{
\sigma_k
\pm
\sqrt{\sigma_k^2+4\tau}
}{2}.
$$

Because

$$
\sqrt{\sigma_k^2+4\tau}
>
\sigma_k,
$$

the negative root satisfies

$$
\frac{
\sigma_k-\sqrt{\sigma_k^2+4\tau}
}{2}
<0.
$$

Since $\sigma$ is a standard deviation,

$$
\sigma>0.
$$

Therefore, the negative root must be discarded.

The JKO update is

$$
\boxed{
\sigma_{k+1}
=
\frac{
\sigma_k+\sqrt{\sigma_k^2+4\tau}
}{2}.
}
$$

## Convexity

The second derivative is

$$
J''(\sigma)
=
\frac1{\sigma^2}
+
\frac1{\tau}.
$$

For

$$
\sigma>0,
\qquad
\tau>0,
$$

we have

$$
J''(\sigma)>0.
$$

Therefore, the objective is strictly convex, and the positive critical point is the unique global minimizer.

## Numerical Verification

The explicit formula was compared with an independent numerical optimization for several values of $\sigma_k$ and $\tau$.

The maximum difference between the explicit formula and the numerical optimizer was

$$
2.39\times10^{-9}.
$$

The largest value of

$$
|J'(\sigma_{k+1})|
$$

was

$$
1.45\times10^{-14}.
$$

The maximum residual in

$$
\sigma_{k+1}^2
-
\sigma_k\sigma_{k+1}
-
\tau
=
0
$$

was

$$
5.27\times10^{-16}.
$$

The smallest computed value of

$$
J''(\sigma_{k+1})
$$

was

$$
1.09167>0.
$$

Thus the analytic minimizer is strongly verified.

For the example

$$
\sigma_k=1,
\qquad
\tau=0.1,
$$

the minimizer is

$$
\sigma_{k+1}
\approx
1.091608.
$$

![JKO scalar minimization](hw42_part3_jko_minimization.png)

The minimum lies to the right of

$$
\sigma_k=1,
$$

showing that the Gaussian spreads after one JKO step.

For fixed $\sigma_k=1$, the update increases as $\tau$ increases.

![JKO update versus time step](hw42_part3_jko_update_vs_tau.png)

The optimality condition can also be written as

$$
\boxed{
\sigma_{k+1}-\sigma_k
=
\frac{\tau}{\sigma_{k+1}}.
}
$$

Thus every JKO step satisfies

$$
\sigma_{k+1}>\sigma_k.
$$

# Part IV: Compare with the Heat Equation

The exact heat equation solution starting from

$$
\rho^0
=
\mathcal N(0,\sigma_0^2)
$$

is

$$
\rho(t)
=
\mathcal N
\left(
0,\sigma_0^2+2t
\right).
$$

Therefore,

$$
\boxed{
\sigma_{\mathrm{exact}}^2(t)
=
\sigma_0^2+2t.
}
$$

At the JKO time levels

$$
t_k=k\tau,
$$

the exact variance is

$$
\boxed{
\sigma_0^2+2k\tau.
}
$$

## Variance Increment of the JKO Scheme

From Part III,

$$
\sigma_{k+1}-\sigma_k
=
\frac{\tau}{\sigma_{k+1}}.
$$

Then

$$
\sigma_{k+1}^2-\sigma_k^2
=
(\sigma_{k+1}-\sigma_k)
(\sigma_{k+1}+\sigma_k).
$$

Therefore,

$$
\sigma_{k+1}^2-\sigma_k^2
=
\frac{\tau}{\sigma_{k+1}}
(\sigma_{k+1}+\sigma_k).
$$

Since

$$
\sigma_k
=
\sigma_{k+1}
-
\frac{\tau}{\sigma_{k+1}},
$$

we obtain

$$
\boxed{
\sigma_{k+1}^2-\sigma_k^2
=
2\tau
-
\frac{\tau^2}{\sigma_{k+1}^2}.
}
$$

The exact heat equation increases the variance by

$$
2\tau.
$$

Therefore,

$$
\sigma_{k+1}^2-\sigma_k^2
<
2\tau.
$$

This shows that the finite-step JKO scheme slightly underestimates the exact variance growth.

## Numerical Comparison

For the numerical experiment, I chose

$$
\sigma_0=1
$$

and

$$
T=2.
$$

For

$$
\tau=0.1,
$$

the JKO variance closely follows the exact heat equation.

![JKO variance versus exact heat equation](hw42_part4_jko_vs_heat.png)

The JKO curve remains slightly below the exact variance, as predicted analytically.

The experiment was repeated for

$$
\tau
=
0.5,\ 0.25,\ 0.1,\ 0.05,\ 0.025.
$$

The final-time results were:

The error decreases consistently as $\tau$ becomes smaller.

$$
2.11\times10^{-15}.
$$

## Convergence Rate

The observed convergence orders were

$$
0.896,
\qquad
0.948,
\qquad
0.976,
\qquad
0.988.
$$

These approach

$$
1
$$

as

$$
\tau\to0.
$$

![JKO convergence](hw42_part4_convergence.png)

Therefore, the numerical experiment indicates approximately first-order convergence:

$$
\boxed{
\text{error}
=
O(\tau).
}
$$

This is also consistent with the optimality condition

$$
\frac{
\sigma_{k+1}-\sigma_k
}{\tau}
=
\frac{1}{\sigma_{k+1}},
$$

which is the backward Euler discretization of

$$
\frac{d\sigma}{dt}
=
\frac1{\sigma}.
$$

Multiplying by $2\sigma$ gives

$$
\frac{d}{dt}\sigma^2
=
2,
$$

so

$$
\sigma^2(t)
=
\sigma_0^2+2t.
$$

Thus the JKO iteration converges to the exact Gaussian heat flow as

$$
\tau\to0.
$$

# Conclusion

For a centered Gaussian

$$
\rho
=
\mathcal N(0,\sigma^2),
$$

the entropy functional is

$$
\boxed{
\mathcal E(\rho)
=
-\frac12\log(2\pi e\sigma^2)
}.
$$

Up to constants independent of $\sigma$,

$$
\boxed{
\mathcal E(\rho)
=
-\log\sigma.
}
$$

The squared Wasserstein distance between two centered Gaussians is

$$
\boxed{
W_2^2
\left(
\mathcal N(0,\sigma^2),
\mathcal N(0,\sigma_k^2)
\right)
=
(\sigma-\sigma_k)^2.
}
$$

The JKO minimization therefore reduces to

$$
\operatorname*{arg\,min}_{\sigma>0}
\left\{
-\log\sigma
+
\frac{1}{2\tau}
(\sigma-\sigma_k)^2
\right\},
$$

whose unique minimizer is

$$
\boxed{
\sigma_{k+1}
=
\frac{
\sigma_k+\sqrt{\sigma_k^2+4\tau}
}{2}.
}
$$

The exact heat equation satisfies

$$
\boxed{
\sigma^2(t)
=
\sigma_0^2+2t.
}
$$

For finite $\tau$, the JKO variance grows slightly more slowly than the exact heat-equation variance.

However, the numerical experiments show that the error decreases approximately linearly with $\tau$.

The observed convergence orders approach

$$
1,
$$

confirming approximately first-order convergence.

Therefore, the explicit Gaussian JKO scheme provides a consistent discrete approximation of the Wasserstein gradient flow of entropy, and in the limit

$$
\tau\to0,
$$

it reproduces the exact Gaussian solution of the heat equation.
